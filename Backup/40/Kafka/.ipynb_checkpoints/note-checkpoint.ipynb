```python
Broker
******

A broker in Apache Kafka is a Kafka server

✅ Responsibilities:
Receives messages from producers.

Stores messages in topics (organized into partitions).

Serves messages to consumers when requested.

Manages replication (if it's the leader of a partition).

Handles offsets (tracking message position for consumers, if configured).
                     
                     
A Kafka cluster is made up of multiple brokers.
```

```python
Topic	A category or feed name to which records are published. 

Topics are split into partitions.
```

```python
Partition	A topic is divided into partitions for parallel processing.

Each partition is an ordered, immutable sequence of records.
```

```python
Offset	A unique ID assigned to each message in a partition. 
```

```python
How kafka works?

1. Apache kafka is a distributed streaming platform
2. At its core, It allows system that generate data (called producers) to persist
their data in real tiem in an apache kafka topic.
3. Any topic can then be read by any number of systems who need that data in real-time(called Consumers)
4. Kafka is a pub/sub system.

```

```python
Kafka Advantages :-

1. High throughput
2. Scalability
3. No data loss once confugured properly
4. Stream Processing
5. Durability
6. Replication
```

```python
Kafka's architecture consists of 
            Producers (publish data), 
            Consumers (subscribe to topics), 
            Brokers (manage data storage and retrieval), 
            Topics (categories for data), 
            Partitions (sub-divisions of topics for parallelism), 
and ZooKeeper (coordinates and manages the Kafka cluster).
```

```python
How does Kafka ensure fault tolerance? with example

 1. ✅ Replication
Kafka replicates each partition of a topic across multiple brokers. 
This ensures that if one broker fails, another has a full copy of the data.


2. 🔁 Leader Election
Each partition has a leader broker, and others act as followers.
Only the leader handles all reads/writes for the partition. 
If the leader fails, Kafka automatically elects a new one from the in-sync replicas (ISR).

3.  Acknowledgments

3.  Consumer Offsets in Kafka

6.  🧹 Log Compaction / Retention
```
