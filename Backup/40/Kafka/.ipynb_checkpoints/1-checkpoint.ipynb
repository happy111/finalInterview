```python
kafka : is a 
   . Distributed
   . Fault-tolerant event streaming platform
   . It follow publish and subscribe and distributed log models
```

```python
Core components:-

  1. Producer
     . Send messages to kafka topics
  2. Consumer
     . subscribes to topic
     . A client that reads data from Kafka topics.
  3. Topics:-
     . Logical category 
     . Topics are split into partitions.
        
    Why Kafka Topics Are Split into Partitions
    Ans : Partitioning is a core design feature in 
            Kafka that enables 
            scalability, 
            parallelism, 
            and fault tolerance.
       1. Scalability (Horizontal Scaling)
       2. Parallelism (Faster Processing)
        
  4. Partitions
    . Kafka topics are divided into partitions for parallelism
    . Each partition is an ordered, immutable sequence of messages.
    . Each message in a partition has an offset (a unique ID).


                https://chatgpt.com/c/6854e2c0-e1fc-8010-bfef-626a2bc60560
```

```python
5. Broker
    A Kafka server that stores messages.
    Kafka cluster typically has multiple brokers.
    Each broker handles data for one or more topic partitions.
6. Cluster
    A group of Kafka brokers.
```

```python
7. ZooKeeper (older versions) / KRaft (newer versions)
   ZooKeeper manages cluster metadata, leader election, etc.
   KRaft (Kafka Raft) replaces ZooKeeper in newer Kafka versions.
```

```python
✅ Key Features
High throughput for publishing and subscribing.

Scalable horizontally by adding brokers and partitions.

Fault-tolerant: replication of partitions across brokers.

Durable: uses disk storage and log-based architecture.
```

```python
Producer: Sends messages to Kafka topics.

Consumer: Subscribes and reads messages from topics.

Broker: Kafka server that stores data and serves clients.

Topic: Logical channel to which messages are sent.

Partition: Kafka splits topics into partitions for scalability.

Zookeeper: Manages cluster metadata (optional in newer versions).

Kafka Connect: For integrating Kafka with external systems.

Kafka Streams: For stream processing using Kafka topics.
```

```python
An offset is a unique ID for a message within a partition.
```

```python
Kafka consumers use it to track which messages have been read.
```
