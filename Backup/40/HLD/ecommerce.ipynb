```python
1️⃣ Requirements:-

Functional Requirements

    User Registration & Login
    Browse Products
    Product Search & Filters
    Add to Cart
    Place Order
    Payments (UPI, Card, Wallet)
    Order Tracking
    Notifications (SMS/Email)
    Admin: Add/Update products
    Reviews & Ratings
```

```python
Non-functional Requirements

    High availability
    Low latency
    Eventual consistency
    Scalable (millions of users)
    Secure (PCI-DSS compliance)
    Fault-tolerant
    Observability
```
