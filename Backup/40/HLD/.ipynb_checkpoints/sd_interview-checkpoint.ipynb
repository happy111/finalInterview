```python
3️⃣ DATABASE DESIGN (Production-Ready)

USERS:-
users (
  id UUID PRIMARY KEY,
  name VARCHAR,
  phone VARCHAR UNIQUE,
  created_at TIMESTAMP
)

DRIVERS:-

drivers (
  id UUID PRIMARY KEY,
  name VARCHAR,
  vehicle_type ENUM('BIKE','CAR'),
  is_available BOOLEAN,
  latitude DOUBLE,
  longitude DOUBLE
)

RIDES:-

rides (
  id UUID PRIMARY KEY,
  user_id UUID,
  driver_id UUID,
  status ENUM('REQUESTED','ONGOING','COMPLETED'),
  fare DECIMAL,
  created_at TIMESTAMP
)

Indexes

CREATE INDEX idx_driver_location ON drivers(latitude, longitude);
CREATE INDEX idx_ride_status ON rides(status);
```

```python
| Failure             | Solution         |
| ------------------- | ---------------- |
| Payment fails       | Retry + rollback |
| Notification fails  | DLQ              |
| Driver service down | Circuit breaker  |
| Kafka consumer down | Replay events    |
```

```python
2️⃣9️⃣ How do you handle background tasks in FastAPI?

Answer:
Using BackgroundTasks for light jobs and Celery or Kafka consumers for heavy tasks.
```

```python
2️⃣8️⃣ What is async vs sync in Python?

Answer:
Sync blocks execution; async allows handling multiple requests concurrently using non-blocking I/O
```

```python
2️⃣7️⃣ How do you design a highly available system?

Answer:
Using load balancers, replicas, multi-AZ deployment, stateless services, and health checks.
```

```python
2️⃣6️⃣ What is CAP theorem?

Answer:
In distributed systems, you can only guarantee two of Consistency, Availability, and Partition tolerance.
```

```python
2️⃣5️⃣ How does Kubernetes help scaling?

Answer:
Kubernetes provides auto-scaling, self-healing, load balancing, and rolling deployments.
```

```python
2️⃣2️⃣ What is observability?

Answer:
The ability to understand system behavior using logs, metrics, and traces.
```

```python
2️⃣1️⃣ How do you secure APIs?

Answer:
JWT/OAuth2 authentication, HTTPS, input validation, rate limiting, and secrets management.
```

```python
2️⃣0️⃣ How do you implement rate limiting?

Answer:
Using token bucket or sliding window algorithms with Redis or API Gateway.
```

```python
1️⃣9️⃣ What is circuit breaker pattern?

Answer:
It stops sending requests to a failing service temporarily to prevent cascading failures.
```

```python
1️⃣8️⃣ How do you handle failures in microservices?

Answer:
Using retries, circuit breakers, timeouts, fallbacks, and dead-letter queues.
```

```python
1️⃣7️⃣ Kafka vs SQS?

Answer:
Kafka is used for high-throughput streaming. SQS is simpler and managed, suitable for background tasks.
```

```python
✅ Kafka Failure: Causes, Identification & Recovery

    Failure :-
      🔴 Broker-level failures
          Disk full
          JVM out of memory (OOM)
          High GC pauses
        Impact :-
             Partition leader goes down
             Producers/consumers fail
    
     🔴 Zookeeper / KRaft issues

            Zookeeper unavailable
            Session timeout
            Leader election stuck
            👉 Impact:
        
        No leader for partitions
        Cluster becomes unstable
     🔴 Disk / Storage issues
     🔴 Producer-side failures
```
