```python
Microservice Architecture:-

. Microservice Architecture is a software design approach.
. All the services are independent.
. 
🔹 Key Characteristics

Independent Services – Each service can be developed, deployed, and scaled independently

Single Responsibility – One service = one business function

Decentralized Data – Each service owns its own database

Loose Coupling – Services communicate via APIs or events

Technology Agnostic – Different services can use different languages or frameworks
```

```python
E-commerce Application

    User Service – user registration, login
    
    Product Service – product catalog
    
    Order Service – order management
    
    Payment Service – payments
    
    Notification Service – emails, SMS

Each service runs independently and talks via REST APIs or message queues (Kafka, RabbitMQ).
```

```python
🔹 Communication Types

Synchronous
    REST (HTTP)
    gRPC
    👉 Used when immediate response is required

Asynchronous
    Kafka, SQS, RabbitMQ
    👉 Used for high scalability & loose coupling
```

```python
🔹 Advantages

✅ Scalability (scale only required services)
✅ Faster development & deployment
✅ Fault isolation (failure of one service doesn’t crash entire system)
✅ Technology flexibility
✅ Better team ownership
```

```python
❌ Complex architecture
❌ Network latency
❌ Data consistency challenges
❌ Distributed debugging is hard
❌ Requires DevOps maturity
```

```python
🔹 Common Design Patterns

API Gateway
Service Discovery (Eureka, Consul)
Circuit Breaker (Resilience4j)
Saga Pattern (distributed transactions)
CQRS
Bulkhead
Retry & Timeout
```

```python
🔹 Database Strategy

Database per Service (recommended)

No shared database

Use eventual consistency
```

```python
Security

OAuth2 / JWT

mTLS (service-to-service)

API Gateway security

RBAC
```
