```python
#  A microservice can fail due to:

    High latency
    Container crash
    Pod restart
    Network timeout
    Dependency failure
    Circuit breaker open
    Health check failure

To detect these issues automatically, microservices use observability + resilience mechanisms.
```

```python
✅ 1. Health Checks (Liveness & Readiness Probes)

Used in Kubernetes, Docker, load balancers.

Liveness Probe

Checks if the service is alive.
If it fails → container is restarted.

Readiness Probe

Checks if the service is ready to receive requests.
If it fails → service is removed from load balancer routing.

/health
/live
/ready
```

```python
✅ 2. Heartbeat / Ping Mechanism

Service A periodically “pings” or calls Service B.

If Service B does not respond within timeout → considered unreachable/failing.

Used by:

Service discovery systems

API gateways

Monitoring agents
```

```python
✅ 3. Circuit Breaker (Resilience Pattern)

Tools: Resilience4j, Hystrix, Envoy, Istio

How it detects failure:

Checks for repeated timeouts

Counts failures

If failures cross threshold → circuit opens

Signals that the service is failing or unstable
```
