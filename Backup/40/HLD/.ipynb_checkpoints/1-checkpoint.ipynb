```python
✅ How to Scale a High-Traffic Microservice in Kubernetes?

Ans :- 1️⃣ Horizontal Pod Autoscaling (HPA)

HPA automatically increases or decreases the number of pods based on metrics.

You can scale based on:

    CPU usage
    Memory usage
    Custom metrics (QPS, latency, queue length, Kafka lag, etc.)
    External metrics (Prometheus, SQS queue size, RabbitMQ backlog)

apiVersion: autoscaling/v2
kind: HorizontalPodAutoscaler
metadata:
  name: payment-hpa
spec:
  scaleTargetRef:
    apiVersion: apps/v1
    kind: Deployment
    name: payment-service
  minReplicas: 3
  maxReplicas: 20
  metrics:
  - type: Resource
    resource:
      name: cpu
      target:
        type: Utilization
        averageUtilization: 70

HPA automatically adds more pods if load increases.
```

```python
2️⃣ Cluster Autoscaler — Add new worker nodes automatically

If HPA increases pods, but your cluster has no space, Kubernetes will not schedule new pods.

Cluster Autoscaler solves this by adding nodes automatically.

📌 Works on:

AWS: EKS Node Group Autoscaling
```

```python
3️⃣ Load Balancing Using Kubernetes Service

Kubernetes Service distributes traffic across all pods automatically.

kind: Service
apiVersion: v1
metadata:
  name: payment-service
spec:
  selector:
    app: payment
  ports:
  - port: 80
    targetPort: 8080
  type: ClusterIP

💡 If pods increase, traffic distribution automatically updates.
```

```python
1️⃣ What is High-Level Design (HLD)?

High-Level Design (HLD) describes the overall system architecture, focusing on:

    System components
    How components communicate
    Databases & storage choices
    APIs and integration points
    Scaling strategies
    Load balancing & failover
    Security and observability
```

```python
2️⃣ What are key components included in an HLD document?

A good HLD typically includes:

    Architecture diagram
    Components description
    Data flow diagram
    API layer design
    Database selection and schema overview
    Scaling strategy (HPA, Load balancers, sharding)
    Caching layer
    Message queues/event-driven architecture
    Security design
    Observability (logs, metrics, tracing)
    High availability & disaster recovery
```

```python
| Feature      | HLD                     | LLD                          |
| ------------ | ----------------------- | ---------------------------- |
| Focus        | Entire system           | Class-level / function-level |
| Detail level | Abstract                | Detailed                     |
| Diagram      | Architecture            | UML Class diagrams           |
| Audience     | Architects, senior devs | Developers                   |
| Output       | Tech stack, modules     | Classes, methods, DB tables  |
```

```python
4️⃣ How do you design a scalable system?


To design a scalable system:

    Horizontal Scaling first (add more servers/pods)
    Stateless services (easy scaling)
    Use Load Balancers (Round Robin, Least Connections)
    Distributed caching (Redis, Memcached)

    Database scaling
        Read replicas
        Sharding
        Partitioning

    Async processing using message queues (Kafka, SQS)
    Autoscaling
        HPA (pods)
        Cluster Autoscaler (nodes)
    CDN for static content
```

```python
5️⃣ How do you ensure High Availability in your system?

    ✔ Replicate services across AZ/Regions
    ✔ Load balancer for failover
    ✔ Database replication (Master–Replica)
    ✔ Health checks + auto-restart
    ✔ Zero-downtime deployments (Blue/green, Canary)
    ✔ Distributed caching
    ✔ Kubernetes + multiple nodes
```

```python
6️⃣ What is the role of caching in system design?

Caching reduces latency and database load.

Types:

Application-level cache (in-memory)
Distributed cache (Redis, Memcached)
CDN cache for static content
Database cache (query caching)

Helps with:
    
    Faster response times
    Less database pressure
    Lower cost
```

```python
7️⃣ How to design an API rate limiter?

Approaches:

Token Bucket Algorithm (most common)
Leaky Bucket Algorithm
Fixed Window / Sliding Window Counter

Use tools like:
    
    Redis atomic counters
    NGINX rate limiting
    API Gateway throttling
```

```python
8️⃣ How do you design a fault-tolerant system?
Answer:

    Replicate services across regions
    Use retry policies with backoff
    Use circuit breaker pattern
    Store state externally
    Graceful degradation
    Distributed message queues
    Health checks & self-healing
```

```python
9️⃣ How do microservices communicate with each other?
Answer:

✔ Synchronous Communication

    REST APIs
    gRPC
    GraphQL

✔ Asynchronous Communication

Kafka / RabbitMQ
AWS SQS
Event-driven architecture
```

```python
🔟 How do you design an event-driven system?
Answer:

Components:

    Producers
    Consumers
    Message broker (Kafka, SQS, RabbitMQ)
    Event schema
    Offset management
    DLQ (Dead Letter Queue)

Benefits:

Loose coupling
High scalability
Better performance
```

```python
1️⃣4️⃣ What is the CAP Theorem?
Answer:

A distributed system can provide only two of the three:

| C | Consistency |
| A | Availability |
| P | Partition Tolerance |

Example:

MongoDB → AP

Cassandra → AP

SQL DB → CP
```

```python
1️⃣7️⃣ What is a Circuit Breaker pattern?
Answer:

Prevents a failing service from being called repeatedly.

States:

Closed – calls allowed

Open – calls blocked

Half-open – test calls allowed

Used in:

Netflix Hystrix

Istio

Resilience4j
```

```python
1️⃣8️⃣ What is an API Gateway and why use it?
Answer:

API Gateway acts as a single entry point for all microservices.

Benefits:
    Authentication
    Rate limiting
    Routing
    Logging
    Versioning
    Caching
    SSL termination

Tools:
        Kong
        AWS API Gateway
        NGINX
        Istio
```
