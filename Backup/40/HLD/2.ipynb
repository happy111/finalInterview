```python
# Circuit Breaker

A circuit breaker prevents cascading failures by blocking calls to an unhealthy service 
and restoring them only when the service becomes healthy again.

A circuit breaker stops requests to a failing service to protect the system and
    restore only when the service recovers
    
Python    PyCircuitBreaker, Tenacity

🎯 When to Use Circuit Breakers?

Use when calling:

Payment gateways
Authorization service
Third-party APIs
Inventory service
Email/sms providers
Database or caching microservices
```

```python
✅ 3. Resilience & Reliability Terms
19. Circuit Breaker

Stops calling a failing service to prevent system failures.

20. Retry

Automatically retrying failed requests.

21. Timeout

Stop waiting for response after a time limit.

22. Bulkhead

Isolate resources so one failure doesn’t sink the whole system.

23. Rate Limiting

Restrict number of requests per second.

24. Throttling

Slowing or blocking excessive traffic.

25. Fail Fast

Immediately return error instead of waiting.

26. Fallback

Backup response when service fails.

27. Chaos Engineering

Intentionally breaking services to test resilience (e.g., Chaos Monkey).
```
