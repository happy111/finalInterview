```python
import multiprocessing
import time

def print_number():
    for i in range(4):
        time.sleep(2)
        print(f"Number : {i}")

def print_letter():
    for letter in "asfds":
        time.sleep(2)
        print(f"Letter : {letter}")

if __name__=="__main__":
    p1 = multiprocessing.Process(target=print_number)
    
    p2 = multiprocessing.Process(target=print_letter)
    # Start thread
    p1.start()
    p2.start()
    # Wait for the thread to complete
    p1.join()
    p2.join()
```

```python
🔹 16. How Lambda handles retries?

| Trigger Type       | Retry Behavior                         |
| ------------------ | -------------------------------------- |
| Sync (API Gateway) | No automatic retry                     |
| Async (S3, SNS)    | 2 retries (1min, 5min)                 |
| SQS                | Infinite retries until message expires |
| DLQ or SNS         | Stores failed events                   |
```
