```python
🔹 What is Aggregation?
Ans : -  Group by

db.collection.aggregate([ stage1, stage2, stage3 ])
🔹 Most Important Stages


| Stage        | Purpose                   |
| ------------ | ------------------------- |
| `$match`     | Filter documents          |
| `$group`     | Group & aggregate         |
| `$project`   | Select / transform fields |
| `$sort`      | Sort results              |
| `$limit`     | Limit output              |
| `$skip`      | Pagination                |
| `$unwind`    | Deconstruct array         |
| `$lookup`    | Join collections          |
| `$count`     | Count documents           |
| `$addFields` | Add new fields            |

```

```python
🔹 Pipeline Execution Order (VERY IMPORTANT)

$match → $project → $group → $sort → $limit

✅ Always filter early for performance.
```

```python
1️⃣ Filter documents

{ $match: { status: "DELIVERED" } }
```

```python
Total sales per restaurant
***************************
db.orders.aggregate([
   {$group: {_id : "restaurantId",totalsales : {$sum : "$amount"} }}
])
```

```python
# 3️⃣ Average Order Value
db.orders.aggregate([
  { $group: { _id: null, avgOrder: { $avg: "$amount" } } }
])
```

```python
4️⃣ Sort + Limit (Top N)

db.orders.aggregate([
  { $sort: { amount: -1 } },
  { $limit: 5 }
])
```
