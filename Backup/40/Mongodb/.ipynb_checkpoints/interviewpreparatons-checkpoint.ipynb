```python
2.	 Do you know about the Diamond problem, and what is MRO?

The Diamond Problem occurs in multiple inheritance when a class inherits 
from two classes that both inherit from the same parent class.
```

```python
class A:
    def show(self):
        print("A")

class B(A):
    def show(self):
        print("B")

class C(A):
    def show(self):
        print("C")

class D(B, C):
    pass

d = D()
d.show()
print(D.mro())

```

```python
MRO (Method Resolution Order)

MRO defines the order in which Python looks for methods and attributes in a class

✅ It uses C3 linearization,
which respects left-to-right order but also ensures consistent and unambiguous inheritance.
```

```python
8.	  Can you tell me what is the aggregation pipeline in MongoDB?

It is a sequence of stages, where each stage takes documents as input, 
processes them, and passes the result to the next stage.

    Each stage starts with $ (e.g. $match, $group).
    
Why use Aggregation Pipeline?

Data filtering
Data grouping
Data transformation
Data joining
Data analytics & reporting
```

```python
Common Aggregation Stages (Very Important)
******************************************

1️⃣ $match
    Filters documents (like WHERE)
    
   { $match: { status: "PAID" } }

2️⃣ $project

Select or transform fields
   { $project: { customer: 1, amount: 1, _id: 0 } }

3️⃣ $group

Groups documents and applies aggregations

  { $group: { _id: "$customer", total: { $sum: "$amount" } } }

4️⃣ $sort

Sorts documents
{ $sort: { total: -1 } }

5️⃣ $limit / $skip

  { $limit: 5 }

6️⃣ $lookup

Join with another collection

Documents → $match → $project → $group → $sort → Output
```

```python
22.	How to join two collections

MongoDB $lookup behaves like a LEFT OUTER JOIN.

5️⃣ Self Join ($lookup on Same Collection)
   employee:-
        [
          { "_id": 1, "name": "Alice", "manager_id": 3 },
          { "_id": 2, "name": "Bob",   "manager_id": 3 },
          { "_id": 3, "name": "Carol", "manager_id": null },
          { "_id": 4, "name": "David", "manager_id": 2 }
        ]

    _id → Employee ID
    manager_id → References another employee who is the manager

    3. MongoDB Aggregation Using $lookup

db.employees.aggregate([
  {
    $lookup: {
      from: "employees",      // same collection
      localField: "manager_id", // field in current document
      foreignField: "_id",    // field to match in the same collection
      as: "manager_info"      // name of the array for result
    }
  }
])
```

```python
4️⃣ $lookup with Multiple Conditions

Using $expr:

1️⃣ Simple $lookup (Equality Match)

Users Collection (users)
[
  { "_id": 1, "name": "Alice" },
  { "_id": 2, "name": "Bob" },
  { "_id": 3, "name": "Carol" }
]
Orders Collection (orders)
[
  { "_id": 101, "user_id": 1, "item": "Laptop" },
  { "_id": 102, "user_id": 2, "item": "Mobile" },
  { "_id": 103, "user_id": 1, "item": "Mouse" }
]
Goal: Get all orders for each user.

 db.users.aggregate([
  {
    $lookup: {
      from: "orders",       // Collection to join
      localField: "_id",    // Field from users
      foreignField: "user_id", // Field from orders
      as: "orders"          // Name of the output array field
    }
  }
])

👉 Most commonly used
👉 Matches localField = foreignField
```

```python
In a single (flat) list, shallow copy and deep copy behave the same because there are no nested mutable objects.
The difference appears only when the list contains nested mutable objects.
In a shallow copy, the inner (nested) objects share the same memory reference as the original, whereas in a deep copy, 
all nested objects are copied into new memory locations.
```

```python
# Reverse a number
def solve(nm):
    if len(nm) == 0:
        return []
    return [[nm[-1]] + solve(nm[:-1])

l = [1,2,3,4,5]
res = solve(l)
print()
```

```python
1.	Can you tell me the architecture of your recent project?

My recent project is built using a microservices architecture, where the application is split into multiple independent services. 
    
Each service is responsible for a specific business capability and can be developed, deployed, and scaled independently.

    Client (Web / Mobile)
        ↓
API Gateway
        ↓
Authentication & Authorization Service
        ↓
Microservices Layer
  ├── User Service
  ├── Order Service
  ├── Payment Service
  ├── Notification Service
        ↓
Database (Per Service)
        ↓
Message Broker (Async Communication)
```

```python
1️⃣ Client Layer

Web / Mobile / Third-party clients

Communicate using REST APIs / JSON
```

```python
2️⃣ API Gateway

Single entry point for all requests

Responsibilities:

Routing

Authentication

Rate limiting

Request validation

Tools: Nginx / AWS API Gateway
```

```python
3️⃣ Authentication & Authorization

Implemented using JWT / OAuth2

Validates tokens before forwarding requests

Role-based access control (RBAC)
```

```python
4️⃣ Microservices Layer

Each service:
    Owns its business logic
    Has its own database

Communicates via:
    REST (sync)
    Message Queue (async)

    Examples:
    User Service → User management
    Order Service → Order processing
    Payment Service → Transactions
    Notification Service → Email/SMS

5️⃣ Database per Service

        Each microservice has an independent database
        Avoids tight coupling

        Technologies:
            MySQL / PostgreSQL
            MongoDB (where applicable)

6️⃣ Asynchronous Communication

Used Kafka / RabbitMQ / SQS

For:
    Event-driven processing
    Decoupling services
    Better scalability

    Example:
        Order Created → Event → Notification Service

7️⃣ Containerization & Deployment

    Services packaged using Docker
    Deployed on Kubernetes
    Features:
        Auto-scaling
        Rolling deployments
        Self-healing

8️⃣ CI/CD Pipeline
    Tools: Git, Jenkins, Docker, Kubernetes, ArgoCD
    Flow:
        Code → Build → Test → Docker Image → Deploy

9️⃣ Observability & Monitoring
    
    Logging: ELK Stack
    Monitoring: Prometheus + Grafana

    Tracing: Distributed tracing (optional)

🔹 Security Measures
    HTTPS
    JWT authentication
    Role-based authorization
    Secrets managed via environment variables / vault

```
