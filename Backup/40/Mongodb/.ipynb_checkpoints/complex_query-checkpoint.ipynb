```python
[
  { "_id": 1, "user_id": 101, "amount": 500, "created_at": ISODate("2024-12-01T10:00:00Z") },
  { "_id": 2, "user_id": 101, "amount": 700, "created_at": ISODate("2024-12-05T09:00:00Z") },
  { "_id": 3, "user_id": 102, "amount": 300, "created_at": ISODate("2024-12-03T12:00:00Z") },
  { "_id": 4, "user_id": 102, "amount": 900, "created_at": ISODate("2024-12-06T08:00:00Z") }
]
```

```python
{
  $sort: { user_id: 1, created_at: -1 }
}

Sorts documents:
    user_id: 1 → ascending
    created_at: -1 → newest orders first

After sort:-
[
  { "_id": 2, "user_id": 101, "amount": 700, "created_at": "2024-12-05" },
  { "_id": 1, "user_id": 101, "amount": 500, "created_at": "2024-12-01" },

  { "_id": 4, "user_id": 102, "amount": 900, "created_at": "2024-12-06" },
  { "_id": 3, "user_id": 102, "amount": 300, "created_at": "2024-12-03" }
]
```

```python
🔹 2️⃣ $group
{
  $group: {
    _id: "$user_id",
    latestOrder: { $first: "$$ROOT" }
  }
}

✔ What it does
    Groups documents by user_id
    $first picks the first document from each group
    $$ROOT means the entire document
```

```python
✔ Result after $group
**********************
[
  {
    "_id": 101,
    "latestOrder": {
      "_id": 2,
      "user_id": 101,
      "amount": 700,
      "created_at": "2024-12-05"
    }
  },
  {
    "_id": 102,
    "latestOrder": {
      "_id": 4,
      "user_id": 102,
      "amount": 900,
      "created_at": "2024-12-06"
    }
  }
]
```

```python
2 :-  How to fetch the latest PAID order for each user

[
  { "_id": 1, "user_id": 101, "amount": 500, "status": "PENDING", "created_at": ISODate("2024-12-01T10:00:00Z") },
  { "_id": 2, "user_id": 101, "amount": 700, "status": "PAID",    "created_at": ISODate("2024-12-05T09:00:00Z") },
  { "_id": 3, "user_id": 101, "amount": 900, "status": "PAID",    "created_at": ISODate("2024-12-07T08:00:00Z") },

  { "_id": 4, "user_id": 102, "amount": 300, "status": "PAID",    "created_at": ISODate("2024-12-03T12:00:00Z") },
  { "_id": 5, "user_id": 102, "amount": 800, "status": "CANCELLED","created_at": ISODate("2024-12-06T11:00:00Z") }
]
```

```python
  db.orders.aggregate([
  {
    $match: { status: "PAID" }
  },
  {
    $sort: { user_id: 1, created_at: -1 }
  },
  {
    $group: {
      _id: "$user_id",
      latestPaidOrder: { $first: "$$ROOT" }
    }
  },
  {
    $replaceRoot: { newRoot: "$latestPaidOrder" }
  }
])

{
  $match: { status: "PAID" }
}

Filters documents
Keeps only PAID orders

✔ Result after $match

[
  { "_id": 2, "user_id": 101, "amount": 700, "status": "PAID", "created_at": "2024-12-05" },
  { "_id": 3, "user_id": 101, "amount": 900, "status": "PAID", "created_at": "2024-12-07" },
  { "_id": 4, "user_id": 102, "amount": 300, "status": "PAID", "created_at": "2024-12-03" }
]

$sort

{
  $sort: { user_id: 1, created_at: -1 }
}

Sorts by:

    user_id ascending
    created_at descending (latest first)

[
  { "_id": 3, "user_id": 101, "amount": 900, "status": "PAID", "created_at": "2024-12-07" },
  { "_id": 2, "user_id": 101, "amount": 700, "status": "PAID", "created_at": "2024-12-05" },

  { "_id": 4, "user_id": 102, "amount": 300, "status": "PAID", "created_at": "2024-12-03" }
]
```
