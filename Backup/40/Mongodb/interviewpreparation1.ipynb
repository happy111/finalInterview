```python
✅ Why RDS Proxy is needed

Without proxy:
    Each app / Lambda opens new DB connections
    DB hits connection limits
    Slower performance & timeouts
    
With RDS Proxy:
    Reuses & pools connections
    Controls concurrent connections
    Improves scalability & reliability
```

```python
16️⃣ Scenario: Find top 5 selling items
****************************************

db.orders.aggregate([
  { $unwind: "$items" },
  { $group: { _id: "$items.name", qty: { $sum: 1 } } },
  { $sort: { qty: -1 } },
  { $limit: 5 }
])
```

```python
17️⃣ Scenario: Aggregation is slow

Q: How do you optimize?

In MongoDB aggregation, earlier stages reduce the amount of data passed to later stages.

    1. 1️⃣ Use $match as early as possible
    2. 2️⃣ Create Proper Indexes
         Indexes help MongoDB avoid scanning the entire collection (COLLSCAN) and instead use IXSCAN, which is much faster.
        🔹 A. Index for $match
            What $match does
            
            Filters documents (like WHERE in SQL)
            
            If indexed, MongoDB can directly jump to matching documents
             *************************************
            ❌ Without index on $match
                 MongoDB:
                        Scans every document
       3️⃣ Reduce $unwind cost
                $unwind explodes documents
                Filter before unwind
        4️⃣ Use $project to reduce document size
        
        ✔ Remove unused fields early
        5️⃣ Avoid large in-memory sorting
        7️⃣ Use $group efficiently
        8️⃣ Analyze using explain()
        🔟 Sharding (for huge datasets)

I optimize slow aggregation by pushing $match and $project early, using proper indexes,
minimizing $unwind, avoiding in-memory sorts, analyzing with explain(), 
and redesigning schema or pre-aggregating data if needed.
```
