```python
Why use FastAPI over Flask or Django?
*************************************
1. Faster performance (based on ASGI & Starlette).
2. Built-in data validation with Pydantic.
3. Automatic generation of OpenAPI docs.
4. Asynchronous support using async/await.
5. Type safety and editor support.
```

```python
from fastapi import FastAPI, Depends
app = FastAPI()

# Dependency function
def get_db():
    db = "DB_CONNECTION"
    return db

# Route using the dependency
@app.get("/items/")
def read_items(db=Depends(get_db)):
    return {"db_connection": db}
```

```python
Why Use Dependencies?
Avoid duplication – Centralize logic like DB access, auth, config.

Testability – Easily mock/override dependencies in tests.

Separation of concerns – Cleaner and more maintainable code.


    
```

```python
Multithreading
************* 
1. Multiple threads running concurrently
2. Multiple OS-level threads
3. Uses threading.Thread
4. Overhead due to threads switching

Asyncio (Async/Await)
*********************
1. Single-threaded cooperative multitasking
2. Single thread, uses event loop
3. Best for I/O-bound & non-blocking async tasks
4. Uses async def, await, and asyncio module
5. More efficient (no thread switching)
```

```python
def solve(nested_dict,parent_key=" ",sep="."):
    temp = {}
    for key,v in nested_dict.items():
        new_key = parent_key + sep + key if parent_key else key
        if isinstance(v,dict):
            temp.update(solve(v,new_key,sep=sep))
        else:
            temp[new_key] = v
    return temp

nested_dict = {'a': 1, 'b': {'c': 2, 'd': {'e': 3}}}
res = solve(nested_dict)
print(res)
```
