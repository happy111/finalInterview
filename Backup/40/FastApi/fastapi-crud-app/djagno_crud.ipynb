```python
Model Layer
***********
from django.db import models

class Employee(models.Model):
    name = models.CharField(max_length=100)
    department = models.CharField(max_length=50)
```

```python
# repositories/employee_repository.py
from .models import Employee

class EmployeeRepository:
    def get_all_employees(self):
        return Employee.objects.all()

    def get_by_department(self, dept):
        return Employee.objects.filter(department=dept)

    def create_employee(self, name, department):
        return Employee.objects.create(name=name, department=department)
```

```python
Service Layer
************
# services/employee_service.py
from repositories.employee_repository import EmployeeRepository

class EmployeeService:
    def __init__(self):
        self.repo = EmployeeRepository()

    def list_all_employees(self):
        return self.repo.get_all_employees()

    def get_employees_in_department(self, dept):
        return self.repo.get_by_department(dept)

    def add_employee(self, name, dept):
        # Add any business logic/validation here
        return self.repo.create_employee(name, dept)
```

```python
# views.py
from django.http import JsonResponse
from services.employee_service import EmployeeService

service = EmployeeService()

def all_employees_view(request):
    employees = service.list_all_employees()
    data = list(employees.values())
    return JsonResponse(data, safe=False)

def add_employee_view(request):
    if request.method == "POST":
        name = request.POST["name"]
        dept = request.POST["department"]
        new_emp = service.add_employee(name, dept)
        return JsonResponse({"id": new_emp.id, "message": "Employee added"})
```

```python
class DatabaseConnection:
    _instance = None

_instance is a class variable, initialized as None.
It will store the single shared instance of the class.
```

```python
__new__ Method: Core of Singleton

def __new__(cls, *args, **kwargs):
    if cls._instance is None:
        print("Creating new database connection...")
        cls._instance = super(DatabaseConnection, cls).__new__(cls)
        cls._instance.init_connection()
    return cls._instance

__new__() is a special method in Python that’s called before __init__,
and is responsible for creating the object.
```
