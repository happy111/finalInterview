```python
from fastapi import FastAPI, HTTPException, Depends

FastAPI: Main class to create the app.
HTTPException: Used to return error messages with custom status codes.
Depends: Helps manage dependencies (like database sessions).
```

```python
from pydantic import BaseModel
from typing import List, Optional

BaseModel: Used to define request and response schemas (data validation).
List, Optional: Type hinting. Optional means the field is not required.
```

```python
from sqlalchemy import create_engine, Column, Integer, String, ForeignKey
from sqlalchemy.ext.declarative import declarative_base
from sqlalchemy.orm import sessionmaker, relationship, Session

create_engine: Connects to the database.
Column, Integer, String, ForeignKey: Define table columns.
declarative_base: Base class for ORM models.
sessionmaker: Creates a DB session factory.
relationship: Links two tables (foreign key relation).
Session: Represents a DB session.
```

```python
 Database Connection

DATABASE_URL = "mysql+pymysql://username:password@localhost:3306/your_db_name"

engine = create_engine(DATABASE_URL)
SessionLocal = sessionmaker(autocommit=False, autoflush=False, bind=engine)
Base = declarative_base()
```

```python
engine: DB engine created using the URL.
SessionLocal: Used to interact with the database.
Base: Used to create ORM models (tables).
```

```python
Models :

class Category(Base):
    __tablename__ = "categories"
    id = Column(Integer, primary_key=True, index=True)
    name = Column(String(100), unique=True)
    products = relationship("Product", back_populates="category")
```

```python
Creates a categories table with:
id: Primary key.
name: Unique name for category.
products: Related products (one-to-many).
```

```python
class Product(Base):
    __tablename__ = "products"
    id = Column(Integer, primary_key=True, index=True)
    name = Column(String(100))
    price = Column(Integer)
    category_id = Column(Integer, ForeignKey("categories.id"))
    category = relationship("Category", back_populates="products")
```
