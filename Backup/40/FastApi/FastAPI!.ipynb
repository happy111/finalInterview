```python
2. What are the key features of FastAPI?
Answer:

Fast (using Starlette + Uvicorn)
Based on ASGI (Async)
Auto-generated interactive docs (Swagger and ReDoc)
Type-safe (using Pydantic)
Dependency injection support
   
Built-in data validation
Easy to use
```

```python
What is Dependency Injection in FastAPI?

Dependency Injection (DI) in FastAPI is a way to declare reusable logic — such as:
Database sessions
Authentication
Configuration

FastAPI automatically resolves and injects those dependencies using the Depends() function.
```
