```python
FastAPI is built on asynchronous programming. Can you explain the difference 
between using def vs async def in FastAPI endpoints?
Also, when should you use each one in real-world production APIs?

| Type        | Works On   | Blocking?      | Internal Behavior             | Best Use                                    |
| ----------- | ---------- | -------------- | ---------------------------   | ------------------------------------------- |
| `async def` | event loop | ❌ non-blocking | runs directly in event loop  | async DB, async HTTP, high-concurrency APIs |
| `def`       | threadpool | ✔ blocking     | executed in threadpool        | CPU work, old sync libraries                |
```

```python
The event loop is a single-threaded scheduler that runs asynchronous tasks.
It switches between tasks when they hit an await, allowing other I/O operations to run concurrently.
FastAPI uses the event loop to serve thousands of requests without creating new threads.
```

```python
1. What happens if we use def in a FastAPI endpoint?

FastAPI runs def endpoints in a threadpool (from Starlette's run_in_threadpool()).

It blocks the thread until the function completes.

Best used for CPU-bound tasks or libraries that are not async, like:
    SQLAlchemy (non-async)
    boto3
    requests library
    Heavy CPU computation

2. What happens if we use async def?

    Runs on the event loop (no thread).
    
    Non-blocking, cooperative multitasking.

Best for I/O-bound async operations, such as:

    async DB drivers
    async HTTP calls
    async Redis
    async file operations

3. When to use which?
Case                                                   Use
Calling synchronous libraries	                       def
CPU-heavy logic	                                       def
High concurrency (DB calls, HTTP calls)	               async def
Using async frameworks (Tortoise ORM, httpx async)	   async def
```
