```python
1. What is FastAPI?

  .  FastAPI is a modern
  .  High-performance Python web framework used to build APIs.
  .  It is built on Starlette
  .  Pydantic
```

```python
2. Why is FastAPI so fast?
    Uses ASGI instead of WSGI
    Supports async/await natively
    Built on uvicorn + Starlette (high performance)
    Automatic request validation using Pydantic
```

```python
3. What are the key features of FastAPI?

    Automatic OpenAPI documentation
    Pydantic data validation
    Async support
    Dependency Injection
    Fast performance 
    Easy to use
```

```python
4. How do you install FastAPI?

pip install fastapi uvicorn
```
