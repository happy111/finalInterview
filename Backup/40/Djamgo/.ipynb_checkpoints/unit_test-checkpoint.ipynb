```python
7. What is setUp() and tearDown()?

        Executed before/after each test.

        def setUp(self):
            self.x = 10
        
        def tearDown(self):
            pass
```

```python
setUpClass()
    Runs only once before all test cases in the class
    Must be declared with @classmethod
    Used for heavy or expensive setup
      (DB connections, API clients, docker containers, reading big files)
    @classmethod
    def setUpClass(cls):
        print("Run once")

✔ tearDownClass()
        Runs only once after all tests
        Used for closing or cleaning resources
```

```python
11. What is mocking?

    Mocking means replacing real objects (API calls, DB calls) 
       with fake objects to avoid slow or external dependencies.

    ✔ Faster
    ✔ Without internet
    ✔ Without depending on external systems
```

```python
12. How to mock using unittest.mock?
from unittest.mock import patch

```

```python
# file: module.py
import requests

def fetch_users():
    response = requests.get("https://api.example.com/users")
    return response.status_code

            🔥 Test it WITHOUT calling the real API
            Use @patch("module.requests.get")
            It replaces requests.get only inside module.py
            It does NOT hit the real API
            
            Instead, you control what response is returned

# file: test_module.py
from unittest.mock import patch
import unittest
import module

class TestAPI(unittest.TestCase):

    @patch("module.requests.get")          # Patch target in module.py
    def test_api(self, mock_get):          # mock_get = fake requests.get
        # Step 1: Configure mock response
        mock_get.return_value.status_code = 200
        mock_get.return_value.json.return_value = {"message": "success"}

        # Step 2: Call function under test
        result = module.fetch_users()

        # Step 3: Assertions
        self.assertEqual(result, 200)

        # Step 4: Verify mock was called
        mock_get.assert_called_once_with("https://api.example.com/users")

if __name__ == '__main__':
    unittest.main()
```

```python
✅ What Are Fixtures?

Fixtures are reusable setup/cleanup code that prepares the environment required for tests.

👉 They help you avoid duplicate code
👉 They make tests clean, readable, and DRY (Don’t Repeat Yourself)
```

```python
| Method            | When it runs                          | Purpose                    |
| ----------------- | ------------------------------------- | -------------------------- |
| `setUpClass()`    | Runs **once** before all test methods | Expensive, shared setup    |
| `tearDownClass()` | Runs **once** after all test methods  | Cleanup                    |
| `setUp()`         | Runs **before each test method**      | Fresh setup for every test |
| `tearDown()`      | Runs **after each test method**       | Cleanup                    |
```

```python
import smtplib

def send_email(to_email, subject, message):
    server = smtplib.SMTP('smtp.example.com', 587)
    server.starttls()
    server.login('user@example.com', 'password')
    server.sendmail('user@example.com', to_email, f"Subject: {subject}\n\n{message}")
    server.quit()
    return True
```

```python
import unittest
from unittest.mock import patch, MagicMock
from emailer import send_email

class TestSendEmail(unittest.TestCase):
    
    @patch('emailer.smtplib.SMTP')  # Mock the SMTP class
    def test_send_email_success(self, mock_smtp):
        mock_server_instance = MagicMock()
        mock_smtp.return_value = mock_server_instance
        result = send_email('test@example.com', 'Test Subject', 'Hello!')
        # Assertions
        mock_smtp.assert_called_with('smtp.example.com', 587)
        mock_server_instance.starttls.assert_called_once()
        mock_server_instance.login.assert_called_once_with('user@example.com', 'password')
        mock_server_instance.sendmail.assert_called_once()
        mock_server_instance.quit.assert_called_once()
        self.assertTrue(result)

if __name__ == '__main__':
    unittest.main()
```
