```python
# How do we authenticate a microservice?

ANS : -  1. Architecture (Simple and Practical)

[User] → [API Gateway] → [Service A] → [Service B]
                         |                
                         → (calls Auth Service to get JWT)

1. Service A needs to call Service B.
2. Before calling, Service A gets a JWT token from the Auth Service.
3. Service A sends the JWT in the header.
4. Service B verifies the JWT using a shared secret or public key.
```

```python
 What is Idempotency?

Idempotency means that repeating the same operation multiple times 
produces the same result as
    doing it once — without any unintended side effects.

Pressing a "Stop" button in an elevator multiple times —
same effect: the elevator stops.
```

```python
CQRS is a software design pattern that separates read and write operations
into different models, allowing them to be optimized,
scaled, and maintained independently.


Key Concept:
Command = Write operation (e.g., create, update, delete)
Query = Read operation (e.g., get, list, search)
They don’t share the same model or database logic

🧩 Why Use CQRS?
✅ Improve performance and scalability (read-heavy apps)
✅ Optimize read and write models separately
✅ Makes complex domains easier to manage
✅ Supports event sourcing and microservices
```

```python
The @dataclass decorator in Python automatically generates special methods like __init__,
__repr__, __eq__, etc., for classes that are mainly used to store data.

from dataclasses import dataclass

@dataclass
class Book:
    title: str
    author: str
    pages: int


class Book:
    def __init__(self, title, author, pages):
        self.title = title
        self.author = author
        self.pages = pages

    def __repr__(self):
        return f'Book(title={self.title}, author={self.author}, pages={self.pages})'
```
