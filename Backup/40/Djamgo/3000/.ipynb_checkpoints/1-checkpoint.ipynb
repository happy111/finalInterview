```python
ASGi  Daphne
 
Type: Asynchronous (also supports synchronous code)

Standard for: Modern Python web frameworks like FastAPI, Django (3.0+), Starlette.

Handles: HTTP, WebSockets, HTTP/2, and more.

Advantage: Great for real-time apps, chat apps, live dashboards, etc.

Concurrency: Non-blocking — can handle thousands of simultaneous 
connections using async/await.

Ex : Notification, Realtime message
```

```python

wsgi   Gunicorn


Type: Synchronous

Standard for: Traditional Python web apps like Django (pre-3.0), Flask, etc.

Handles: Only HTTP 1.1 requests.

Limitation: Can’t handle asynchronous code like WebSockets, long-lived connections, or real-time apps.

Concurrency: Blocking — each request is handled one at a time per thread/process.

Use case: Simple websites, REST APIs where async features are not needed.
```

```python
client -> Nginx -> gunicorn -> wsgi
```
