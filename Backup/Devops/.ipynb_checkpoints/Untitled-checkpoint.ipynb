```python
VPC
We create separate VPCs in both the primary and disaster recovery (DR) 
regions, ensuring proper network isolation and high availability.

Storage (Amazon S3)
Amazon S3 is configured with Cross-Region Replication (CRR) to
replicate data automatically from the primary region to a secondary region, 
ensuring data durability and availability.

Database (Amazon RDS)
For relational databases, cross-region read replicas are enabled in Amazon RDS to
support disaster recovery and reduce recovery time in case of regional failure.

DynamoDB
Amazon DynamoDB Global Tables are used to provide multi-region, 
active-active replication, enabling low-latency access and seamless failover.

Amazon EKS
We deploy separate EKS clusters in both the primary and secondary regions.
The primary cluster handles live traffic, 
while the secondary cluster is kept ready for failover during regional outages.

Monitoring and Observability
Amazon CloudWatch, along with Prometheus and Grafana, 
is used for monitoring, alerting, and visualizing metrics across both regions.

Traffic Management (Route 53)
Amazon Route 53 failover routing policies are configured so
that if the primary region becomes unavailable,
traffic is automatically redirected to the DR region.

Secrets Management
Application secrets are securely managed using AWS Secrets Manager 
and replicated or synchronized across regions to ensure availability during failover.

CI/CD Pipeline and GitOps
CI/CD pipelines are implemented for automated deployments, 
and Argo CD is used for GitOps-based continuous delivery, 
ensuring consistent application state across both regions.
```

```python
hYBRID NETWORKING:- connect bet onpremise to aws

AWS Direct Connect
AWS Direct Connect provides a dedicated, private network connection
between the on-premises environment and AWS. It offers low latency, high bandwidth,
and consistent performance, making it ideal for large-scale or latency-sensitive workloads.

Site-to-Site VPN
A Site-to-Site VPN establishes an encrypted IPsec tunnel between the Customer
Gateway (on-premises) and the AWS Virtual Private Gateway or Transit Gateway.
    It is a cost-effective solution and commonly used as a primary connection
for small workloads or as a backup to Direct Connect.

AWS Transit Gateway
AWS Transit Gateway acts as a central hub to connect multiple VPCs and on-premises networks.
    It simplifies network architecture by reducing complex peering connections and supports
    scalable, secure, and centralized routing for hybrid and multi-VPC environments.
```

```python
Disaster recovery: Design ecommere website for disaster recovery
******************

    
1. Downtime Allowed (Active–Passive / Backup-Based DR)
    In this approach, brief downtime is acceptable.
    All critical services are backed up regularly.
    In the event of a disaster,
      infrastructure is recreated in the DR region, including:
        Amazon EKS (Kubernetes clusters)
        Amazon RDS (from automated backups or snapshots)
        Amazon S3 (using backups or replicated data)
        Traffic is redirected to the DR region after services are restored.
        This approach has higher RTO and RPO but is cost-effective.

2. No Downtime (Active–Passive / Near-Zero Downtime DR)

        This approach is designed for high availability and minimal service disruption.
        Database Replication
        Enable cross-region read replicas for Amazon RDS.
        Use DynamoDB Global Tables for active-active database replication (if applicable).

        Traffic Failover
        Configure Amazon Route 53 failover routing policies.

        If the primary region becomes unavailable, traffic is automatically routed to the DR region.

        Application Layer
        
        Maintain a standby EKS cluster in the DR region with pre-deployed workloads.
        Use CI/CD pipelines and Argo CD to keep configurations in sync across regions.

Storage

Enable S3 Cross-Region Replication (CRR) to keep static assets and backups available.

Monitoring

Use CloudWatch, Prometheus, and Grafana to detect failures and trigger alerts instantly.
```

```python
Cross account :-

    AWS STS (Security Token Service) is a service that provides temporary,
limited-privilege credentials to access AWS resources.

    
I have 2 accounts

EC2 is in Account A
S3 bucket is in Account B

Create an IAM Role in Account B
    Give S3 permissions (GetObject / PutObject)
    This role will be used to access the S3 bucket

Allow Account A to Assume the Role
    Update the role’s trust policy
    Allow EC2 role from Account A

Create an IAM Role for EC2 in Account A
    Attach permission to assume the role in Account B

Attach IAM Role to EC2
    Assign the EC2 role to the instance

STS Provides Temporary Credentials
    EC2 assumes the role
    AWS STS generates a temporary security token

EC2 Accesses S3 Securely
    No access keys needed
    Access is secure and temporary
```
