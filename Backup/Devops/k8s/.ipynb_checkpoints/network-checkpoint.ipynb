```python
Securing Kubernetes:-

1. Namespace:-
    Logical Isolation in a cluster 

  Namespaces in Kubernetes provide logical isolation, which helps manage a
single Kubernetes cluster across different teams within a company.
        
       May be issue :-
            . OwnerShip
            . resource quota
     When we create namespace, define resource quota[cpu,memory]
       
           
         
    How to grand permission to pod. then came RBAC
    
2. RBAC : Role based access control
    RoleBind : Role binds to service account.

        Three resources :
           . service account : User Account
           . Role 
           . Role Binding

               Pod run service account : kubectl edit pod <name> -n <namespace>
        
        Every pod has service account. Kubernet assigns default service account.

******************************
        1. Create service account
            kubectl create serviceaccount payment-user -n payments
        2.  Then create role
                kind: Role
                metadata:
                    name: pod-reader
                    namespace: payments-ns
                rules:
                - apiGroups: [""]
                  resources: ["pods"]
                  verbs:["get","list"]
         3. Then Bind
                      apiVersion: rbac.authorization.k8s.io/v1
                      kind: RoleBinding
                      metadata:
                          name: read-pods-binding
                          namespace: payments
                      subjects:
                      - kind: ServiceAccount
                        name: payments-user
                        namespace: payments
                      roleRef:
                            kind: Role
                            name: pod-reader
                            apiGroup: rbac.authorization.k8s.io

            4. kubectl auth can-i create pods --as=system:serviceaccount:payment-ns:payment-user -n payment-ns
                output : no
               
3. Network policies
4. Advanced (kyverno)
5. Secrets
6. GIT  (External secrets Operator)
```
