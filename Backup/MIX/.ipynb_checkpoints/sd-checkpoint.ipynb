```python
Question 1: What is System Design?

    System design is the process of defining architecture, components, modules,
    interfaces, and data flow of a system to meet business and technical requirements such as:
    
    Scalability
    Reliability
    Availability
    Performance
    Security

There are two types:

High-Level Design (HLD) – overall architecture

Low-Level Design (LLD) – class-level and code-level design
```

```python
🔹 Question 2: What is High-Level Design (HLD)?

HLD focuses on overall system architecture and interactions between components.

It includes:
    Load balancer
    API Gateway
    Databases
    Caching
    Message queues
    Microservices
    Data flow

Client → Load Balancer → API Service → Database
                    → Cache
```

```python
🔹 Question 3: What is Low-Level Design (LLD)?
✅ Answer:

LLD focuses on class design, object interactions, and data structures.

It includes:

    Classes
    Interfaces
    Design patterns
    Method signatures
    Database schema
```

```python
Factory Pattern → Create different notification channels (Email, SMS, Push) dynamically

Observer Pattern → Notify multiple subscribers when a new notification is created

Singleton Pattern → Ensure there’s only one instance of Notification Service
```

```python
from abc import ABC, abstractmethod

# =========================
# Observer Pattern
# =========================
class Observer(ABC):
    @abstractmethod
    def notify(self, message: str):
        pass


# =========================
# Concrete Observers (Channels)
# =========================
class EmailNotifier(Observer):
    def notify(self, message: str):
        print(f"Email sent: {message}")


class SMSNotifier(Observer):
    def notify(self, message: str):
        print(f"SMS sent: {message}")


class PushNotifier(Observer):
    def notify(self, message: str):
        print(f"Push notification sent: {message}")


# =========================
# Factory Pattern
# =========================
class NotifierFactory:
    @staticmethod
    def get_notifier(channel: str) -> Observer:
        if channel == "email":
            return EmailNotifier()
        elif channel == "sms":
            return SMSNotifier()
        elif channel == "push":
            return PushNotifier()
        else:
            raise ValueError(f"Unknown channel: {channel}")


# =========================
# Singleton Pattern
# =========================
class NotificationService:
    _instance = None

    def __new__(cls):
        if cls._instance is None:
            cls._instance = super(NotificationService,cls).__new__(cls)
            cls._instance.subscribers = []
        return cls._instance

    # Subscribe channel
    def subscribe(self, observer: Observer):
        self.subscribers.append(observer)

    # Send notification
    def send_notification(self, message: str):
        for observer in self.subscribers:
            observer.notify(message)


# =========================
# Client Code
# =========================
if __name__ == "__main__":
    service = NotificationService()

    # Create channels dynamically using Factory
    email_channel = NotifierFactory.get_notifier("email")
    sms_channel = NotifierFactory.get_notifier("sms")
    push_channel = NotifierFactory.get_notifier("push")

    # Subscribe channels (Observer Pattern)
    service.subscribe(email_channel)
    service.subscribe(sms_channel)
    service.subscribe(push_channel)

    # Send a notification
    service.send_notification("New notification: System update at 10 PM")
```
