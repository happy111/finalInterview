```python
A cold start happens when AWS Lambda creates a brand-new execution environment 
to run your function for the first time or after being idle

What happens during a Cold Start?

When a cold start occurs, AWS does the following:

    Creates a new container
    Loads the runtime (Python / Node / Java)
    Mounts Lambda Layers (/opt)
    Loads your function code (/var/task)
    Executes global code (outside the handler)
    Calls the handler function
```

```python
A warm start happens when AWS Lambda reuses an existing execution environment (container)
to handle a new request after a previous invocation.

Because everything is already initialized, the function runs much faster.

What happens during a Warm Start?

During a warm start:

No new container is created
Runtime is already loaded
Layers are already mounted
Global code is NOT re-executed

Only the handler function runs

⚡ Result: Low latency
```

```python
# 4️⃣ Step-by-Step Implementation
# Step 1: Component Interface

from abc import ABC, abstractmethod

class Notifier(ABC):
    @abstractmethod
    def send(self, message: str):
        pass
```

```python
# Step 2: Concrete Component
class EmailNotifier(Notifier):
    def send(self, message: str):
        print(f"Email sent: {message}")
```

```python
# Step 3: Base Decorator

class NotifierDecorator(Notifier):
    def __init__(self, notifier: Notifier):
        self._notifier = notifier

    def send(self, message: str):
        self._notifier.send(message)
```

```python
# Step 4: Concrete Decorators
# SMS Decorator
class SMSNotifier(NotifierDecorator):
    def send(self, message: str):
        super().send(message)
        print(f"SMS sent: {message}")
```

```python
notifier = EmailNotifier()
notifier = SMSNotifier(notifier)
# notifier = SlackNotifier(notifier)

notifier.send("Server is down!")
```

```python
import pandas as pd

df1 = pd.DataFrame({
    "emp_id": [1, 2, 3],
    "name": ["Umesh", "Ravi", "Amit"]
})

df2 = pd.DataFrame({
    "emp_id": [1, 2, 4],
    "salary": [50000, 60000, 70000]
})

result = pd.merge(df1, df2, on="emp_id", how="inner")
print(result)
```
