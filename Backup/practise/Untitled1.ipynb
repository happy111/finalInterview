```python
#   TTL strategy [Time to live]

TTL (Time-To-Live) defines 
 how long a piece of data 
 should live in cache, memory, or storage before it expires.

After TTL expires, the data is invalidated or evicted.

Use cases:

    URL shorteners → temporary short links.
    Caches → avoid stale data.
    Session management → auto-expire user sessions.
    Rate limiter → counters reset after a window

A. Fixed TTL
B. Sliding / Idle TTL
C. Dynamic / Custom TTL
```

```python
“Cache invalidation is removing or updating cached data when the original data changes, to avoid stale results.
Common strategies are:

TTL (time-based expiration) – automatically expire after some time.

Manual or event-driven invalidation – delete or refresh cache when data changes.

Write-through caching – update cache whenever DB is updated.
It ensures users always see fresh and consistent data.”
```

```python
A hot key is a cache or database item that is accessed very frequently,
which can overload the backend if not handled properly.
```

```python
“A distributed cache is a cache system shared across multiple servers to store frequently accessed data,
improving performance and scalability in large applications.”
```

```python
Cache-Aside is a caching strategy where the application reads from the cache first, and if the data is missing,
it loads from the database and then updates the cache.
```

```python
Write-Through is a caching strategy where every time the application writes data,
it updates both the cache and the database
at the same time, keeping them always consistent
```

```python
Write-Back is a caching strategy where data is written only to the cache first, and the cache asynchronously
updates the database later, improving write performance.”
```

```python
Read-Through is a caching strategy where the application always reads data from the cache, 
and if the data is missing, the cache itself loads it from the database and returns it,
keeping the cache updated automatically
```

```python
indexing strategy (BTREE vs HASH)

An index is a data structure that improves the speed of data retrieval from a database
at the cost of extra storage and slower writes

1️⃣ B-Tree Index:
    Stores data in a sorted tree structure.
    Good for range queries (<, >, BETWEEN).
    Supports ORDER BY efficiently.
    Slightly slower for exact matches than hash.
    Used as default index in MySQL/PostgreSQL.

2️⃣ Hash Index:
    Uses a hash function to map keys to buckets.
    Very fast for exact equality queries (=).
    Cannot do range queries or sorting.
    Best for high-speed lookups by key.

Lookup by exact key → Hash (or B-Tree)

Lookup by range or order → B-Tree
```

```python
composite index:-

  A composite index is an index on two or more columns of a table.
  It helps queries that filter or sort using multiple columns together.
```

```python
covering index :- 

A covering index is an index that contains all the columns needed by a query.
```

```python
The N+1 query problem happens when an application executes one query to get N records,
and then executes an additional
query for each of the N records, causing many unnecessary queries.
```

```python
normalization vs denormalization

Normalization: Organizing database tables to reduce redundancy and avoid data anomalies.

Denormalization: Adding redundant data intentionally to improve read performance.

Normalization: Ensures data consistency and integrity.

Denormalization: Optimizes query performance, especially for read-heavy systems

Normalization: Split a large table into multiple smaller tables with relationships.

Denormalization: Combine related tables into fewer tables to avoid joins.

| Normalization         | Denormalization     |
| --------------------- | ------------------- |
| Less data redundancy  | Faster read queries |
| Easy to maintain data | Fewer joins needed  |
| Ensures consistency   | Simpler query logic |

```

```python
“Rate limiting is a technique to control 
how many requests a client can make to a service in a given time period

Fixed Window: Allow N requests per fixed time window (e.g., 100 requests/minute).
Sliding Window: Smooths request limits over time, avoids bursts at window boundaries.
Token Bucket: Each request consumes a token; tokens are refilled over time.
Leaky Bucket: Requests flow at a constant rate; excess requests are queued or dropped.
```
