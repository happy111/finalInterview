```python
__new__ → creates object (used in Singleton, immutables)

__init__ → initializes object
```

```python
class SingleTon:
    __instance__= None
    def __new__(cls):
        if cls.__instance__ is None:
            cls.__instance__ = super().__new__(cls)
        return cls.__instance__
```

```python
obj = SingleTon()
obj1 = SingleTon()
obj == obj1
```

```python
#Q1. Find first non-repeating character in a string
from collections import defaultdict,OrderedDict

def solve(input):
    data = OrderedDict()
    for s in input:
        data[s] = data.get(s,0)+1
    for k,v in data.items():
        if v == 1:
            return k
input = "swiss"
Output = "w"
res = solve(input)
print(res)
```

```python
#check anagram
def solve(str,str1):
    new_str = str.replace(" ","").lower()
    new_str1 = str1.replace(" ","").lower()
    if len(new_str) != len(new_str1):
        return False
    if sorted(new_str) != sorted(new_str1):
        return False
    return True

str = "listen"
str1 = "silent"
res = solve(str,str1)
print(res)
```

```python
#🔹 4. Find Duplicate Elements
def findDuplicate(input):
    seen = set()
    duplicates = set()

    for k in input:
        if k in seen:
            duplicates.add(k)
        else:
            seen.add(k)
    return list(duplicates)

    
Input =  [1, 2, 3, 2, 4, 1]
Output = [1, 2]
res = findDuplicate(Input)
print(res)
```
