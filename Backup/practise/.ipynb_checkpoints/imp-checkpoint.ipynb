```python
agent.py
********
from strands import Agent
from strands.models.ollama import ollamaModel

ollama_model = ollamaModel(
    host = "https://localhosy:11434",
    model_id="llama3.2lb"
    
)

system_prompt = "You are a respectful agent."\
"you give answers in a kind and humble way."\
"You can use correct data sources to give outputs."

agent = Agent(model=ollama_model, system_prompt = system_prompt)
# By default it runs Amazon Bedrock in the background

pip install 'strands-agents[ollama]'

agent("Hello how are you")
```
