```python
What is GIL? Why is it needed?

GIL (Global Interpreter Lock) is a mutex (lock) in CPython that:
Allows only ONE thread to execute Python bytecode at a time

Why is GIL needed?

1️⃣ Memory management is NOT thread-safe
   . Python uses reference counting for memory management
   . Multiple threads changing reference counts at the same time → data corruption

  GIL prevents this by allowing only one thread to modify memory at a time.
      
3️⃣ Provides basic thread safety
```

```python
Why are strings immutable?

1️⃣ Memory efficiency
   Python reuses strings to save memory.
2️⃣ Hashing & Dictionary keys
3️⃣ Thread safety 🧵
   Immutable objects are naturally thread-safe.
   Multiple threads can read the same string
    No lock needed
    No race condition

Strings are immutable in Python to improve memory efficiency,
ensure hashing safety, provide thread safety, and allow safe reuse of string objects.
```

```python
What is MRO (Method Resolution Order)? ⭐ (Method Resolution Order)?

MRO is the order in which Python searches for a method or attribute in a class hierarchy.

When a class inherits from multiple parent classes and:

    Both parents have the same method name
    Python must decide which method to call

How Python decides MRO?
Python uses the C3 Linearization Algorithm.
    This ensures:
  ✔ Left-to-right order
```

```python
def solve(s):
    n = len(s) - 1
    pos = n
    for i in range(n,-1,-1):
        if s[i] != 0:
            s[pos] = s[i]
            pos -= 1
    for i in range(pos+1):
        s[i] = 0
    print(s)

s = [1,2,3,0,4,0,5]
res = solve(s)
print(res)
```

```python
🚨 Common Causes of Slow API Performance
1. Database Issues
   
   . Slow SQl queries
   . Missing wrong Indexes
   . N+1 query problem
   . Fetching unnecessary columns (select *)
   . Large result sets without pagination
   . Connection pool exhaustion

  HOW to Identify :
    . Enable DB query logging
    . Check execution plans (EXPLAIN)
    . Monitor slow query logs

2. Application Code Issues
  . Heavy loops or inefficient algorithms
  . Blocking I/O
  . Large JSON serialization/deserialization

  FIX:
  . Optimize logic
  . More heavy work to background tasks
  . Use async

3. No caching

   FIX:
       Implementing caching

4. Concurrency & Threading Issues
   . Limited worker processes
   . Blocking synchronous code
   . GIL contention in CPU heavy tasks
   
    FIX:-
     . Increase workers
     . Use async / multiprocessing
5. Large Payload size
  . Huge request/response bodies
  FIX
     pagination
6. Infrastructure / Network Issues
   . Low CPU/RAM
   . Disk I/o bottlenecks
   . Network latency
   . Improper load balancer config
   FIX
     . Auto scaling
     . Proper resource limits
     . CDN for static cata
```

```python
# Two Sum
def solve(arr,target):
    n = len(arr)
    res = []
    temp = {}
    for i in range(n):
        s = target - arr[i]
        if s in temp:
            res.append([arr[i],s])
        else:
            temp[arr[i]] = i
    return res
arr =  [10, 5, 20, 8, 2,10]
target = 13
res = solve(arr,target)
print(res)
```
