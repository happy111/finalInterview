```python
# Find max/min in array
def solve(arr):
    max_val = float('-inf')
    min_val = float('inf')
    for k in arr:
        if k > max_val:
            max_val = k
        if k < min_val:
            min_val = k
    return max_val,min_val

arr = [10, 5, 20, 8, 2]
max_val,min_val = solve(arr)
print(max_val,min_val)
```

```python
# Two Sum
def solve(arr,target):
    temp = {}
    res = []
    for k in range(len(arr)):
        se = target - arr[k]
        if se in temp:
            res.append((se,arr[k]))
        temp[arr[k]] = k
    return res

arr =  [10, 5, 20, 8, 2,10]
target = 13
res = solve(arr,target)
print(res)
```

```python
def solve(arr,target):
    n = len(arr)
    res = []
    temp = {}
    for i in range(n):
        s = target - arr[i]
        if s in temp:
            res.append([arr[i],s])
        else:
            temp[arr[i]] = i
    return res
arr =  [10, 5, 20, 8, 2,10]
target = 13
res = solve(arr,target)
print(res)
```

```python
# Kadane’s Algorithm (Max Subarray)

# Optimize  Imp
# Find the contiguous subarray with the largest sum.
def solve(nums):
    max_sum = curr_sum = nums[0]

    for i in range(1, len(nums)):
        curr_sum = max(nums[i], curr_sum + nums[i])
        max_sum = max(max_sum, curr_sum)

    return max_sum
nums = [-2, 1, -3, 4, -1, 2, 1, -5, 4]
res = solve(nums)
print(res)
```

```python
def solve(nums):
    max_sum = cur_sum = nums[0]

    for k in range(1,len(nums)):
        cur_sum = max(nums[k],nums[k] + cur_sum)
        max_sum = max(cur_sum,max_sum)
    return max_sum


nums = [-2, 1, -3, 4, -1, 2, 1, -5, 4]
res = solve(nums)
print(res)
```

```python
# Move zeros to end
def solve(s):
    without_zero = [k for k in s if k != 0]
    return without_zero + [0] * (len(s)-len(without_zero))

s = [1,2,3,4,0,1]
res = solve(s)
print(res)
```

```python
#🔥 Sliding Window Programs for Interview Preparation (Python)
def max_sum_subarray(arr,k):
    max_val= float('-inf')
    for i in range(len(arr)):
        value = len(arr[i:k+i])
        if value == k:
            max_val = max(max_val,sum(arr[i:k+i]))
    return max_val

arr = [2, 1, 5, 1, 3, 2]
print(max_sum_subarray(arr, 3))  # 9
```

```python
def max_sum_subarray(arr, k):
    window_sum = sum(arr[:k])
    max_sum = window_sum
    for i in range(k, len(arr)):
        window_sum += arr[i]      # add next element
        window_sum -= arr[i - k]  # remove left element
        max_sum = max(max_sum, window_sum)
    return max_sum
arr = [2, 1, 5, 1, 3, 2]
print(max_sum_subarray(arr, 3))  # 9
```

```python
# ✅ 2. Average of Subarrays of Size k
def solve(arr,k):
    first_sum = sum(arr[:k])
    res = [first_sum / k]
    window_sum = first_sum
    
    for i in range(k,len(arr)):
        window_sum = window_sum + arr[i]
        window_sum -= arr[i-k]
        avg = window_sum / k
        res.append(avg)
    return res

arr = [1, 3, 2, 6, -1, 4, 1, 8, 2]
k = 5
res = solve(arr,k)
print(res)
```

```python
# ✅ 3. Maximum Element in Each Window (Important)

def solve(arr, k):
    res = []
    for i in range(len(arr) - k + 1):
        res.append(max(arr[i:i+k]))

    return res
arr = [1, 3, -1, -3, 5, 3, 6, 7]
k = 3
res = solve(arr,k)
print(res)
```

```python
#✅ 4. Longest Subarray with Sum ≤ K   Imp

def longest_subarray_sum_k(arr, k):
    window_sum = 0
    window_start = 0
    max_length = 0

    for window_end in range(len(arr)):
        window_sum += arr[window_end]

        while window_sum > k:
            window_sum -= arr[window_start]
            window_start += 1

        max_length = max(max_length, window_end - window_start + 1)

    return max_length

arr = [3, 1, 2, 7, 4, 2, 1, 1, 5]
k = 8
print(longest_subarray_sum_k(arr, k))
```

```python
# ✅ 5. Longest Substring Without Repeating Characters
def solve(st):
    temp = {}
    start = 0
    max_len = 0
    max_str =""
    for end in range(len(st)):
        if st[end] in temp and temp[st[end]] >= start:
            start = temp[st[end]] + 1
        temp[st[end]] = end
        cur_len = end - start + 1
        if max_len < cur_len:
            max_len = cur_len
            max_str = st[start:end+1]
    return max_str

str = "abcabcbb"
res = solve(str)
print(res)
```

```python
# ✅ 6. Minimum Size Subarray Sum ≥ Target

def minSubArrayLen(target,nums):
    left = 0
    current_sum = 0
    min_len = float('inf')
    for i in range(len(nums)):
        current_sum += nums[i]
        while current_sum >= target:
            min_len = min(min_len, i - left + 1)
            current_sum -= nums[left]
            left += 1
    return 0 if min_len == float('inf') else min_len
target = 7
nums = [2,3,1,2,4,3]
print(minSubArrayLen(target, nums))  # Output: 2
```

```python
#✅ 7. Longest Substring with At Most K Distinct Characters
def longestSubstringKDistinct(s,k):
    left = 0
    freq = {}
    max_len = 0
    for right in range(len(s)):
        freq[s[right]] = freq.get(s[right],0)+1
        while len(freq) > k:
            freq[s[left]] -= 1
            if freq[s[left]] == 0:
                del freq[s[left]]
            left += 1
        max_len = max(max_len,right-left+1)
    return max_len
s = "eceba"
k = 2
print(longestSubstringKDistinct(s, k))  # Output: 3
```

```python
# 1️⃣ Two Sum
def solve(arr,k):
    print(arr)
    res = []
    left = 0
    right = len(arr) - 1
    while left <= right:
        s = arr[left] + arr[right]
        if s == k:
            res.append((arr[left],arr[right]))
            return res
        elif s < k:
            left += 1
        else:
            right -=1
    return [-1,-1]

li = [1,2,3,4,5,6]
target = 7
res = solve(li,target)
print(res)
```
