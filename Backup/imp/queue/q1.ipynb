```python
# Queue
 1. A Queue is a linear data structure
 2. FIFO – First In First Out

| Operation            | Meaning                   |
| -------------------- | ------------------------- |
| **enqueue(x)**       | Insert element at rear    |
| **dequeue()**        | Remove element from front |
| **peek() / front()** | Get first element         |
| **isEmpty()**        | Check if queue is empty   |
| **size()**           | Return number of elements |

You can implement a queue in 4 ways:

    Using List (not recommended for heavy operations)
    Using Collections.deque (best & fastest)
    Using queue.Queue (thread-safe)
    Custom queue using class (important for interviews)
```

```python
class MyQueue:
    def __init__(self):
        self.items = []

    def enqueue(self, data):        # Insert element
        self.items.append(data)

    def dequeue(self):             #  Remove
        if self.isEmpty():
            return "Queue is empty"
        return self.items.pop(0)

    def peek(self):                  # peek First element
        if self.isEmpty():
            return "Queue is empty"
        return self.items[0]

    def isEmpty(self):
        return len(self.items) == 0

    def size(self):
        return len(self.items)


q = MyQueue()
q.enqueue(5)
q.enqueue(10)
q.enqueue(15)
print(q.dequeue())  # 5
print(q.peek())     # 10
```

```python
# 1️⃣ Reverse a Queue

from collections import deque
def reverse_queue(q):
    if not q:
        return
    temp = q.popleft()
    reverse_queue(q)
    q.append(temp)

q = deque([1,2,3,4])
print(q)
reverse_queue(q)
print(q)   # deque([4,3,2,1])
```

```python
from collections import deque

def reverseFirstK(q, k):
    if k > len(q):
        return q
    stack = []
    # Step 1: push first k elements to stack
    for _ in range(k):
        stack.append(q.popleft())

    # Step 2: pop from stack and add back to queue

    while stack:
        q.append(stack.pop())

    # Step 3: rotate remaining elements
    size = len(q)
    for _ in range(size - k):
        q.append(q.popleft())

    return list(q)


q = deque([1,2,3,4,5])
print(reverseFirstK(q, 3))  
```

```python
def timeRequiredToBuy(tickets, k):
    time = 0
    for i in range(len(tickets)):
        if i <= k:
            time += min(tickets[i], tickets[k])
        else:
            time += min(tickets[i], tickets[k] - 1)
    return time

r = [1,5,2,3,7]
res = timeRequiredToBuy(r,2)
print(res)
```

```python
from collections import deque

def timeRequiredToBuy(tickets, k):
    q = deque()
    # Put (index, tickets_needed) into queue
    for i, t in enumerate(tickets):
        q.append((i, t))
    time = 0
    while q:
        index, needed = q.popleft()
        needed -= 1       # person buys 1 ticket
        time += 1
        # If this is person k and they finished → stop
        if index == k and needed == 0:
            return time
        # Otherwise if they still need tickets, put them back
        if needed > 0:
            q.append((index, needed))
```

```python
r = [1,5,2,3,7]
res = timeRequiredToBuy(r,2)
print(res)
```

```python
class MyQueue:

    def __init__(self):
        self.s1 = []  # stack1
        self.s2 = []  # stack2

    def push(self, x):
        # Always push into s1
        self.s1.append(x)

    def pop(self):
        # Move all elements from s1 to s2 (reverse order)
        if not self.s2:
            while self.s1:
                self.s2.append(self.s1.pop())
        return self.s2.pop()

    def peek(self):
        # Move elements if s2 is empty
        if not self.s2:
            while self.s1:
                self.s2.append(self.s1.pop())
        return self.s2[-1]

    def empty(self):
        return not self.s1 and not self.s2
```

```python
q = MyQueue()
q.push(10)
q.push(20)
q.push(30)

q.pop() # 10
q.peek() # 20
q.empty() # False
```

```python
from collections import deque

class MyStack:

    def __init__(self):
        self.q = deque()

    def push(self, x: int) -> None:
        self.q.append(x)
        # Rotate all other elements behind the new element
        for _ in range(len(self.q) - 1):
            self.q.append(self.q.popleft())
       

    def pop(self) -> int:
        return self.q.popleft()

    def top(self) -> int:
        return self.q[0]

    def empty(self) -> bool:
        return len(self.q) == 0
```

```python
obj = MyStack()
obj.push(1)
obj.push(2)
obj.push(3)
```
