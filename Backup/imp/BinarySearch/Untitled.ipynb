```python
IAM (Identity and Access Management) is used to control who can access what in AWS.

Key components:
    Users
    Groups
    Roles
    Policies
```

```python
🔐 Amazon S3 – All Permissions
******************************
1️⃣ Bucket-Level Permissions
    s3:CreateBucket – Create a new bucket
    s3:DeleteBucket – Delete a bucket
    s3:ListBucket – List objects in a bucket
2️⃣ Object-Level Permissions
    s3:GetObject – Read/download object
    s3:PutObject – Upload object
    s3:DeleteObject – Delete object
3️⃣ Multipart Upload Permissions
```

```python
User Based permission in s3
***************************
{
  "Version": "2012-10-17",
  "Statement": [
    {
      "Effect": "Allow",
      "Principal": {
        "AWS": [
          "arn:aws:iam::123456789012:user/UserA",
          "arn:aws:iam::123456789012:user/UserB"
        ]
      },
      "Action": [
        "s3:ListBucket",
        "s3:GetObject"
      ],
      "Resource": [
        "arn:aws:s3:::my-bucket",
        "arn:aws:s3:::my-bucket/*"
      ]
    }
  ]
}

```

```python
✅ What is a Bucket Policy (Clear Answer)

    A bucket policy is a resource-based policy written in JSON that defines:
    Who can access the bucket
    What actions they can perform
    On which resources
    Under what conditions

Option 1️⃣ AWS Console
Go to S3 → Bucket → Permissions
Open Bucket policy
Paste your custom JSON policy
Save
```
