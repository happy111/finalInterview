```python
def rev(i):
    s = list(i)
    l = 0
    r = len(s)-1
    while l < r:
        s[l],s[r] = s[r],s[l]
        l += 1
        r -= 1
    return ''.join(s)


def reverse_str(str1):
    res = []
    out = ""
    for i in str1:
        if i == " ":
            res.append(out)
            out=""
        else:
            out += i
    res.append(out)
    put = ""
    for i in res:
        s = rev(i)
        if put == "":
            put += s
        else:
            put += " " + s
    return put

str1 = "i love Python Programing"
result = reverse_str(str1)
print(result)
```

```python

from collections import defaultdict

# Build the hierarchy
hierarchy = defaultdict(list)
hierarchy["A"] = ["B", "C", "D"]
hierarchy["B"] = ["P"]
hierarchy["C"] = ["Q", "R"]
hierarchy["R"] = ["T", "U"]
hierarchy["P"] = ["X"]

def is_reportee(manager, reportee):
    if manager == reportee:
        return True  # optional, depending on if a person is their own reportee
    stack = [manager]
    
    while stack:
        current = stack.pop()
        for child in hierarchy.get(current, []):
            if child == reportee:
                return True
            stack.append(child)
    
    return False

# Example usage
manager = input("Enter manager name: ").strip()
reportee = input("Enter reportee name: ").strip()

print(is_reportee(manager, reportee))

```

```python
def dfs_recursive(adj_list, node, visited, result):
    visited.add(node)
    result.append(node)
    for neighbor in adj_list.get(node, []):
        if isinstance(neighbor, tuple):
            neighbor = neighbor[0]
        if neighbor not in visited:
            dfs_recursive(adj_list, neighbor, visited, result)

def dfs_traversal(adj_list, start):
    visited = set()
    result = []
    dfs_recursive(adj_list, start, visited, result)
    return result

# Example graph
adj_list = {1: [2, 3, 4], 2: [1, 3], 3: [2, 1], 4: [1]}
start_node = int(input("Enter starting node for DFS: "))
res = dfs_traversal(adj_list, start_node)
print("DFS Traversal:", res)
```

```python
def has_cycle_dfs(adj_list):
    visited = set()
    def dfs(node, parent):
        visited.add(node)
        for neighbor in adj_list.get(node, []):
            if neighbor not in visited:
                if dfs(neighbor, node):
                    return True
            elif neighbor != parent:
                return True  # Found a back edge (cycle)
        return False

    for node in adj_list:
        print("vvvvvvvvvvvvvv",node)
        if node not in visited:
            if dfs(node, -1):
                return True
    return False

# Example graph with a cycle
adj_list = {
    1: [2, 3],
    2: [1, 4],
    3: [1, 4],
    4: [2, 3]
}

# Example graph without a cycle
# adj_list = {
#     1: [2],
#     2: [1, 3],
#     3: [2, 4],
#     4: [3]
# }

print("Cycle Detected" if has_cycle_dfs(adj_list) else "No Cycle Found")
```

```python
2. Cycle Detection Using Union-Find (Disjoint Set)
```

```python
def find(parent, node):
    if parent[node] != node:
        parent[node] = find(parent, parent[node])
    return parent[node]

def union(parent, rank, u, v):
    root_u = find(parent, u)
    root_v = find(parent, v)
    if root_u == root_v:
        return True  # Cycle detected
    if rank[root_u] < rank[root_v]:
        parent[root_u] = root_v
    elif rank[root_u] > rank[root_v]:
        parent[root_v] = root_u
    else:
        parent[root_v] = root_u
        rank[root_u] += 1
    return False

def has_cycle_union_find(edges, num_nodes):
    parent = {i: i for i in range(1, num_nodes + 1)}
    rank = {i: 0 for i in range(1, num_nodes + 1)}
    for u, v in edges:
        if union(parent, rank, u, v):
            return True
    return False

# List of edges
edges = [(1, 2), (2, 3), (3, 1)]  # cycle
# edges = [(1, 2), (2, 3), (3, 4)]  # no cycle

print("Cycle Detected" if has_cycle_union_find(edges, 4) else "No Cycle Found")
```

```python
# 🧠 Example 2: Count Smaller Elements on Right

def count_smaller(nums):
    n = len(nums)
    result = [0] * n

    for i in range(n - 2, -1, -1):   # move from right to left
        count = 0
        for j in range(i + 1, n):    # check all elements to the right
            print(j)
            if nums[j] < nums[i]:
                count += 1
        result[i] = count

    return result


print(count_smaller([5, 2, 6, 1]))  # Output: 2
```

```python
# Example 1: Count Pairs (A before B)
def solve(input):
    c = 0
    carryA = 0
    for i in input:
        if i == 'A':
            carryA += 1
        if i == 'B':
            c += carryA
    return c
input = "ABBA"
res = solve(input)
print(res)
```

```python
# Q6. Move all zeros to end
def move_zeros(nums):
    non_zero = [x for x in nums if x != 0]
    return non_zero + [0] * (len(nums) - len(non_zero))

nums = [0,2,3,4,0,1,24,55]
res = move_zeros(nums)
print(res)
```

```python
#Question 3: Find Second Highest Number
def solve(li):
    max_v = float('-inf')
    sec_v = float('-inf')
    for k in li:
        if max_v < k:
            sec_v = max_v
            max_v = k
        elif k > sec_v and max_v < sec_v:
            sec_v = k
    return sec_v

s = [1,22,2,12,21,33,3]
res = solve(s)
print(res)
```

```python
# Question 2: Find First Non-Repeating Character
# input =  "swiss"
# output =  "w"

from collections import Counter
def solve(input):
    freq = Counter(input)
    for k,v in freq.items():
        if v == 1:
            return k
            break

input =  "swiss"
res = solve(input)
print(res)
```

```python
import string
import random

class Codec:
    def __init__(self):
        self.url_map = {}
        self.base_url = "http://tinyurl.com/"

    def _generate_key(self):
        chars = string.ascii_letters + string.digits
        return ''.join(random.choice(chars) for _ in range(6))

    def encode(self, longUrl: str) -> str:
        key = self._generate_key()
        self.url_map[key] = longUrl
        return self.base_url + key

    def decode(self, shortUrl: str) -> str:
        key = shortUrl.split("/")[-1]
        return self.url_map.get(key)
```
