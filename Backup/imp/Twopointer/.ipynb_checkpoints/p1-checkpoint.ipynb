```python
def twoSum(numbers: 'List[int]', target: 'int') -> 'List[int]':
    left, right = 0, len(numbers) - 1

    while left < right:
        if numbers[left] + numbers[right] == target:
            return [left + 1, right + 1]
        if numbers[left] + numbers[right] < target:
            left += 1
        else:
            right -= 1

    return [0, 0]
           
numbers = [2, 7, 11, 15]
target = 9
res = twoSum(numbers,target)
print(res)
```

```python
def next_greater_element(nums):
    n = len(nums)
    result = [-1] * n
    stack = []

    for i in range(n):
        while stack and nums[i] > nums[stack[-1]]:
            index = stack.pop()
            result[index] = nums[i]
        stack.append(i)
    return result


# Example
print(next_greater_element([4, 5, 2, 25]))
```

```python
from collections import OrderedDict

class LRUCache:
    def __init__(self, capacity):
        self.cache = OrderedDict()
        self.cap = capacity

    def get(self, key):
        if key not in self.cache:
            return -1
        self.cache.move_to_end(key)
        return self.cache[key]

    def put(self, key, value):
        if key in self.cache:
            self.cache.move_to_end(key)
        self.cache[key] = value
        if len(self.cache) > self.cap:
            self.cache.popitem(last=False)
```

```python
# ---------------- RUNNING EXAMPLE ----------------

cache = LRUCache(2)   # capacity = 2

cache.put(1, 10)
print(cache.cache)    # OrderedDict([(1, 10)])

cache.put(2, 20)
print(cache.cache)    # OrderedDict([(1, 10), (2, 20)])

# print(cache.get(1))   # 10
# print(cache.cache)    # OrderedDict([(2, 20), (1, 10)])

# cache.put(3, 30)
# print(cache.cache)    # OrderedDict([(1, 10), (3, 30)])

# print(cache.get(2))   # -1
```

```python
# Two sum
def solve(nums,target):
    res = []
    temp = {}
    for s in range(len(nums)):
        k = target - nums[s]
        if k in temp:
            res.append([k,nums[s]])
        temp[nums[s]] = s
    return res
nums = [2, 7, 11, 15]
target = 9
res = solve(nums,target)
print(res)
```

```python
# Find the contiguous subarray with the largest sum.

def solve(nums):
    pf = [nums[0]]
    for k in range(1,len(nums)):
        pf.append(nums[k] + pf[k-1])
    m = 0
    for s in range(len(nums)):
        for e in range(s,len(nums)):
            if s == 0:
                csum = pf[e]
            else:
                csum = pf[e] - pf[s-1]
            m = max(m,csum)
    return m

nums = [-2, 1, -3, 4, -1, 2, 1, -5, 4]
sum = 6
res = solve(nums)
print(res)
```

```python
# Optimize  Imp
# Find the contiguous subarray with the largest sum.
def solve(nums):
    max_sum = curr_sum = nums[0]

    for i in range(1, len(nums)):
        curr_sum = max(nums[i], curr_sum + nums[i])
        max_sum = max(max_sum, curr_sum)

    return max_sum
nums = [-2, 1, -3, 4, -1, 2, 1, -5, 4]
res = solve(nums)
print(res)
```

```python
# Move all 0's to the end while maintaining the relative order of non-zero elements.
def solve(s):
    non_zero = [k for k in s if k != 0]
    return non_zero + [0] * (len(s) - len(non_zero))

s = [0,1,2,44,0,2,2]
res = solve(s)
print(res)
```

```python
# ◆ 5. Trapping Rain Water  Need to understabd
def trap(height):
    left, right = 0, len(height) - 1
    left_max, right_max = 0, 0
    water = 0

    while left < right:
        if height[left] < height[right]:
            if height[left] >= left_max:
                left_max = height[left]
            else:
                water += left_max - height[left]
            left += 1
        else:
            if height[right] >= right_max:
                right_max = height[right]
            else:
                water += right_max - height[right]
            right -= 1

    return water
height = [0,1,0,2,1,0,1,3,2,1,2,1]
print(trap(height))
```

```python
# Imp

def longestPalindrome(s):
    if not s:
        return ""
    start = end = 0
    def expand(left, right):
        while left >= 0 and right < len(s) and s[left] == s[right]:
            left -= 1
            right += 1
        return left + 1, right - 1

    for i in range(len(s)):
        l1, r1 = expand(i, i)       # odd length
        l2, r2 = expand(i, i + 1)   # even length

        if r1 - l1 > end - start:
            start, end = l1, r1
        if r2 - l2 > end - start:
            start, end = l2, r2

    return s[start:end + 1]
s = "babad"
res = longestPalindrome(s)
print(res)
```

```python
# ◆ 4. Longest Common Prefix
def solve(strs):
    if not strs:
        return ""
    min_len = min(strs,key=len)
    res = ""
    for i in range(len(min_len)):
        for k in strs:
            if min_len[i] != k[i]:
                return res
        res += min_len[i]
    return res


strs = ["flower", "flow", "flight"]
res = solve(strs)
print(res)
```

```python
# 5. Roman to Integer
def romanToInt(s):
    roman = {
        'I': 1,
        'V': 5,
        'X': 10,
        'L': 50,
        'C': 100,
        'D': 500,
        'M': 1000
    }

    total = 0

    for i in range(len(s)):
        if i + 1 < len(s) and roman[s[i]] < roman[s[i + 1]]:
            total -= roman[s[i]]
        else:
            total += roman[s[i]]

    return total
print(romanToInt("V"))
```
