```python
# Missing Number
def solve(s):
    n = max(s)
    actual_sum = n* (n + 1) // 2
    ex = 0
    for i in s:
        ex += i
    return actual_sum - ex
s = [1,2,3,5]
res = solve(s)
print(res)
```

```python
# Convert Lower to Higher
def solve(st):
    ans = ""
    for i in st:
        ch = ord(i) - 32
        ans = chr(ch) + ans 
    return ans


st = input("Enter a string")
res = solve(st)
print(res)
```

```python
#sqrt
def solve(num):
    left  = 0
    right  = num
    ans = 0
    while left <= right:
        mid = left + (right - left) // 2
        if mid * mid == num:
            return mid
        elif mid * mid < num:
            ans = mid
            left = mid + 1
        else:
            right = mid - 1
    return ans
num = int(input("Enter a number"))
res = solve(num)
print(res)
```

```python
#272 = 2
def solve(num):
    while num > 9:
        ans = 0
        while num > 0:
            rem = num % 10
            ans += rem
            num = num // 10
        num = ans
    return num
num = int(input("Enter a number"))
res = solve(num)
print(res)
```

```python
# Decimal to binary
def solve(num):
    ans = ""
    while num > 0:
        rem = num % 2
        ans = str(rem) + ans
        num = num //2
    return ans

num = int(input("Enter a number"))
res = solve(num)
print(res)
```

```python
def find_maxarray(arr):
    result = []
    max_sum = 0
    for i in range(len(arr)):
        for j in range(i+1,len(arr)+1):
            result.append(arr[i:j])
            max_sum = max(max_sum,sum(arr[i:j]))
    return max_sum


arr = [2,1,6,4,-2]
result = find_maxarray(arr)
print("Sum of subarray",result)
```

```python
def convertChar(str):
    result = ""
    for k in str:
        if k.isupper():
            result += chr(ord(k)+32)
        else:
            result += chr(ord(k)-32)
    return result
            
result = convertChar("Hello")
print(result)
```

```python
# Check pallindrome
def check_pallindrome(st):
    start = 0
    end = len(st) - 1
    while start < end:
        if st[start] != st[end]:
            return False
        start += 1
        end -= 1
    else:
        return True



st = "madam1"
result = check_pallindrome(st)
print(result)
```

```python
# Decimal to Binary
def convertBinary(n):
    result = ""
    while n > 0:
        rem = n % 2;
        result =  str(rem) + result
        n = n//2
    return result

n = 25
result = convertBinary(n)
print(result)
```

```python
# Addition of decimal number
def sumOfDecimal(num, num1):
    s = ''
    carry = 0

    while num > 0 or num1 > 0 or carry:
        r1 = num % 10 if num > 0 else 0
        r2 = num1 % 10 if num1 > 0 else 0

        num = num // 10 if num > 0 else 0
        num1 = num1 // 10 if num1 > 0 else 0

        sum_digits = r1 + r2 + carry
        carry = sum_digits // 10
        s = str(sum_digits % 10) + s

    return s

num = int(input("Enter a number: "))
num1 = int(input("Enter a number: "))
result = sumOfDecimal(num, num1)
print(result)
```
