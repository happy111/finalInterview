```python
# Bubble sort
def solve(li):
    n = len(li)
    for i in range(n):
        for j in range(i+1,n):
            if li[i] < li[j]:
                li[i],li[j] = li[j],li[i]
    return li
li = [5,6,1,88,2]
res = solve(li)
print(res)
```

```python
# Selection sort
# Selection sort
def solve(li):
    n = len(li)
    for i in range(n):
        index = i
        for j in range(i+1,n):
            if li[index] > li[j]:
                index = j
        li[i],li[index] = li[index],li[i]
    return li
li = [3,4,1,3,11]
res = solve(li)
print(res)
```

```python
def solve(l1, l2):
    len_l1 = len(l1)
    len_l2 = len(l2)
    ans = []
    i = 0
    j = 0
    while i < len_l1 and j < len_l2:
        if l1[i] < l2[j]:
            ans.append(l1[i])
            i += 1
        else:
            ans.append(l2[j])
            j += 1
    while i < len_l1:
        ans.append(l1[i])
        i += 1
    while j < len_l2:
        ans.append(l2[j])
        j += 1
    return ans

# l1 = [2, 3, 4, 5, 10]
# l2 = [11, 12, 23]
# res = solve(l1, l2)
# print(res)
```

```python
def merge_sort(arr):
    if len(arr) <= 1:
        return arr
    mid = len(arr) // 2
    left = merge_sort(arr[:mid])
    right = merge_sort(arr[mid:])
    return solve(left, right)  # reusing your `solve()` function

arr = [5, 2, 10, 3, 12, 4, 11, 23]
sorted_arr = merge_sort(arr)
print(sorted_arr)
```
