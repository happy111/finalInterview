```python
# 9️⃣ Trapping Rain Water
def solve(height):
    l, r = 0, len(height)-1
    left_max = right_max = 0
    water = 0
    while l < r:
        if height[l] < height[r]:
            left_max = max(left_max, height[l])
            water += left_max - height[l]
            l += 1
        else:
            right_max = max(right_max, height[r])
            water += right_max - height[r]
            r -= 1
    return water

height = [4, 2, 0, 3, 2, 5]
res = solve(height)
print(res)
```

```python
from collections import Counter
def solve(st):
    s = Counter(st)
    res = ""
    for k,v in s.items():
        res += k+str(v)
    return res
    

st = "aabcccccaaa"
res = solve(st)
print(res)
```

```python
def compress_string(st):
    res = []
    count = 1
    s = st[0]
    for k in range(1,len(st)):
        if s == st[k]:
            count += 1
        else:
            res.append(s+str(count))
            count = 1
            s = st[k]
    res.append(s+str(count))
    return ''.join(res)

print(compress_string("aabcccccaaa"))
```

```python
def sort_sentence(sentence):
    words = sentence.split()
    sorted_arr = [""] * len(words)
    for w in words:
        # extract last character (digit)
        pos = int(w[-1]) - 1   # zero-based index
        print(pos)
        # remove the digit from the word
        sorted_arr[pos] = w[:-1]

    return " ".join(sorted_arr)


# Example
s = "is2 sentence4 This1 a3"
print(sort_sentence(s))  
```

```python
def clock_wise(s, n):
    if not s:
        return s
    n = n % len(s)   # handle n > length
    return s[-n:] + s[:-n]

# print(clockwise_rotate_n("amazon", 2))  # onamaz
# print(clockwise_rotate_n("amazon", 3))  # zonama



# Example
s1 = "amazon"
result = clock_wise(s1,1)
print(result)
```

```python
def anticlockwise_rotate_n(s, n):
    if not s:
        return s
    n = n % len(s)   # handles n > length
    return s[n:] + s[:n]

print(anticlockwise_rotate_n("amazon", 2))  # azonam
print(anticlockwise_rotate_n("amazon", 3))  # zonama
```

```python
def defang_ip_addr(address):
    ans = ""
    for ch in address:
        if ch == '.':
            ans += "[.]"
        else:
            ans += ch
    return ans
print(defang_ip_addr("1.1.1.1"))
```

```python
# sorted string
def solve(s):
    s = list(s)
    for i in range(len(s)):
        for j in range(i,len(s)):
            if s[i] < s[j]:
                s[i],s[j] = s[j],s[i]
    return ''.join(s)

s = "dsfsa"
res = solve(s)
print(res)
```

```python
def reverse(li, start, end):
    while start < end:
        li[start], li[end] = li[end], li[start]
        start += 1
        end -= 1
def solve(li, k):
    n = len(li)
    k %= n
    reverse(li, 0, k - 1)
    reverse(li, k, n - 1)
    reverse(li, 0, n - 1)
    return li
k = 2
li = [1,2,3,4,5]
res = solve(li,k)
print(res)
```
