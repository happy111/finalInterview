```python
def singleNumber(nums):
    result = 0
    for num in nums:
        result ^= num
    return result
nums = [2, 2, 1]
print(singleNumber(nums))
```

```python
def gcd(a, b):
    if b == 0:
        return a
    return gcd(b, a % b)
print(gcd(48, 18))
```
