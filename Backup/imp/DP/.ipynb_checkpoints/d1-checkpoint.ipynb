```python
# Fibo
def fibo(num):
    a, b = 0, 1
    if num <= 0:
        return
    print(a, end=" ")
    if num == 1:
        return
    print(b, end=" ")
    for _ in range(2, num):
        c = a + b
        print(c, end=" ")
        a, b = b, c
# print first 5 Fibonacci numbers
fibo(5)
```

```python
def fibo_recursive(n, a=0, b=1):
    if n <= 0:
        return
    print(a, end=" ")
    fibo_recursive(n-1, b, a+b)

# print first 5 Fibonacci numbers
res = fibo_recursive(5)
```

```python
# Dynamic programming
def fibo_memo(n, memo=None):
    if memo is None:
        memo = {}
    if n in memo:
        return memo[n]
    if n <= 1:
        return n
    memo[n] = fibo_memo(n-1, memo) + fibo_memo(n-2, memo)
    return memo[n]


print(fibo_memo(6))  # 55
```

```python
def fibo_tab(n):
    if n <= 1:
        return n
    dp = [0] * (n + 1)
    dp[0] = 0
    dp[1] = 1
    for i in range(2, n + 1):
        dp[i] = dp[i-1] + dp[i-2]
    return dp[n]

print(fibo_tab(10))  # 55
```

```python
Climbing Stairs
***************
Problem:
You are climbing stairs.
Each time you can climb 1 step or 2 steps.
How many distinct ways can you climb n steps?
ways(n) = ways(n-1) + ways(n-2)
```

```python
✅ 1. Brute Force Recursion (Naive)
```

```python
def climb(n):
    if n <= 2:
        return n
    return climb(n-1) + climb(n-2)
print(climb(5))
```

```python
# ✅ 2. Top-Down DP (Memoization)
```

```python
def climb(n,memo=None):
    if memo is None:
        memo = {}
    if n is memo:
        return memo[n]
    if n <= 2:
        return n
    memo[n] =  climb(n-1,memo) + climb(n-2,memo)
    return memo[n]
print(climb(5))
```

```python
# ✅ 2. Buttom-Down DP (Memoization)

def climb(n):
    if n <= 2:
        return n
    dp = [0] * (n+1)
    dp[1] = 1
    dp[2] = 2
    dp[n] =  climb(n-1) + climb(n-2)
    return dp[n]
print(climb(5))
```

```python
🐸 Count Number of Hops Problem

A frog can jump 1 step, 2 steps, or 3 steps at a time.

Find how many ways the frog can reach the top of n stairs.

This is a Dynamic Programming (DP) problem.
```

```python
✅ 1. Brute Force Recursion (Very slow)
```

```python
def count_hops(n):
    if n < 0:
        return 0
    if n == 0:
        return 1  # 1 way to stay at start

    return count_hops(n-1) + count_hops(n-2) + count_hops(n-3)
print(count_hops(4))

✔ Complexity
Time: O(3ⁿ) (exponential)
Space: O(n) recursion depth
❌ Very slow for large n.
```

```python
✅ 2. Top-Down DP (Memoization)
```

```python
def count_hops_memo(n, memo=None):
    if memo is None:
        memo = {}
    if n in memo:
        return memo[n]
    if n < 0:
        return 0
    if n == 0:
        return 1
    memo[n] = (count_hops_memo(n-1, memo) +
               count_hops_memo(n-2, memo) +
               count_hops_memo(n-3, memo))
    return memo[n]
print(count_hops(4))
```

```python
# Space optimized

def count_hops_opt(n):
    if n == 0:
        return 1
    if n == 1:
        return 1
    if n == 2:
        return 2  # (1+1), (2), (1+1) duplicates removed

    a, b, c = 1, 1, 2  # ways to reach step 0,1,2
    for _ in range(3, n + 1):
        a, b, c = b, c, a + b + c

    return c
```

```python
# 1. Fibonacci Number
def fibo(num,dp):
    if num < 2:
        dp[num] = num
        return num
    if num in dp:
        return dp[num]
    dp[num] = fibo(num-1,dp) + fibo(num-2,dp)
    return dp[num]
    
dp = {}
res = fibo(2,dp)
print(res)
```

```python
# 2. Climbing Stairs
def climb(n):
    dp = [0]*(n+1)
    dp[0]=dp[1]=1
    for i in range(2,n+1):
        dp[i]=dp[i-1]+dp[i-2]
    return dp[n]

res = climb(4)
print(res)
```

```python
def rev(l,r):
    while l < r:
        li[l],li[r] = li[r],li[l]
        l += 1
        r -= 1

def solve(li,k):
    k = k % len(li)
    rev(0,len(li)-1)
    rev(0,k-1)
    rev(k,len(li)-1)
    return li
li = [1,2,3,4,5]
k = 2
res = solve(li,k)
print(res)
```

```python
# Min Cost Climbing Stairs

def minCostClimbingStairs(cost):
    n = len(cost)
    dp = [0] * n
    dp[0] = cost[0]
    dp[1] = cost[1]
    
    for i in range(2, n):
        dp[i] = cost[i] + min(dp[i-1], dp[i-2])
    print(dp)
    return min(dp[n-1], dp[n-2])

cost = [10, 15, 20,30]
res = minCostClimbingStairs(cost)
print(res)
```

```python
def minCostClimbingStairs(cost):
    prev2 = cost[0]
    prev1 = cost[1]
    
    for i in range(2, len(cost)):
        curr = cost[i] + min(prev1, prev2)
        prev2 = prev1
        prev1 = curr
    
    return min(prev1, prev2)

cost = [10, 15, 20,30]
res = minCostClimbingStairs(cost)
print(res)
```

```python
# Tribonacci
🔢 What is Tribonacci?

In Tribonacci, each number is the sum of the previous 3 numbers.
```

```python
def tribonacci(n):
    if n == 0:
        return 0
    if n == 1 or n == 2:
        return 1

    dp = [0] * (n + 1)
    dp[0] = 0
    dp[1] = 1
    dp[2] = 1

    for i in range(3, n + 1):
        dp[i] = dp[i-1] + dp[i-2] + dp[i-3]

    return dp[n]

res = tribonacci(5)
print(res)
```
