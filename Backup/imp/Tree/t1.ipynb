```python
from queue import Queue

class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None

class ConstructTree:
    def constructNode(self, data):
        q = Queue()
        root = Node(data)
        q.put(root)
        while not q.empty():
            temp = q.get()
            leftchild = int(input(f"Enter left child of {temp.data} (-1 for no child): "))
            if leftchild != -1:
                temp.left = Node(leftchild)
                q.put(temp.left)
            rightchild = int(input(f"Enter right child of {temp.data} (-1 for no child): "))
            if rightchild != -1:
                temp.right = Node(rightchild)
                q.put(temp.right)
        return root

# Example usage:
if __name__ == "__main__":
    ct = ConstructTree()
    root = ct.constructNode(int(input("Enter root node: ")))
    # Simple BFS to check tree construction
    q = Queue()
    q.put(root)
    print("\nLevel Order Traversal of Constructed Tree:")
    while not q.empty():
        node = q.get()
        print(node.data, end=" ")
        if node.left:
            q.put(node.left)
        if node.right:
            q.put(node.right)
```

```python
from queue import Queue

class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None
class TreeInfo:
    def __init__(self, height, diameter):
        self.height = height
        self.diameter = diameter
class BinaryTree:
    def __init__(self):
        self.index = 0  # keeps track of position in list

    def build_tree(self, nodes):
        if self.index >= len(nodes) or nodes[self.index] == -1:
            self.index += 1
            return None
        
        root = Node(nodes[self.index])
        self.index += 1
        root.left = self.build_tree(nodes)
        root.right = self.build_tree(nodes)
        return root

    def preorder(self, root):
        if not root:
            return
        print(root.data, end=" ")
        self.preorder(root.left)
        self.preorder(root.right)

    def inorder(self, root):
        if not root:
            return

        self.inorder(root.left)
        print(root.data, end=" ")
        self.inorder(root.right)

    def postorder(self, root):
        if not root:
            return

        self.postorder(root.left)
        self.postorder(root.right)
        print(root.data, end=" ")

    def countNode(self,root):
        if not root:
            return 0
        leftNodes = self.countNode(root.left)
        rightNodes = self.countNode(root.right)
        return leftNodes + rightNodes + 1
    
    def sumNode(self,root):
        if not root:
            return 0
        leftSum = self.sumNode(root.left)
        rightSum = self.sumNode(root.right)
        return leftSum + rightSum+root.data

    def heightOfTree(self,root):
        if not root:
            return 0
        leftHeight = self.heightOfTree(root.left)
        rightHeight = self.heightOfTree(root.right)

        myheight = max(leftHeight,rightHeight) + 1
        return myheight

    def diameter1(self,root):
        if not root:
            return 0
        diam1 = self.diameter(root.left);
        diam2 = self.diameter(root.right);
        diam3 = self.heightOfTree(root.left) + self.heightOfTree(root.right) + 1
        return max(diam3,max(diam1,diam2))
    
    def diameter(self, root):
        if root is None:
            return TreeInfo(0, 0)
        # Get TreeInfo for left and right subtrees
        left_info = self.diameter(root.left)
        right_info = self.diameter(root.right)

        # Calculate height for current node
        my_height = max(left_info.height, right_info.height) + 1

        # Diameters: left, right, and through current node
        diam1 = left_info.diameter
        diam2 = right_info.diameter
        diam3 = left_info.height + right_info.height + 1

        # Best diameter
        my_diameter = max(diam1, diam2, diam3)
        return TreeInfo(my_height, my_diameter)
        
    def levelorder(self, root):
        if not root:
            return
        q = Queue()
        q.put(root)
        q.put(None)  # Marker for end of level
        while not q.empty():
            node = q.get()
            if node is None:
                print()  # Newline for level separation
                if not q.empty():
                    q.put(None)
            else:
                print(node.data, end=" ")
                if node.left:
                    q.put(node.left)
                if node.right:
                    q.put(node.right)

# Example usage
nodes = [1, 2, 4, -1, -1, 5, -1, -1, 3, -1, 6, -1, -1]
tree = BinaryTree()
root = tree.build_tree(nodes)

# print("Preorder Traversal:",end="\n")
# tree.preorder(root)
# print() 

# print("inorder Traversal:",end="\n")
# tree.inorder(root)
# print() 
# print("postorder Traversal:")
# tree.postorder(root)
# print()

print("Level order Traversal")
tree.levelorder(root)
print()

# print("Count of nodes")
# print(tree.countNode(root))
# print()


# print("Sum of nodes")
# print(tree.sumNode(root))
# print()

# print("Height of tree")
# print(tree.heightOfTree(root))
# print()


# print("Diameter of tree")
# # Number of nodes in the longest path between any 2 nodes
# result = tree.diameter(root)
# print()
# print("Height:", result.height)
# print("Diameter:", result.diameter)
```
