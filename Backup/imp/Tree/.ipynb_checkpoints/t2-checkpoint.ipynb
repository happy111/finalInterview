```python
from queue import Queue

class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None

class BinaryTree:
    def __init__(self):
        self.index = -1  # start from -1 for correct increment

    def build_tree(self, nodes):
        self.index += 1
        if self.index >= len(nodes) or nodes[self.index] == -1:
            return None
        root = Node(nodes[self.index])
        root.left = self.build_tree(nodes)
        root.right = self.build_tree(nodes)
        return root
    def is_identical(self,root1, root2):
        # if both trees are empty
        if root1 is None and root2 is None:
            return True
        # if one is empty and other is not
        if root1 is None or root2 is None:
            return False
        # check data and recurse for children
        return (root1.data == root2.data and
                self.is_identical(root1.left, root2.left) and
                self.is_identical(root1.right, root2.right))
    def levelorder(self, root):
        if not root:
            return
        q = Queue()
        q.put(root)
        q.put(None)
        while not q.empty():
            node = q.get()
            if node is None:
                print()
                if not q.empty():
                    q.put(None)
            else:
                print(node.data, end=" ")
                if node.left:
                    q.put(node.left)
                if node.right:
                    q.put(node.right)

    

# --------------------
# Create & print both trees
# --------------------
n1 = [1, 2, 4, -1, -1, 5, -1, -1, 3, -1, 6, -1, -1]
n2 = [1, 2, 4, -1, -1, 5, -1, -1, 3, -1, 6, -1, -1]

tree1 = BinaryTree()
root1 = tree1.build_tree(n1)
tree1.levelorder(root1)

print("-------------")

tree2 = BinaryTree()
root2 = tree2.build_tree(n2)
tree2.levelorder(root2)
tree2.is_identical(root1,root2)
```

```python
from queue import Queue

class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None


class BinaryTree:
    def __init__(self):
        self.index = -1

    def build_tree(self, nodes):
        self.index += 1
        if self.index >= len(nodes) or nodes[self.index] == -1:
            return None

        root = Node(nodes[self.index])
        root.left = self.build_tree(nodes)
        root.right = self.build_tree(nodes)
        return root

    def levelorder(self, root):
        if not root:
            return
        q = Queue()
        q.put(root)
        q.put(None)

        while not q.empty():
            node = q.get()
            if node is None:
                print()
                if not q.empty():
                    q.put(None)
            else:
                print(node.data, end=" ")
                if node.left:
                    q.put(node.left)
                if node.right:
                    q.put(node.right)

    def mirror(self, root):
        if root is None:
            return None
        # swap the children
        root.left, root.right = root.right, root.left
        # recursively mirror the subtrees
        self.mirror(root.left)
        self.mirror(root.right)
        return root


# Example Usage
tree = BinaryTree()
nodes = [1, 2, 4, -1, -1, 5, -1, -1, 3, 6, -1, -1, 7, -1, -1]

root = tree.build_tree(nodes)

print("Original Tree (Level Order):")
tree.levelorder(root)

print("\n\nMirror Tree (Level Order):")
tree.mirror(root)
tree.levelorder(root)
```

```python
# Check for balance TREE
  -1 <= height(L)-height(R) <= 1
```

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None
class BinaryTree:
    def __init__(self):
        self.index = -1
    def build_tree(self, nodes):
        self.index += 1
        if self.index >= len(nodes) or nodes[self.index] == -1:
            return None
        root = Node(nodes[self.index])
        root.left = self.build_tree(nodes)
        root.right = self.build_tree(nodes)
        return root
def is_balanced(root):
    """
    Returns (isBalanced, height)
    """
    if root is None:
        return True, 0
    left_balanced, left_height = is_balanced(root.left)
    right_balanced, right_height = is_balanced(root.right)
    # Check balance condition for current node
    balanced = (left_balanced and right_balanced and
                abs(left_height - right_height) <= 1)
    height = 1 + max(left_height, right_height)
    return balanced, height


# Example usage
tree = BinaryTree()
nodes = [1, 2, 4, -1, -1, 5, -1, -1, 3, -1, 6, -1, -1]
root = tree.build_tree(nodes)

balanced, height = is_balanced(root)
print("Is tree balanced?", balanced)
print("Height of tree:", height)
```
