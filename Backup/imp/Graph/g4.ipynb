```python
1) Directed graph — DFS + recursion stack
```

```python
from typing import Dict, List, Set

def dfs_directed(u: int, graph: Dict[int, List[int]], visited: Set[int], on_stack: Set[int]) -> bool:
    visited.add(u)
    on_stack.add(u)
    for v in graph.get(u, []):
        if v not in visited:
            if dfs_directed(v, graph, visited, on_stack):
                return True
        elif v in on_stack:
            # found a back edge → cycle detected
            return True
    on_stack.remove(u)
    return False
def has_cycle_directed(graph: Dict[int, List[int]]) -> bool:
    visited = set()
    on_stack = set()

    for node in graph.keys():
        if node not in visited:
            if dfs_directed(node, graph, visited, on_stack):
                return True
    return False

# Example
g1 = {1: [2], 2: [3], 3: [4], 4: []}        # acyclic
g2 = {1: [2], 2: [3], 3: [1]}               # cycle

print("directed g1:", has_cycle_directed(g1))  # False
print("directed g2:", has_cycle_directed(g2))  # True
```

```python
# 2) Undirected graph — DFS + parent check
# Undirected graph cycle detection (adjacency list)

from typing import Dict, List, Set
def dfs_undirected(u: int, parent: int, graph: Dict[int, List[int]], visited: Set[int]) -> bool:
    visited.add(u)
    for v in graph.get(u, []):
        if v == parent:
            continue
        if v not in visited:
            if dfs_undirected(v, u, graph, visited):
                return True
        else:
            # visited neighbor that's not parent -> cycle detected
            return True
    return False


def has_cycle_undirected(graph: Dict[int, List[int]]) -> bool:
    visited = set()
    for node in graph.keys():
        if node not in visited:
            if dfs_undirected(node, -1, graph, visited):
                return True
    return False


# Example
ug1 = {1: [2], 2: [1, 3], 3: [2]}            # no cycle (a path)
ug2 = {1: [2, 3], 2: [1, 3], 3: [1, 2]}      # triangle -> cycle

print("undirected ug1:", has_cycle_undirected(ug1))  # False
print("undirected ug2:", has_cycle_undirected(ug2))  # True

```
