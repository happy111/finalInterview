```python
from collections import defaultdict, deque

def resolve_tasks(tasks):
    # Step 1: Initialize graph and in-degree map
    graph = defaultdict(list)
    in_degree = defaultdict(int)
    
    # Ensure every task is in the in_degree map
    for task in tasks:
        in_degree[task['id']] = 0

    # Step 2: Build graph and update in-degrees
    for task in tasks:
        for dep in task['depends_on']:
            graph[dep].append(task['id'])
            in_degree[task['id']] += 1

    # Step 3: Topological sort using Kahn's algorithm
    queue = deque([task_id for task_id in in_degree if in_degree[task_id] == 0])
    execution_order = []

    while queue:
        current = queue.popleft()
        execution_order.append(current)
        for neighbor in graph[current]:
            in_degree[neighbor] -= 1
            if in_degree[neighbor] == 0:
                queue.append(neighbor)

    # Step 4: Check for cycles
    if len(execution_order) != len(in_degree):
        raise Exception("Cyclic dependency detected")

    return {"execution_order": execution_order}

# Example input
input_data = {
    "tasks": [
        {"id": "A", "depends_on": ["B", "C"]},
        {"id": "B", "depends_on": ["C"]},
        {"id": "C", "depends_on": []},
        {"id": "D", "depends_on": ["A"]}
    ]
}

# Run the function and print result
output = resolve_tasks(input_data["tasks"])
print(output)
```

```python
# Topological sort
# ****************
# DFS Based Topological sort
# **************************

def topological_sort_dfs(num_nodes, edges):
    graph = [[] for _ in range(num_nodes)]
    for u, v in edges:
        graph[u].append(v)
    visited = [False] * num_nodes
    stack = []
    def dfs(node):
        visited[node] = True
        for neighbor in graph[node]:
            if not visited[neighbor]:
                dfs(neighbor)
        stack.append(node)
    for i in range(num_nodes):
        if not visited[i]:
            dfs(i)

    return stack[::-1]  # Reverse the stack for correct order

# Example usage:
edges = [(5, 2), (5, 0), (4, 0), (4, 1), (2, 3), (3, 1)]
num_nodes = 6
print(topological_sort_dfs(num_nodes, edges))
```

```python
from queue import Queue,deque

def bfsTraversal(adj_list, start):
    visited = set()
    queue = Queue()
    traversal_order = []

    queue.put(start)

    while not queue.empty():
        node = queue.get()
        if node not in visited:
            visited.add(node)
            traversal_order.append(node)
            for neighbor in adj_list.get(node, []):
                # If it's a weighted graph: (neighbor, weight)
                if isinstance(neighbor, tuple):
                    neighbor = neighbor[0]
                if neighbor not in visited:
                    queue.put(neighbor)

    return traversal_order

# Sample input
start_node = int(input("Enter starting node for BFS: "))
s = {1: [2, 2, 4, 3], 2: [1, 1, 3, 3], 3: [2, 2, 1], 4: [1]}
res = bfsTraversal(s, start_node)
print("BFS Traversal:", res)
```

```python
def bfsTraversal(adj_list, start):
    visited = set()
    queue = deque([start])
    traversal_order = []
    while queue:
        node = queue.popleft()
        if node not in visited:
            visited.add(node)
            traversal_order.append(node)
            for neighbor in adj_list[node]:
                # If weighted graph: neighbor is a tuple (v, w)
                if isinstance(neighbor, tuple):
                    neighbor = neighbor[0]
                if neighbor not in visited:
                    queue.append(neighbor)
    return traversal_order
    

start_node = int(input("Enter starting node for BFS: "))
s =  {1: [2, 2, 4, 3], 2: [1, 1, 3, 3], 3: [2, 2, 1], 4: [1]}
res = bfsTraversal(s,start_node)
print(res)
```

```python
from collections import defaultdict
class Graph:
    def __init__(self, vertices):
        self.graph = defaultdict(list)
        self.V = vertices

    def add_edge(self, u, v):
        self.graph[u].append(v)

    def topological_sort_util(self, v, visited, stack):
        visited[v] = True
        for neighbour in self.graph[v]:
            if not visited[neighbour]:
                self.topological_sort_util(neighbour, visited, stack)
        stack.insert(0, v)

    def topological_sort(self):
        visited = [False] * self.V
        stack = []

        for i in range(self.V):
            if not visited[i]:
                self.topological_sort_util(i, visited, stack)

        print("Topological Sort:", stack)

# Example Usage:
g = Graph(6)
g.add_edge(5, 2)
g.add_edge(5, 0)
g.add_edge(4, 0)
g.add_edge(4, 1)
g.add_edge(2, 3)
g.add_edge(3, 1)

g.topological_sort()
```

```python
from queue import Queue

class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None
class TreeInfo:
    def __init__(self, height, diameter):
        self.height = height
        self.diameter = diameter
class BinaryTree:
    def __init__(self):
        self.index = 0  # keeps track of position in list

    def build_tree(self, nodes):
        if self.index >= len(nodes) or nodes[self.index] == -1:
            self.index += 1
            return None
        
        root = Node(nodes[self.index])
        self.index += 1
        root.left = self.build_tree(nodes)
        root.right = self.build_tree(nodes)
        return root

    def preorder(self, root):
        if not root:
            return
        print(root.data, end=" ")
        self.preorder(root.left)
        self.preorder(root.right)

    def inorder(self, root):
        if not root:
            return

        self.inorder(root.left)
        print(root.data, end=" ")
        self.inorder(root.right)

    def postorder(self, root):
        if not root:
            return

        self.postorder(root.left)
        self.postorder(root.right)
        print(root.data, end=" ")

    def countNode(self,root):
        if not root:
            return 0
        leftNodes = self.countNode(root.left)
        rightNodes = self.countNode(root.right)
        return leftNodes + rightNodes + 1
    
    def sumNode(self,root):
        if not root:
            return 0
        leftSum = self.sumNode(root.left)
        rightSum = self.sumNode(root.right)
        return leftSum + rightSum+root.data

    def heightOfTree(self,root):
        if not root:
            return 0
        leftHeight = self.heightOfTree(root.left)
        rightHeight = self.heightOfTree(root.right)

        myheight = max(leftHeight,rightHeight) + 1
        return myheight

    def diameter1(self,root):
        if not root:
            return 0
        diam1 = self.diameter(root.left);
        diam2 = self.diameter(root.right);
        diam3 = self.heightOfTree(root.left) + self.heightOfTree(root.right) + 1
        return max(diam3,max(diam1,diam2))
    
    def diameter(self, root):
        if root is None:
            return TreeInfo(0, 0)
        # Get TreeInfo for left and right subtrees
        left_info = self.diameter(root.left)
        right_info = self.diameter(root.right)

        # Calculate height for current node
        my_height = max(left_info.height, right_info.height) + 1

        # Diameters: left, right, and through current node
        diam1 = left_info.diameter
        diam2 = right_info.diameter
        diam3 = left_info.height + right_info.height + 1

        # Best diameter
        my_diameter = max(diam1, diam2, diam3)
        return TreeInfo(my_height, my_diameter)
        
    def levelorder(self, root):
        if not root:
            return
        q = Queue()
        q.put(root)
        q.put(None)  # Marker for end of level
        while not q.empty():
            node = q.get()
            if node is None:
                print()  # Newline for level separation
                if not q.empty():
                    q.put(None)
            else:
                print(node.data, end=" ")
                if node.left:
                    q.put(node.left)
                if node.right:
                    q.put(node.right)

# Example usage
nodes = [1, 2, 4, -1, -1, 5, -1, -1, 3, -1, 6, -1, -1]
tree = BinaryTree()
root = tree.build_tree(nodes)

print("Preorder Traversal:",end="\n")
tree.preorder(root)
print() 

print("inorder Traversal:",end="\n")
tree.inorder(root)
print() 
print("postorder Traversal:")
tree.postorder(root)
print()

print("Level order Traversal")
tree.levelorder(root)
print()

print("Count of nodes")
print(tree.countNode(root))
print()


print("Sum of nodes")
print(tree.sumNode(root))
print()

print("Height of tree")
print(tree.heightOfTree(root))
print()


print("Diameter of tree")
# Number of nodes in the longest path between any 2 nodes
result = tree.diameter(root)
print()
print("Height:", result.height)
print("Diameter:", result.diameter)
```

```python
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

def isIdentical(t1, t2):
    if not t1 and not t2:
        return True
    if not t1 or not t2:
        return False
    return (t1.val == t2.val and 
            isIdentical(t1.left, t2.left) and 
            isIdentical(t1.right, t2.right))

def isSubtree(root, subRoot):
    if not subRoot:
        return True
    if not root:
        return False
    if isIdentical(root, subRoot):
        return True
    return isSubtree(root.left, subRoot) or isSubtree(root.right, subRoot)

# Example Usage:
T = TreeNode(3)
T.left = TreeNode(4, TreeNode(1), TreeNode(2, TreeNode(0)))
T.right = TreeNode(5)



S = TreeNode(4, TreeNode(1), TreeNode(2))

print(isSubtree(T, S))  # Output: True
```
