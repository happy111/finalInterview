```python
# 1️⃣ BFS Traversal
from queue import Queue
def bfs(graph,start):
    visited = set()
    q = Queue()
    q.put(start)
    traversal = []
    while not q.empty():
        node = q.get()
        if node not in visited:
            visited.add(node)
            traversal.append(node)
            for n in graph.get(node,[]):
                if n not in visited:
                    q.put(n)
    return traversal
graph = {
    "A": ["B", "C"],
    "B": ["D"],
    "C": ["E"],
    "D": [],
    "E": []
}
res = bfs(graph,"A")
print(res)
```

```python
# 1️⃣ DFS Traversal
from queue import Queue
def dfs(graph,node,visited,res):
    visited.add(node)
    res.append(node)
    for nei in graph[node]:
        if nei not in visited:
            dfs(graph,nei,visited,res)

graph = {
    "A": ["B", "C"],
    "B": ["D"],
    "C": ["E"],
    "D": [],
    "E": []
}
res = []
dfs(graph, "A", set(), res)
print(res)
```

```python
def dfs(graph, visited, node, path, tar):
    # Mark node as visited and add to path
    visited.add(node)
    path.append(node)
    # If destination reached, print path
    if node == tar:
        print(" -> ".join(path))
    else:
        for neighbour in graph.get(node, []):
            if neighbour not in visited:
                dfs(graph, visited, neighbour, path, tar)

    # Backtrack
    path.pop()
    visited.remove(node)


# Example graph
graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [],
    'E': ['F'],
    'F': []
}

src = 'A'
tar = 'F'
visited = set()
path = []

print(f"All paths from {src} to {tar}:")
dfs(graph, visited, src, path, tar)
```
