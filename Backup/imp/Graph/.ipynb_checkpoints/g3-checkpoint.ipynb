```python
class Edge:
    def __init__(self,src,dest):
        self.src = src
        self.dest = dest
def create_graph(graph):
    for i in range(v):
        graph[i] = []
    graph[0].append(Edge("A","B"))
    graph[0].append(Edge("A","C"))

    graph[1].append(Edge("B","D"))
    graph[1].append(Edge("B","E"))

    graph[2].append(Edge("C","F"))
    graph[4].append(Edge("E","F"))
    
def print_graph(graph):
    temp = ["A","B","C","D","E","F"]
    
    for i in range(len(graph)):
        print(f"Vertex {temp[i]}:", end=" ")
        for edge in graph[i]:
            print(edge.dest, end=" ")
        print()

if __name__ == "__main__":
    v = 6
    graph = [None] * v
    create_graph(graph)
    print_graph(graph)
```

```python
from queue import Queue
def bfsTraversal(adj_list,start_node):
    visited = set()
    q = Queue()
    traversal_order = []
    q.put(start_node)
    while not q.empty():
        node = q.get()
        if node not in visited:
            visited.add(node)
            traversal_order.append(node)
            for neighbour in adj_list.get(node,[]):
                if isinstance(neighbour,tuple):
                    neighbour = neighbor[0]
                if neighbour not in visited:
                    q.put(neighbour)
    return traversal_order

# Sample input
start_node = int(input("Enter starting node for BFS: "))
s = {1: [2, 2, 4, 3], 2: [1, 1, 3, 3], 3: [2, 2, 1], 4: [1]}
res = bfsTraversal(s, start_node)
print("BFS Traversal:", res)
```

```python
from collections import deque

def bfs_traverse(all_graphs, visited):
    for start_node in all_graphs:
        if start_node not in visited:
            queue = deque([start_node])
            while queue:
                node = queue.popleft()
                if node not in visited:
                    visited.add(node)
                    for neighbour in all_graphs.get(node, []):
                        queue.append(neighbour)
            print()  # new line after each graph traversal


# -------- Graph 1 --------
graph1 = {
    'A': ['B', 'C'],
    'B': ['D'],
    'C': [],
    'D': []
}

# -------- Graph 2 --------
graph2 = {
    'P': ['Q'],
    'Q': ['R'],
    'R': []
}

# -------- Main Method --------
if __name__ == "__main__":
    all_graphs = {**graph1, **graph2}   # merge both graphs
    visited = set()                     # define visited in main
    print("Traversal of all graphs:")
    bfs_traverse(all_graphs, visited)

    print("\nVisited nodes:", visited)
```

```python
# 2. Depth First Search (Recursion)
```

```python
def dfs(graph, node, visited):
    if node not in visited:
        print(node, end=' ')
        visited.add(node)
        for neighbour in graph.get(node, []):
            dfs(graph, neighbour, visited)


# -------- Graph --------
graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [],
    'E': ['F'],
    'F': []
}

# -------- Main --------
if __name__ == "__main__":
    visited = set()
    print("DFS Traversal:")
    dfs(graph, 'A', visited)
    print("\nVisited nodes:", visited)
```

```python
### All paths from source to target

For given src & tar, print all paths that exists from src to target

```

```python
def print_graph(graph,src,tar,visited):
    if src not in visited:
        if src == tar:
            return True
        else:
            print(src,end=" ")
            visited.add(src)
            for neighbour in graph.get(src,[]):
                if print_graph(graph,neighbour,tar,visited):
                    return True
                    

graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [],
    'E': ['F'],
    'F': []
}
visited = set()
src = 'A'
tar = 'E'
print_graph(graph,src,tar,visited)
```

```python
#  100
def print_graph(graph, start, dest, visited):
    if start not in visited:
        print(start, end=' ')
        visited.add(start)

        if start == dest:   # stop when destination is found
            return True     # signal that destination was found

        for neighbour in graph.get(start, []):
            if print_graph(graph, neighbour, dest, visited):
                return True  # stop further recursion if found

    return False  # not found in this path


graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [],
    'E': ['F'],
    'F': []
}

visited = set()
start = 'A'
dest = 'D'

print_graph(graph, start, dest, visited)
```

```python
def dfs(graph, visited, node, path, tar):
    # Mark node as visited and add to path
    visited.add(node)
    path.append(node)
    # If destination reached, print path
    if node == tar:
        print(" -> ".join(path))
    else:
        for neighbour in graph.get(node, []):
            if neighbour not in visited:
                dfs(graph, visited, neighbour, path, tar)

    # Backtrack
    path.pop()
    visited.remove(node)


# Example graph
graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [],
    'E': ['F'],
    'F': []
}

src = 'A'
tar = 'F'
visited = set()
path = []

print(f"All paths from {src} to {tar}:")
dfs(graph, visited, src, path, tar)
```
