```python
def solve():
    u = int(input("Enter number of vertices:")) 
    v = int(input("Enter number of edges:"))

    adj_matrix = [[0 for _ in range(u)]for _ in range(u)]

    print(adj_matrix)

    print("Enter the edges (in the form: u,v):")
    for i in range(v):
        a,b = map(int,input().split())
        adj_matrix[a][b] = 1
        adj_matrix[b][a] = 1
    
    print("\nAdjacency Matrix:")
    for row in adj_matrix:
        print(row)
# Run the function
solve()
```

```python
# Adjacency matrix
# undirected graph
# Weight

def solve():
    u = int(input("Enter number of vertices"))
    v = int(input("Enter number of edges"))
    adj_matrix = [[0 for i in range(u)] for _ in range(u)]
    print(adj_matrix)
    print("Enter edges (in the form: u v):")
    for i in range(v):
        a,b,w = map(int,input().split())
        adj_matrix[a][b] = w
        adj_matrix[b][a] = w 

    print("\nAdjacency Matrix:")
    for row in adj_matrix:
        print(row)
    

solve()
```

```python
from queue import Queue

def bfs(graph,start):
    visited = set()
    q = Queue()
    ans = []
    q.put(start)
    while not q.empty():
        node = q.get()
        if node not in visited:
            visited.add(node)
            ans.append(node)
            for neighbour in graph.get(node,[]):
                if neighbour not in visited:
                    q.put(neighbour)
    return ans

graph = {
    "A": ["B", "C"],
    "B": ["D"],
    "C": ["E"],
    "D": [],
    "E": []
}
res = bfs(graph,"A")
print(res)
```

```python
from typing import Dict, List, Set

def dfs_direct(v,graph,visited,onstack):
    visited.add(v)
    onstack.add(v)
    for u in graph.get(v,[]):
        if u not in visited:
            if dfs_direct(u,graph,visited,onstack):
                return True
        elif u in onstack:
            return True
    onstack.remove(v)
    return False
                
            
def has_cycle_directed(graph:Dict[int,List[int]])->bool:
    visited = set()
    onstack= set()
    for v in graph.keys():
        if v not in visited:
            if dfs_direct(v,graph,visited,onstack):
                return True
    return False

# Example
g1 = {1: [2], 2: [3], 3: [4], 4: []}        # acyclic
g2 = {1: [2], 2: [3], 3: [1]}               # cycle

print("directed g1:", has_cycle_directed(g1))  # False
print("directed g2:", has_cycle_directed(g2))  # True
```
