```python
from collections import defaultdict, deque

def topological_sort(input_data):
    # Step 1: Build graph and in-degree map
    graph = defaultdict(list)
    in_degree = {}

    for task in input_data['tasks']:
        task_id = task['id']
        in_degree[task_id] = 0  # Initialize in-degree

    for task in input_data['tasks']:
        task_id = task['id']
        for dep in task['depends_on']:
            graph[dep].append(task_id)
            in_degree[task_id] += 1
    print("ddddddddddddddddddddddd", in_degree)
    
    # Step 2: Initialize queue with nodes having in-degree 0
    queue = deque([node for node in in_degree if in_degree[node] == 0])

    topo_order = []

    while queue:
        node = queue.popleft()
        topo_order.append(node)

        for neighbor in graph[node]:
            in_degree[neighbor] -= 1
            if in_degree[neighbor] == 0:
                queue.append(neighbor)

    if len(topo_order) == len(in_degree):
        return topo_order
    else:
        raise ValueError("Cycle detected in the graph!")

# Example Usage:
input_data = {
    "tasks": [
        {"id": "A", "depends_on": ["B", "C"]},
        {"id": "B", "depends_on": ["C"]},
        {"id": "C", "depends_on": []},
        {"id": "D", "depends_on": ["A"]}
    ]
}

output = topological_sort(input_data)
print("OUTPUT =", output)
```

```python
def dfs(node, adj, visited, stack):
    visited[node] = True
    for neighbor in adj[node]:
        if not visited[neighbor]:
            dfs(neighbor, adj, visited, stack)
    stack.append(node)

def topo_sort(v, adj):
    visited = [False] * v
    stack = []
    for i in range(v):
        if not visited[i]:
            dfs(i, adj, visited, stack)
    ans = []
    while stack:
        ans.append(stack.pop())
    return ans

# Method to call topological sort
def main():
    # Example: Graph with 6 vertices (0 to 5)
    v = 6
    adj = [[] for _ in range(v)]
    
    # Adding edges (Directed graph)
    adj[5].append(0)
    adj[0].append(2)
    adj[4].append(2)
    adj[1].append(1)
    adj[2].append(3)
    adj[3].append(1)

    print("dddddddddddddddddddd",adj)
    result = topo_sort(v, adj)
    print("Topological Sort Order:", result)

if __name__ == "__main__":
    main()
```
