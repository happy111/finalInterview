```python
def longest_unique_substring(s):
    temp = {}
    start = 0
    max_len = 0
    large_str = ""
    for end in range(len(s)):
        if s[end] in temp and temp[s[end]] >= start:
            start = temp[s[end]] + 1
        temp[s[end]] = end
        cur = end - start + 1
        if cur > max_len:
            max_len = cur
            large_str = s[start : end+ 1]

    return large_str


# Example usage:
s = "asfdsafsadewrwrdddsdssf"
print(longest_unique_substring(s))
```

```python
# Give a code for upload file in s3
import boto3
from botocore.exceptions import NoCredentialsError, ClientError

def upload_to_s3(file_name, bucket_name, object_name=None, region="ap-south-1"):
    """
    Upload a file to an S3 bucket

    :param file_name: Local file path
    :param bucket_name: S3 bucket name
    :param object_name: S3 file name (if None, same as local file name)
    :param region: AWS region
    """
    if object_name is None:
        object_name = file_name

    try:
        s3_client = boto3.client("s3", region_name=region)
        s3_client.upload_file(file_name, bucket_name, object_name)
        print(f"✔ File uploaded successfully to s3://{bucket_name}/{object_name}")
        return True

    except FileNotFoundError:
        print("❌ File not found!")
        return False
    except NoCredentialsError:
        print("❌ AWS credentials not found!")
        return False
    except ClientError as e:
        print(f"❌ AWS Error: {e}")
        return False


# Example usage
upload_to_s3(
    file_name="sample.txt",
    bucket_name="your-bucket-name",
    object_name="uploads/sample.txt"
)
```

```python
def solve(input,parent=None,sep='.'):
    if not input:
        return 
    temp = {}
    for key,val in input.items():
        pkey = parent + sep + key if parent else key
        if isinstance(val,dict):
            temp.update(solve(val,pkey,sep=sep))
        else:
            temp[pkey] = val
    return temp

input  = {'a': 1, 'b': {'c': 2, 'd': {'e': 3}}}
res = solve(input)
print(res)
```

```python
def solve(res):
    if not res:
        return ""
    min_str = min(res,key=len)
    output = ""
    for i in range(len(min_str)):
        for word in res:
            if word[i] != min_str[i]:
                return output
        output += min_str[i]
    return output


# Test cases
res1 = ['car', 'carbon', 'vehicle']
res2 = ['carbon', 'carbonmonoxide']
res12 = ["abcd", "abcc", "abcde"]

print(solve(res1))   # Output: ""
print(solve(res2))   # Output: "carbon"
print(solve(res12))  # Output: "abc"
```

```python
def solve(s,n):
    min_val = 1 * n
    max_val = 9 * n
    if s < min_val or s > max_val:
        return "Not allowed"
    res = []


    for i in range(n):
        for d in range(9,0,-1):
            if s - d >= (n-i-1) * 1 and s-d <= (n-i-1) * 9:
                res.append(d)
                s -= d
                break
    print(res)


print(solve(20, 3))  # Output: "992"
```

```python
def solve(logs):
    result = {
        "GET": {},
        "POST": {}
    }
    for entry in logs:
        parts = entry.split()
        method = parts[2]     # GET or POST
        status = parts[-1]    # 200, 401, etc.

        # Initialize dict keys if not present
        if status not in result[method]:
            result[method][status] = 0
        # Increment count
        result[method][status] += 1

    return result


# Input Logs
logs = [
    "2025-07-18 12:00:01 GET /index.html 200",
    "2025-07-18 12:00:02 POST /login 401",
    "2025-07-18 12:00:03 GET /dashboard 200",
    "2025-07-18 12:00:04 GET /index.html 200",
    "2025-07-18 12:00:05 POST /login 200",
]

res = solve(logs)
print(res)
```

```python
import re

def solve(inp):
    # Extract tokens like MR, CT, FP
    tokens = re.findall(r"[A-Za-z0-9]+", inp)
    # Remove duplicates while keeping order
    seen = set()
    unique = []
    for t in tokens:
        if t not in seen:
            seen.add(t)
            unique.append(t)
    
    return ", ".join(unique)


inp = "'MR', ['CT', 'MR'], 'FP', 'MR'"
print(solve(inp))
```
