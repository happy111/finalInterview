```python
def clockwise_shift(lst, k):
    n = len(lst)
    k = k % n  # handle k > n
    return lst[-k:] + lst[:-k]

# Example
arr = "amazon"
k = 1
shifted = clockwise_shift(arr, k)
print(shifted)
```

```python
def anticlockwise_shift(lst, k):
    n = len(lst)
    k = k % n  # handle k > n
    return lst[k:] + lst[:k]

st = "amazon"
res = anticlockwise_shift(st,2)
print(res)
```

```python
def sort_char(s):
    s = list(s)
    n = len(s)
    for i in range(n):
        for j in range(i+1,n):
            if s[i] > s[j]:
                s[i],s[j] = s[j],s[i]
    return ''.join(s)

s = "eabcabd"
res = sort_char(s)
print(res)
```

```python
def solve(A,k):
    ps=[A[0]]
    for i in range(1,len(A)):
        ps.append(A[i]+ps[i-1])
       
    for s in range(len(A)):
        for e in range(s,len(A)):
            sum = 0
            if s == 0:
                sum += ps[e]
            else:
                sum += ps[e] - ps[s-1]
            if sum == k:
                return A[s:e+1]

# subarray sum is k
A = [1,3,15,10,20,3,23]
k = 33
result = solve(A,k)
print(result)
```

```python
def solve(s,n):
    min_v = 1 * n
    max_v = 9 * n
    res = []
    if s > max_v or s < min_v:
        return "It is not possible"

    for i in range(n):
        for d in range(9,0,-1):
            if s - d <= (n-i-1) * 9 and s - d >= (n-i-1) * 1:
                res.append(d)
                s -= d
                break
    return res


print(solve(2, 3)) 
```

```python
def merge_meetings(m):
    start,end = m[0]
    res = []
    for k in m[1:]:
        if k[0] <= end:
            end = max(k[1],end)
        else:
            res.append([start,end])
            start,end = k[0],k[1]
    res.append([start,end])
    return res

meetings = [
    (9, 10),
    (9.5, 11),
    (10, 12),
    (13, 14),
    (13.5, 15)
]

print(merge_meetings(meetings))
```

```python
def longest_unique_substring(s):
    start = 0
    temp = {}
    max_len = 0
    max_str = ""
    for end in range(len(s)):
        if s[end] in temp and temp[s[end]] >= start:
            start = temp[s[end]] + 1
        temp[s[end]] = end
        cur = end - start+1
        if cur > max_len:
            max_len = cur
            max_str = s[start:end+1]
    return max_str

s = "asfdsafsadewrwrdddsdssf"
print(longest_unique_substring(s))
```

```python
def is_in_chain(h,m,r):
    if m not in h or r not in h:
        return False
    stack = [m]
    while stack:
        cur = stack.pop()
        if cur == r:
            return True
        stack.extend(h.get(cur,[]))
    return False



hierarchy = {
    'A': ['B', 'C', 'D'],
    'B': ['P'],
    'C': ['Q', 'R'],
    'R': ['T', 'U'],
    'P': ['X']
}

# Example usage
manager_input = input("Enter manager name: ")
reportee_input = input("Enter reportee name: ")

result = is_in_chain(hierarchy, manager_input, reportee_input)
print("Related in chain?", result)
```

```python
Q1: What is the difference between a Process and a Thread? Explain with Python examples.

✔ Process
    A process is an independent unit of execution.
    It has its own memory space.
    
✔ Thread

  A thread runs inside a process.
  Multiple threads of the same process share memory, file descriptors, variables.
  Good for I/O-bound tasks.
      
# MULTI-THREADING EXAMPLE (best for I/O-bound tasks)
import threading
import time

def task():
    time.sleep(1)
    print("Thread executed")

threads = []
for _ in range(5):
    t = threading.Thread(target=task)
    t.start()
    threads.append(t)

for t in threads:
    t.join()
```

```python

# MULTI-PROCESSING EXAMPLE (best for CPU-bound tasks)
from multiprocessing import Process
import time

def task():
    time.sleep(1)
    print("Process executed")

processes = []
for _ in range(5):
    p = Process(target=task)
    p.start()
    processes.append(p)

for p in processes:
    p.join()
```

```python
| Process                     | Thread                    |
| --------------------------- | ------------------------- |
| Separate memory             | Shared memory             |
| Heavyweight                 | Lightweight               |
| Safe (no shared state bugs) | Can cause race conditions |
| Good for CPU-bound tasks    | Good for I/O-bound tasks  |
| Slower to start             | Fast to start             |
```
