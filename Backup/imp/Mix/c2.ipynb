```python
# How do you remove duplicates from a list of dictionaries?

def remove(dicts):
    seen = set()
    result = []
    for d in dicts:
        # Convert dict to a sorted tuple of key-value pairs
        tup = tuple(sorted(d.items()))
        if tup not in seen:
            seen.add(tup)
            result.append(d)
    return result

data = [
    {'name': 'Alice', 'age': 30},
    {'name': 'Bob', 'age': 25},
    {'name': 'Alice', 'age': 30},  # duplicate
    {'name': 'Charlie', 'age': 35}
]
res = remove(data)
print(res)
```

```python
employees = [
    {'name': 'Alice', 'department': 'HR'},
    {'name': 'Bob', 'department': 'IT'},
    {'name': 'Charlie', 'department': 'HR'},
    {'name': 'Dave', 'department': 'Finance'}
]

temp = {}
for i in employees:
    s = i.get('department')
    if s in temp:
        temp[s].append(i)
    else:
        temp[s] = [i]
print(temp)
```

```python

from collections import defaultdict

def group_by_field(data, field):
    grouped = defaultdict(list)
    for item in data:
        key = item.get(field)
        grouped[key].append(item)
    return dict(grouped)

employees = [
    {'name': 'Alice', 'department': 'HR'},
    {'name': 'Bob', 'department': 'IT'},
    {'name': 'Charlie', 'department': 'HR'},
    {'name': 'Dave', 'department': 'Finance'}
]

grouped = group_by_field(employees, 'department')
print(grouped)
```

```python
 # How to update all values in a nested dictionary where keys match a given pattern?

def update_nested_values(d, target_key, new_value):
    for key, value in d.items():
        if key == target_key:
            d[key] = new_value
        elif isinstance(value, dict):
            update_nested_values(value, target_key, new_value)

data = {
    'id': 1,
    'info': {
        'status': 'pending',
        'details': {
            'status': 'in_progress',
            'comments': 'Work started'
        }
    }
}

update_nested_values(data, 'status', 'complete')
print(data)
```

```python
# Given a nested dict, count total number of leaf key-value pairs.
def count_leaf_pairs(nested_dict):
    count = 0
    for key,value in nested_dict.items():
        if isinstance(value,dict):
            count+= count_leaf_pairs(value)
        else:
            count +=1
    return count
nested_dict = {
    'a': 1,
    'b': {
        'c': 2,
        'd': {
            'e': 3,
            'f': 4
        }
    },
    'g': 5
}

print(count_leaf_pairs(nested_dict)) 
```

```python
# Write a recursive function to print all values in a nested dictionary.

def print_values(d):
    for key, value in d.items():
        if isinstance(value, dict):
            print_values(value)  # recursive call
        else:
            print(value)

# Example usage:
nested_dict = {
    'a': 1,
    'b': {
        'c': 2,
        'd': {
            'e': 3,
            'f': 4
        }
    },
    'g': 5
}

print_values(nested_dict)
```

```python
Iterable:-
  1. An object that can return an iterator.
  2. Lists, tuples, sets, dictionaries, strings.
  3. It has the __iter__() method.
my_list = [1, 2, 3]  # This is an iterable
for item in my_list:
    print(item)
 


Iterator:-
  1. An object that represents a stream of data
  2.  It has both __iter__() and __next__() methods.
  3.  Use next() to manually get elements one by one.

my_list = [1, 2, 3]
it = iter(my_list)  # This is an iterator

print(next(it))  # 1
print(next(it))  # 2
print(next(it))  # 3
```

```python
import threading

def display(n):
    for i in range(n):
        print("Current Thread is ", i , threading.current_thread().name)

t1 = threading.Thread(target=display,args=(4,))
t2 = threading.Thread(target=display,args=(4,))


t1.start()
t2.start()

t1.join()
t2.join()
```

```python
def solve(str):
    n = len(str)
    for i in range(n-1,-1,-1):
        yield str[i]

str = "umesh"
res = solve(str)
try:
    while True:
        print(next(res), end = "")
except Exception as e:
    print(e)
```
