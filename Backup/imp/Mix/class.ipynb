```python
Q Given an integer sorted array A and an integer K, Find any pair(i,j)
such that
  A[i] + A[j] = k i != j

A = [-5,-2,1,8,10,12,15]
```

```python
# Brute Force
def pair_sum(A):
    n = len(A)
    li = []
    for i in range(n):
        for j in range(i+1,n):
            if A[i] + A[j] == 11:
                li.append([A[i],A[j]])
    return li

A = [-5,-2,1,8,10,12,15]
result = pair_sum(A)
print(result)
```

```python
# Using Binary search
def pair_sum_binary(A,k):
    li = []
    for i in A:
        j = k - i
        left = 0
        right = len(A) - 1
        while left <= right:
            mid = left + (right-left) // 2
            if A[mid] == j:
                li.append([i,j])
                break
            elif A[mid] < j:
                left = mid+1
            else:
                right = mid - 1
    return li
                


A = [-5,-2,1,8,10,12,15]
target = 11
result = pair_sum_binary(A,target)
print(result)
```

```python
def Two_pointer(A,k):
    i = 0
    j = len(A) - 1
    li = []
    while i < j:
        if A[i] + A[j] == k:
            li.append([i,j])
            break
        elif A[i] + A[j] > k:
            j = j - 1
        else:
            i = i + 1
    return li


A = [-5,-2,1,8,10,12,15]
target = 11
result = Two_pointer(A,target)
print(result)
```

```python
Find all pairs in a sorted array where sum is k.

A = [1,2,3,4,5,6,8]
k = 10
```

```python
def All_pairs(A,k):
    i = 0
    j = len(A) - 1
    count = 0
    while i < j:
        if A[i] + A[j] == k:
            count += 1
            i += 1
            j -= 1
        elif A[i] + A[j] > k:
            j -= 1
        else:
            i += 1
    return count

A = [1,2,3,4,5,6,8]
k = 10
result = All_pairs(A,k)
print(result)
```

```python
Given a sorted integer array A and an integer k
Find any pair (i,j) such that

A[j] - A[i] = k

i != j and k > 0
```

```python
# Brute Force
def difference_pair(A,k):
    for i in range(len(A)):
        for j in range(i+1,len(A)):
            if A[j] - A[i] == k:
                return [i,j]
    return [-1,-1]

A = [-5,-2,1,8,10,12,15]
k = 11
result = difference_pair(A,k)
print(result)
```

```python
# Binary search
def difference_pair(A,k):
    for i in range(len(A)):
        j = k + A[i]
        left = i+1
        right = len(A) - 1
        while left <= right:
            mid = left + (right-left) // 2
            if A[mid] == j:
                return [i,left]
            elif A[mid] > j:
                right = mid - 1
            else:
                left = mid + 1
    return [-1,-1]

A = [-5,-2,1,8,10,12,15]
k = 11
result = difference_pair(A,k)
print(result)
```

```python
# Two pointer
def difference_pair(A,k):
    i = 0
    j = 1
    while j < len(A):
        diff = A[j] - A[i]
        if diff == k:
            return [j,i]
        elif diff < k:
            j += 1
        else:
            i += 1
    return [-1]
        



A = [-5,-2,1,8,10,12,15]
k = 11
result = difference_pair(A,k)
print(result)
```

```python
Find if in the given array there exist a subarray summing to k

A = [1,3,15,10,20,3,23]

k = 33
```

```python
def prefix_sum(A,target):
    pf=[A[0]]
    for i in range(1,len(A)):
        pf.append(pf[i-1]+A[i])
    for i in range(len(A)):
        for j in range(i+1,len(A)):
            sum = 0
            if i == 0:
                sum += pf[j]
            else:
                sum += pf[j] - pf[i-1]
            if sum == target:
                return A[i:j+1]
    
A = [1,3,15,10,20,3,23]
target = 26
result = prefix_sum(A,target)
print(result)
```

```python
def subarray_sum_to_k(A,k):
    temp = {}
    start_index = 0
    csum = 0
    for i,num in enumerate(A):
        csum += num
        if csum == k:
            return True,A[start_index:i+1]
        elif csum-k in temp:
            return True, A[temp[csum-k]+1 : i+1]
        temp[csum] = i
        
A = [1, 3, 15, 10, 20, 3, 23]
k = 26
found, subarray = subarray_sum_to_k(A, k)
print(subarray)
```

```python
def solve(string): 
    freq = [0] * 26
    is_odd_present = False
    ans = 0
    for ch in string:
        freq[ord(ch) - ord('a')] += 1
    for count in freq:  # go through all 26 letters
        if count % 2 == 0:
            ans += count
        else:
            is_odd_present = True
            ans += count - 1
    if is_odd_present:
        ans += 1
    return ans

string = "madamda"
result = solve(string)
print(result)  # Output: 3
```
