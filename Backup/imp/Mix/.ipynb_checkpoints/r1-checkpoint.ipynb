```python
# Prime number

def check_prime(n):
    flag = 1
    for i in range(2,n//2):
        if n % i == 0:
            flag = 0
    if flag == 1:
        return "Prime number"
    else:
        return "Not prime number"
n = int(input("Enter a number"))
result = check_prime(n)
print(result)
```

```python
# Sort of the list

li = [4,6,8,9,34,45]
n = len(li)
for i in range(n-1):
    for j in range(n-1-i):
        if li[j] > li[j+1]:
            temp = li[j]
            li[j] = li[j+1]
            li[j+1] = temp
print(li)
```

```python
# Factorial
def fact(n):
    fact = 1
    while n > 0:
        fact = fact * n
        n = n -1
    return fact
        

print(fact(5))
```

```python
def is_valid_parenthese(str1):
    stack, pchar = [], {"(": ")", "{": "}", "[": "]"}
    for parenthese in str1:
        if parenthese in pchar:
            stack.append(parenthese)
        elif len(stack) == 0 or pchar[stack.pop()] != parenthese:
            return False
    return len(stack) == 0

print(is_valid_parenthese("(){}[]"))
```

```python
def list_sorting(li):
    for i in range(len(li)):
        for j in range(len(li)):
            if li[j]>li[i]:
                li[i],li[j]=li[j],li[i]
    return li
                

li = [4,6,1,8,11]
sort = list_sorting(li)
sort
```

```python
def sum1(a, b):
    def decorator(func):
        def wrapper(a,b):
            result = func(a,b)
            return result // 2
        return wrapper
    return decorator


@sum1(4, 6)  # Using sum1 decorator with parameters 4 and 6
def sum(a, b):
    return a + b

print(sum(4, 6))  # Output: 5
```

```python
from abc import ABC, abstractmethod

class Shape(ABC):
    @abstractmethod
    def area(self):
        pass

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius
    
    def area(self):
        return 3.14 * self.radius * self.radius

class Rectangle(Shape):
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height

# Create objects with valid arguments
circle_obj = Circle(4)
rectangle_obj = Rectangle(5, 6)

print("Circle area:", circle_obj.area())
print("Rectangle area:", rectangle_obj.area())
```

```python
def are_cyclic_strings(str1, str2):
    if len(str1) != len(str2):
        return False

    concatenated = str1 + str1
    print(concatenated)
    if str2 in concatenated:
        return True
    else:
        return False

# Test cases
string1 = "abcde"
string2 = "cdeab"
if are_cyclic_strings(string1, string2):
    print(f"'{string1}' and '{string2}' are cyclic strings.")
else:
    print(f"'{string1}' and '{string2}' are not cyclic strings.")
```

```python
# Find greatest common division

def greatest_division(num1,num2):
    smalest_value =num2 if num1 > num2 else num1
    greatest_value = 0
    for i in range(1,smalest_value+1):
        if num1 % i == 0 and num2 % i == 0:
            if greatest_value < i:
                greatest_value = i
    return greatest_value


num1 = int(input("Enter a number"))
num2 = int(input("Enter a number"))
result = greatest_division(num1,num2)
print(result)
```

```python
def find_consecutive(inp):
    inp.sort()
    arr = []
    arr_final = [inp[0]]
    for i in range(1,len(inp)):
        if inp[i] == inp[i-1]+1:
            arr_final.append(inp[i])
        else:
            if len(arr_final) > 1:
                arr.append(arr_final)
            arr_final = [inp[i]]
    return arr
 

inp = [1,112, 320, 230, 431, 233, 231,432,232, 412, 5982]
result = find_consecutive(inp)
print(result)
```

```python
import os

# Define the path using a raw string
path = r'D:\New\text.txt'

# Generator function to yield words from the file line by line
def read_file(file_path):
    with open(file_path, 'r') as file:
        for line in file:
            for word in line.split():
                yield word

# Create a generator
s = read_file(path)

# Use the generator to print each word
try:
    while True:
        print(next(s))
except StopIteration:
    pass
```

```python
class Singleton:
    __instance = None
    def __new__(cls):
        if cls.__instance is None:
            cls.__instance = super(Singleton, cls).__new__(cls)
        return cls.__instance

# Testing
obj1 = Singleton()
obj2 = Singleton()

print(obj1 is obj2)  # ✅ True — both point to the same instance
```

```python
Manager-reportee
A->B,C,D
B->P
C->Q,R
R->T,U
P->X
Write a programme to take a manager and reportee name as 
input and return true if the manager and reportee are related in chain 
else false
```

```python
# Define the manager-reportee relationships


def is_in_chain(hierarchy, manager, reportee):
    # Base case: manager not in hierarchy
    if manager not in hierarchy:
        return False
    # DFS stack
    stack = [manager]
    while stack:
        current = stack.pop()
        if current == reportee:
            return True
        stack.extend(hierarchy.get(current, []))
        print(stack)
    return False

hierarchy = {
    'A': ['B', 'C', 'D'],
    'B': ['P'],
    'C': ['Q', 'R'],
    'R': ['T', 'U'],
    'P': ['X']
}

# Example usage
manager_input = input("Enter manager name: ").strip()
reportee_input = input("Enter reportee name: ").strip()

result = is_in_chain(hierarchy, manager_input, reportee_input)
print("Related in chain?", result)
```

```python
We are making n stone piles! The first pile has n stones. 
If n is even, then all piles have an even number of stones.
If n is odd, all piles have an odd number of stones. 
Each pile must more stones than the previous pile but as few as possible. 
Write a Python program to find the number of stones in each pile.
```

```python
def stone_piles(n):
    piles = []
    current = n

    for _ in range(n):
        piles.append(current)
        current += 2  # minimal increase while keeping even/odd same

    return piles
res = stone_piles(4)
print(res)
```
