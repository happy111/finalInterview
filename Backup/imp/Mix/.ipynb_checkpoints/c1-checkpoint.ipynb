```python
Compiler:-

  1. Translates the entire source code into machine code at once.
  2. Produces an executable file (like .exe).
      Used in languages like C, C++, Java (Java uses both compiler & interpreter).
Interpreter

  1. Translates code line by line and executes it immediately.
  2. Does not create an executable file.
  3. Execution is slower because translation happens during runtime.

      Used in languages like Python,
```

```python
➡️ Python first converts this into bytecode (.pyc files)

2️⃣ Interpretation by Python Virtual Machine (PVM)

The PVM executes the bytecode line by line

This is why Python is called interpreted

📌 So the correct statement is:

Python is a bytecode-interpreted language
```

```python
#  Migration step for database

Database migration is the process of moving data,schema,objects and configuration
from one database to another.

    MySQL → PostgreSQL    SQL Server → AWS RDS
    On-prem → Cloud

2️⃣ Why do companies migrate databases?
    1. Cost reduction
    2. Scalability and performance

3️⃣ What are the main steps in database migration?

1. Requirement analysis
2. Schema assessment
3. Type mapping
4. Backup source DB
5. Schema conversion
6. Data migration
7. Functions/triggers migration
8. Validation
9. Post-migration monitoring
```

```python
4️⃣ What tools are used for MySQL → PostgreSQL migration?

   pgLoader (most popular) 
   AWS Database Migration Service (DMS)
   Custom Python scripts (SQLAlchemy, psycopg2)

5️⃣ What are common datatype issues when migrating MySQL → PostgreSQL?

Answer:
    TINYINT(1) → BOOLEAN
    AUTO_INCREMENT → SERIAL or GENERATED AS IDENTITY
    DATETIME → TIMESTAMP
    FLOAT → REAL, DOUBLE → DOUBLE PRECISION
    MySQL ENUM → PostgreSQL CHECK constraints
    MySQL JSON is not fully compatible with PostgreSQL JSONB

6️⃣ What challenges occur during database migration?

Answer:
    Data type incompatibility
    Syntax differences (LIMIT, JOINs, NULL handling)
    Stored procedure rewrite
    Triggers rewrite
    Foreign key dependency order
    Large dataset performance issues
    Downtime management
    Timezone issues

7️⃣ How do you minimize downtime during migration?

Answer:
Use CDC (Change Data Capture):

    AWS DMS CDC
    Debezium
    Kafka Connect
    pgLogical

    Steps:
    
    Full load first
    Capture incremental changes
    Sync the delta
    Freeze writes
    Cutover

8️⃣ How do you validate data after migration?

Answer:
    
    Compare row counts
    Sample data check
    Checksum comparison
    Run queries on both DBs and match results
    Test foreign keys, constraints, triggers
    Application functional testing

9️⃣ What is schema migration?

Answer:
    Schema migration means converting the entire database structure:
    
    Tables
    
    Columns
    
    Indexes
    
    Constraints
    
    Views
    
    Triggers

Schema migration is done before data migration.

1️⃣2️⃣ What is pgLoader?

Answer:
pgLoader is an open-source migration tool that automates:

Schema conversion

Data migration

Index creation

Constraints

Type mappings

pgloader mysql://user:pass@host/db postgres://user:pass@host/db
```

```python
OAUTH : 

 . It allows a third-party application to access a user’s data without sharing the user’s password.
 . ✅ OAuth does NOT authenticate users directly
   ✅ It authorizes access to resources

   Real-Life Example
        “Login with Google”

Step-by-step (Interview explanation)

    User clicks Login with Google
    App redirects user to Google Authorization Server
    User logs in & approves permissions
    Google sends authorization code to app
    App exchanges code for Access Token
    App calls API using token
    API validates token and returns data

3️⃣ Where OAuth is Used?
    ✅ Third-party login

OAuth handles authorization, JWT handles authentication
```

```python
What is RestAPI ?

 . REST API stands for Representational State Transfer Application Programming Interface
 . It is a set of architectural principles used to build scalable, stateless,
     and lightweight web services that communicate over HTTP.

| HTTP Method | Operation        | Example          |
| ----------- | ---------------- | ---------------- |
| GET         | Read             | GET /users       |
| POST        | Create           | POST /users      |
| PUT         | Update (full)    | PUT /users/10    |
| PATCH       | Update (partial) | PATCH /users/10  |
| DELETE      | Delete           | DELETE /users/10 |
```

```python
🔹 Types of Exception Handling

1️⃣ try–except

      try:
          a = int("abc")
      except ValueError:
         print("Invalid conversion")
          
2️⃣ Multiple except Blocks

try:
    x = int(input("Enter number: "))
    y = 10 / x
except ValueError:
    print("Please enter a number")
except ZeroDivisionError:
    print("Cannot divide by zero")
    
3️⃣ Single except for Multiple Exceptions
try:
    a = int("10")
    b = 10 / 0
except (ValueError, ZeroDivisionError) as e:
    print("Error:", e)
```

```python
Key Differences
Scope:

Decorators: Used at the function/method level within the same application.
Middleware: Used at the request/response level, typically in web applications to handle HTTP requests and responses.
Usage:

Decorators: Applied directly to functions or methods.
Middleware: Configured in the web framework settings and applied globally or to specific routes.
Functionality:

Decorators: Extend or modify the behavior of specific functions or methods.
Middleware: Provide a mechanism to process and modify requests and responses as they move through the application.
Typical Use Cases:

Decorators: Logging, access control, memoization.
Middleware: Authentication, logging, request modification, response compression.
```

```python
In Python, __str__ and __repr__ are special methods used to define how objects are represented as strings.
They are both used for string representation


__str__: For a user-friendly and readable string representation.
__repr__: For an unambiguous string representation useful for developers, often to recreate the object.
```

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __repr__(self):
        return f"Person('{self.name}', {self.age})"

person = Person("Alice", 30)
print(repr(person))  # Output: Person('Alice', 30)
```

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __str__(self):
        return f"Person named {self.name}, aged {self.age}"

person = Person("Alice", 30)
print(person)  # Output: Person named Alice, aged 30
```

```python
def display(file_name):
    with open(file_name,'r') as fr:
        s = fr.read()
        yield(s)

file_name = "demo.txt"
file_data = display(file_name)
print(file_data)
next(file_data)
```
