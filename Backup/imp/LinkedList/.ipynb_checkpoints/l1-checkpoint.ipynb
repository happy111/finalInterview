```python
class Node:
    def __init__(self,data):
        self.next = None
        self.data = data
        
```

```python
class linkedList:
    def __init__(self):
        self.head = None
    def reverse(self):
        prev = None
        fut = None
        temp = self.head
        while temp != None:
            fut = temp.next
            temp.next = prev
            prev = temp
            temp = fut
        self.head = prev
    def insertAtbeg(self,data):
        new_data = Node(data)
        new_data.next = self.head
        self.head= new_data
    def display(self):
        self.temp = self.head
        while self.temp:
            print(self.temp.data,end=" ")
            self.temp = self.temp.next

    def middleElement(self):
        fast = self.head
        slow = self.head
        while fast is not None and fast.next is not None:
            fast = fast.next.next
            slow = slow.next
        print("Middle Element "  , slow.data)

    #Detect loop (Floyd’s cycle)
    def detectLoop(self):
        fast = self.head
        slow = self.head
        while fast is not None and fast.next is not None:
            fast = fast.next.next
            slow = slow.next
            if slow == fast:
                return True
        return False
        print("Middle Element "  , slow.data)
```

```python
obj = linkedList()
obj.insertAtbeg(1)
obj.insertAtbeg(2)
obj.display()
```

```python
class Linekdlist:
    def __init__(self):
        self.head = None
 
    def deleteAtbeg(self):
        if self.head is None:
            print("List is empty, nothing to delete")
            return
        self.head = self.head.next

    def deleteAtEnd(self):
        if self.head is None:
            print("List is empty, nothing to delete")
            return
        # If only one node
        if self.head.next is None:
            self.head = None
            return
        temp = self.head
        # Traverse to the second-last node
        while temp.next.next:
            temp = temp.next
        # Remove last node
        temp.next = None

    def deleteAtBet(self,pos):
        if self.head is None:
            print("List is empty, nothing to delete")
            return
        if pos == 0:
            self.head = self.head.next
            return
        temp = self.head
        i = 0

        # Move to (pos-1)-th node
        while temp is not None and i < pos - 1:
            temp = temp.next
            i += 1

        # If position is out of range
        if temp is None or temp.next is None:
            print("Position out of range")
            return

        # Skip the node at given position
        temp.next = temp.next.next
        
    def insertAtbeg(self,data):
        new_data = Node(data)
        new_data.next = self.head
        self.head = new_data
    def insertAtbet(self,data,pos):
        new_data = Node(data)
        if pos == 0:
            new_data.next = self.head
            self.head = new_data
            return
        
        temp = self.head
        i = 0
        while temp.next and i < pos-1:
            temp = temp.next
            i += 1
     # If position is out of range
        if temp is None:
            print("Position out of range")
            return

        new_data.next = temp.next
        temp.next = new_data
                
    def insertAtEnd(self,data):
        new_data = Node(data)
        if self.head is None:
            new_data.next = self.head
            self.head = new_data
            return
        else:
            temp = self.head               # Use temporary pointer
            while temp.next:               # Traverse till last node
                temp = temp.next
            temp.next = new_data
    def display(self):
        temp = self.head
        while temp:
            print(temp.data,end=" ")
            temp = temp.next
```

```python
obj= Linekdlist()
obj.insertAtbeg(1)
obj.insertAtbeg(2)
obj.insertAtEnd(10)
obj.insertAtbet(12,0)
obj.insertAtbet(14,2)
obj.deleteAtbeg()
obj.deleteAtEnd()
obj.deleteAtBet(2)
obj.display()
```
