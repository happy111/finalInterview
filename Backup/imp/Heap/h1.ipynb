```python
def heapHeight(N, arr):
    if N == 1:
        return 1
    height = 0
    while N > 1:
        N //= 2
        height += 1
    
    return height
arr = [10, 20, 30, 40, 50, 60, 70]
N = len(arr)

print("Heap Height:", heapHeight(N, arr))
```

```python
import heapq
MOD = 1000000007
def nchoc(A, B):
    # Convert list B into a max-heap by negating values
    max_heap = [-x for x in B]
    heapq.heapify(max_heap)
    result = 0

    while A > 0 and max_heap:
        # Pop largest element (remember it's stored negative)
        r = -heapq.heappop(max_heap)

        result = (result + r) % MOD

        # Push half of r back as negative for max-heap
        heapq.heappush(max_heap, -(r // 2))

        A -= 1

    return result
print(nchoc(3, [6, 5]))  
```
