```python
Introduction to stack
*********************

insertion : push
delete    : pop

peek()  : know about peek()



LIFO

Operations: push(), pop(), peek(), isEmpty()

Behavior: Last-In-First-Out (LIFO) order

Whenever stack is empty. (Underflow)
```

```python
def reverseString(s):
    stack = []

    # Push all characters onto the stack
    for c in s:
        stack.append(c)

    # Pop characters back into the list
    for i in range(len(s)):
        s[i] = stack.pop()
```

```python
s = ['h', 'e', 'l', 'l', 'o']
reverseString(s)
print(s)
```

```python
def insert_at_bottom(st, x):
    st2 = []

    # Pop everything from st into st2
    while st:
        st2.append(st.pop())

    # Push the new element to the (now empty) original stack
    st.append(x)

    # Put everything back from st2 to st
    while st2:
        st.append(st2.pop())

    return st
stack = [1, 2, 3, 4]   # 4 is the top of the stack
x = 10

print(insert_at_bottom(stack, x))  
# Output: [1, 2, 3, 4, 10]
```

```python
def insert_at_bottom_recursive(st, x):
    # If stack is empty, just push x
    if not st:
        st.append(x)
        return

    # Pop the top element
    top = st.pop()
    # Recursively insert x at bottom
    insert_at_bottom_recursive(st, x)
    # Push the top element back
    st.append(top)
stack = [1, 2, 3, 4]  # 4 is top
insert_at_bottom_recursive(stack, 10)
print(stack)  # → [10, 1, 2, 3, 4]
```

```python
def make_beautiful(arr):
    stack = []
    for data in arr:
        if stack and ((stack[-1] >= 0 and data < 0) or (stack[-1] < 0 and data >= 0)):
            stack.pop()
        else:
            stack.append(data)
    return stack
arr = [4, -4, 1, -3, 1, -1]
print(make_beautiful(arr))
# Possible output: [1, -1] (depends on how many pairs cancel out)
```
