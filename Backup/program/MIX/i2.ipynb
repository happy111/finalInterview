```python
Why does binary search not work on a linked list?

Binary search works by dividing the search space into halves until the target element is found.

Binary search does not work on a linked list because: 
   .Linked list is not stored in a contiguous block of memory
   .Elements are accessed using node references, not by index
   . Binary search requires random access (O(1) index access)
    .In a linked list, accessing the middle element takes O(n) time
```

```python
Linear Search

    Checks elements one by one
    Works on sorted and unsorted data
    Time Complexity: O(n)

Binary Search

    Divides the search space into half
    Works only on sorted data
    Time Complexity: O(log n)
```

```python
🔹 Question 2: Why Binary Search is O(log n)?

Because in every step:
Search space is reduced to half
Number of steps = log₂(n)
```
