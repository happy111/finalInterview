```python
What is a window function in MySQL?
How is it different from GROUP BY?

✔ Mentioned row numbering
✔ Mentioned PARTITION BY
✔ Mentioned AVG / MAX
✔ Correctly said different from GROUP BY
```

```python
SELECT
    customer_id,
    order_date,
    amount,
    LAG(amount) OVER (
        PARTITION BY customer_id
        ORDER BY order_date
    ) AS previous_order_amount
FROM Orders;
```
