```python
1️⃣ Second Highest Salary per Department

s = Emp.objects.filter(department=OuterRef('department')).order_by('-salary').values('salary')[1:2]
Emp.objects.filter(salary=Subquery(s))

```

```python
OuterRef means:
“Use the value from the outer query’s current row inside the subquery.”
```

```python
2️⃣ Fetch Latest Order per Customer

S = Orders.objects.filter(customer_id=OuterRef('customer_id')).order_by('-order_date').values('id')[:1]

Orders.objects.filter(id=Subquery(s))
```

```python
3️⃣ Top 3 Salaries per Department (Window Function)

👉 annotate() is used to add extra calculated fields to each row in a queryset.

Employee.objects.annotate(
    rank=Window(
        expression=Rank(),
        partition_by=[F('dept')],
        order_by=F('salary').desc()
    )
).filter(rank__lte=3)
```
