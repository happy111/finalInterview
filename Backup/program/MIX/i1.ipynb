```python
What is the difference between a compiled language and an interpreted language?
Where does Python fit?

Compiled Language

    Source code is compiled at once into machine code
    Produces an executable file (e.g., .exe)
    Errors are mostly caught at compile time
    
    Generally faster execution
    Examples: C, C++, Go

Interpreted Language

    Code is executed line by line at runtime
    No separate executable file
    Errors are found during execution
    Generally slower than compiled languages
    Examples: Python, JavaScript, Ruby

Where does Python fit?

    Python is an interpreted language
    Python code is compiled into bytecode (.pyc)
    Bytecode is executed by the Python Virtual Machine (PVM)
```

```python
Mutable Data Types

Can be changed after creation
Memory address remains the same

Examples:

    list
    dict
    set

Immutable Data Types

Cannot be changed after creation
Any modification creates a new object
Examples:

    int
    float
    str
    tuple
    bool
    complex
```

```python
def solve(li):
    pos = 0
    for i in range(len(li)):
        if li[i] != 0:
            li[pos] = li[i]
            pos += 1
    for k in range(pos,len(li)):
        li[k] = 0
    return li


s = [0,0,2,3,1]
res = solve(s)
print(res)
```

```python
What is the Global Interpreter Lock (GIL)?
Why does Python have it?

What is GIL?

The Global Interpreter Lock (GIL) is a mutex in CPython that ensures only one thread
executes Python bytecode at a time, even on multi-core systems.

Why does Python have the GIL?

    To simplify memory management
    To make built-in data structures (list, dict, set) thread-safe
    To avoid race conditions and data inconsistency
    To improve performance for single-threaded execution
```

```python
What is memory management in Python?
How does Python handle garbage collection?

Memory Management in Python
Python handles memory management automatically.

There are two main areas of memory:

1. Stack Memory

    Stores local variables
    Stores function call frames
    Managed automatically when functions are called and returned

2. Heap Memory

    Stores objects and instances
    Managed by the Python Memory Manager
    Objects are accessed via references

Garbage Collection in Python

Python uses reference counting as its primary garbage collection mechanism.
Each object has a reference count
When reference count becomes zero, the object is deallocated automatically
```

```python
What is the difference between @staticmethod, @classmethod, and instance methods?

1️⃣ Instance Method

    Works with object (instance) data
    First parameter is self
    Can access and modify instance variables and class variables

2️⃣ Class Method

    Works with class-level data
    First parameter is cls
    Created using @classmethod
    Can modify class variables
    Called using class or object

3️⃣ Static Method

    Does not access instance or class data
    No self or cls
    Created using @staticmethod
    Used for utility/helper functions
```

```python
What is a Python metaclass?
Why and when would you use it?

A metaclass is a class of a class.

Just like a class defines the behavior of an object,

A metaclass defines the behavior of a class.

In Python:
    Classes are objects
    They are created using a metaclass
    Default metaclass is type

Why do we use Metaclasses?

Metaclasses are used to:
    Control class creation
    Add or modify class attributes automatically

When are Metaclasses used?

    ORM frameworks (like Django models)
    API schema generation
    Validation of class structure
    Singleton pattern (advanced use)

A metaclass is a class that controls how classes are created and behaves, just like a class controls objects.
```

```python
class MyMeta(type):
    def __new__(cls, name, bases, dct):
        dct['created_by_meta'] = True
        return super().__new__(cls, name, bases, dct)

class A(metaclass=MyMeta):
    pass

print(A.created_by_meta)  # True
```

```python
What is a stack and a queue?
Explain their operations and use cases.

Stack

Follows LIFO → Last In, First Out
Operations:
    push → insert element
    pop → remove top element
    peek/top → view top element

Use cases:

    Function call stack
    Undo / Redo operations
    Expression evaluation
    Backtracking (DFS)

Queue

Follows FIFO → First In, First Out

Operations:
    enqueue → insert elment
    dequeue → remove element

Use cases:

    Task scheduling
    Request handling
    Breadth-First Search (BFS)
    Producer–consumer problems
```

```python
What is a linked list?
What are its types and advantages over arrays?

A linked list is a linear data structure where elements are stored in nodes. 

Each node contains:

Data
Reference (pointer) to the next node

Nodes are not stored in contiguous memory locations.

Types of Linked List

    Singly Linked List – points to next node
    Doubly Linked List – points to next and previous nodes
    Circular Linked List – last node points to first

Advantages over Arrays

    Dynamic size
    Efficient insertion and deletion (O(1) if position is known)

Disadvantages

    No direct indexing (access is O(n))
    Extra memory for pointers
```

```python
Array

    Stored in contiguous (continuous) memory
    Supports direct access using index
    Access time is O(1)
    Insertion and deletion are costly (O(n)) because elements shift
    Fixed size

Linked List

    Stored in non-contiguous memory
    Each node contains data + pointer/reference
    No direct indexing → access time is O(n)
    Insertion and deletion are efficient (O(1) if position is known)
    Dynamic size
```

```python
What is a hash table?
How do Python dictionaries work internally?


A hash table is a data structure that stores data in key–value pairs using a hash function.

   . The hash function converts a key into a hash value
   . The hash value determines where the data is stored

Python dictionary is implemented using a hash table
Each key is passed through a hash function
The hash value determines the index in the table
Data is stored as (key, value) pairs
```

```python
What is recursion?
What are its advantages and disadvantages?

What is Recursion?

Recursion is a technique where a function calls itself to solve a problem by breaking it into smaller subproblems.

A recursive function has:

    Base condition (to stop recursion)
    Recursive call
    
Disadvantages of Recursion

    Uses more memory (function call stack)
    Can cause stack overflow if base condition is missing

Advantages of Recursion

    Simplifies complex problems
    Cleaner and more readable code
```

```python
What is the difference between recursion and iteration?
Which one is better and why?

Recursion

    A function calls itself
    Breaks a problem into smaller subproblems
    Uses call stack memory
    Code is often simpler and cleaner
    Risk of stack overflow if not handled properly

Iteration

    Uses loops (for, while)
    Repeats statements until a condition is met
    Uses less memory than recursion
    Generally faster than recursion


Iteration is better for performance and memory

Recursion is better when the problem is naturally recursive (trees, graphs)
```

```python
What is a binary search?
What is its time complexity and requirement?

Binary search is a search algorithm used to find an element in a sorted array by repeatedly dividing the search space into half.

How it Works

Find the middle element
Compare the target with the middle
If equal → element found
If smaller → search left half
If greater → search right half
Repeat until found or search space becomes empty

Time Complexity

Best case: O(1)

Average case: O(log n)

Worst case: O(log n)
```
