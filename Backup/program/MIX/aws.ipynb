```
# Disaster Handling in AWS (Disaster Recovery – DR)

  Disaster handling :
    Region outage 🌍
    Server crash 💥
    Database corruption 🗄️
    Human error

 AWS handles this using DR strategies + managed services.

1️⃣ Disaster Recovery Strategies
  🔹 1. Backup & Restore (Low cost)
        Regular backups
```

```
What is Disaster Recovery (DR)?

Disaster Recovery is the process of restoring applications,
data, and systems after a failure or disaster so that business can continue.
```

```
What is a Disaster?

A disaster can be:

    Server crash 💻
    Database failure 🗄️
    Data loss or corruption ❌
    Power/network outage ⚡
    Cloud region failure 🌍
    Human mistake (accidental delete) 😅
```

```
Why Disaster Recovery is Important?

    Avoid business downtime
    Prevent data loss
    Maintain customer trust
    Meet compliance & SLA requirements
```

```
Disaster Recovery is a strategy to restore 
IT systems and data 
after failures to minimize downtime and data loss.
```

```
Common Disaster Recovery Strategies

1️⃣ Backup & Restore
2️⃣ Pilot Light
3️⃣ Warm Standby
4️⃣ Active–Active (Multi-Region)
```

```
Disaster Recovery in AWS means using AWS services and architectures to recover applications and data

when:
    an EC2 instance fails
    an Availability Zone goes down
    an entire AWS Region fails

👉 Goal: minimum downtime (RTO) and minimum data loss (RPO)

AZ failure → use Multi-AZ

Region failure → use Multi-Region DR
```

```
AWS Disaster Recovery Strategies (Mapped to AWS Services)

1️⃣ Backup & Restore (Basic DR)
  
    AWS services
        S3 (store backups)
        AWS Backup
        RDS Snapshots
        AMIs (EC2 backups)

2️⃣ Pilot Light (Minimal Setup Always ON)

AWS services
    RDS (running)
    EC2 AMI (stopped)
    Auto Scaling
    S3 backup
3️⃣ Warm Standby
```

```
RTO – Recovery Time Objective

RTO is the maximum time allowed to restore a system after a disaster.

RPO – Recovery Point Objective
RPO is the maximum amount of data you can afford to lose after a disaster.
```
