```python
def validate(fun):
    def wrapper(name):
        try:
            if not  isinstance(name,str):
                raise TypeError("Not validate")
            else:
                print("validate")
        except Exception as e:
            print(e)
        s = fun(name)
        return s
    return wrapper

@validate
def getValue(name):
    return name;

getValue(4)
```

```python
def solve(str):
    temp = {}
    s=str[0]
    c = 1
    for i in range(1,len(str)):
        if str[i] == s:
            c += 1
        else:
            if s in temp and temp[s] < c:
                temp[s] = c
                s = str[i]
                c = 1
            else:
                temp[s] = c
                s = str[i]
                c = 1
    if s in temp and temp[s] < c:
        temp[s] = c
    return temp

str = "AAABBCCCDDDDAAAA"
result = solve(str)
print(result)
```

```python
def longest_valid_permutation(input_string):
    max_len = 0
    start = 0
    temp = {}
    for end in range(len(input_string)):
        char = input_string[end]
        # If character is already in the window, shrink from the left
        if char in temp and temp[char] >= start:
            start = temp[char] + 1
        temp[char] = end
        max_len = max(max_len, end - start + 1)
    return max_len

# Example usage:
input_string = "abcabc"
result = longest_valid_permutation(input_string)
print("Longest valid permutation length:", result)
```
