```python
temp = {
    "A": [49, 98, 65, 34],
    "B": [98, 56, 74, 32],
    "C": [65, 35, 56, 98]
}

n = len(temp['A'])
output = [0] * n

for i in temp:
    for j in range(n):
        output[j] += temp[i][j]
print(output)

max_value = max(output)
max_value_index = output.index(max_value)
```

```python
APIView and ModelViewSet are used to build APIs

but they serve different purposes and offer different levels of abstraction.

modelviewset  : -
  1. A ready-made view that gives you full CRUD operations with minimal code.
  2. Combines logic for listing, retrieving, creating, updating, and deleting model
  3. Requires a serializer and a queryset.
  4. Works best with routers.

from rest_framework.viewsets import ModelViewSet
from .models import Book
from .serializers import BookSerializer

class BookViewSet(ModelViewSet):
    queryset = Book.objects.all()
    serializer_class = BookSerializer
    
    
from rest_framework.routers import DefaultRouter
router = DefaultRouter()
router.register('books', BookViewSet)
urlpatterns = router.urls
```

```python
and apiview   : -

1. It’s the base class for all API views in DRF.
2. Gives you full control over request handling (GET, POST, PUT, etc.).
3. You define each method manually.


from rest_framework.views import APIView
from rest_framework.response import Response

class HelloAPIView(APIView):
    def get(self, request):
        return Response({"message": "Hello from APIView"})
```

```python
monkeypatching :-
*************

Sure! Monkeypatching is a powerful (but potentially risky) technique in Python where you dynamically
change or replace attributes or methods at runtime.


Scenario:
You have a function that fetches data from an external API
(e.g., requests to https://api.weather.com),
but you want to test it without actually hitting the API.

import requests

def get_weather(city):
    response = requests.get(f"https://api.weather.com/v1/location/{city}")
    if response.status_code == 200:
        return response.json()["temperature"]
    return None
```

```python
Duck Typing in python
*******************

https://www.youtube.com/watch?v=1YgeU5Afmro


Duck Typing in Python is a type system philosophy that says:

"If it looks like a duck, swims like a duck, and quacks like a duck — then it's a duck."

In Python, you don’t care about the type of an object, you only care if it behaves correctly.
If an object implements the methods or properties you expect, you can use it — no need for 
inheritance or type declarations.
```

```python
 Model Inheritance 
******************

Model inheritance is when one class (model) inherits fields and behaviors from another model


🔹 1. Abstract Base Classes

from django.db import models

class BaseModel(models.Model):
    created_at = models.DateTimeField(auto_now_add=True)
    updated_at = models.DateTimeField(auto_now=True)

    class Meta:
        abstract = True

class Blog(BaseModel):
    title = models.CharField(max_length=100)

class Comment(BaseModel):
    text = models.TextField()
```

```python
 2. Multi-table Inheritance
    
class Employee(models.Model):
    name = models.CharField(max_length=100)

class Manager(Employee):
    department = models.CharField(max_length=50)
```

```python
3. Proxy Models

Use this when you want to change behavior (like default ordering) without changing the model structure.

class Person(models.Model):
    name = models.CharField(max_length=100)

class OrderedPerson(Person):
    class Meta:
        proxy = True
        ordering = ['name']
```

```python
Request-Response Flow in Django
*********************************
Browser (Client) →
    URL →
        Django URL Dispatcher →
            View →
                Model (optional) →
                    Template →
                        Response →
                            Browser

                            
1. 🌐 URL Dispatcher (urls.py)
Routes HTTP requests to the correct view.


2. 🧠 View (views.py)
Handles business logic.

Interacts with models and templates.

3. 🗃️ Model (models.py)
Represents tables in your database.


4. 🧾 Template (.html)
Defines how data is presented to the user.

Client → URL Dispatcher → View → Model → Template → Response → Client
```
