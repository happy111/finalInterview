```python
2. Sharding
Split data across multiple databases based on a sharding key (e.g., user ID, region).

Each shard contains a subset of data, reducing the size and load per DB.
```

```python
Feature	                          Read-Replica	                  Sharding
Use Case	                      High read traffic	              Large datasets
Complexity	                      Medium	                      High
Sync Mechanism	                  Asynchronous replication	      Manual or app-level partitioning
```
