```python
Key Features of MongoDB
*-----------------------

1. Document-Based :- Stores data as documents in collections.
2. Schema-less  :- No strict schema — 
                    each document can have a different structure.
3. High Performance	Optimized for read/write speed and scalability.
4. Horizontal Scaling  : Easy to scale using sharding
5.   Query Language   : Powerful queries,  
                       including filtering, aggregation, and indexing.
```

```python
Basic MongoDB Commands
**********************
1. Create/Insert Document:
    db.users.insertOne({ name: "Alice", age: 25 });
2.  Find Document:
    db.users.find({ name: "Alice" });
3.  Update Document:
   db.users.updateOne({ name: "Alice" }, { $set: { age: 26 } });
```

```python
MongoDB Properties:-
1. NoSQL (Non-relational)
2. Document-Oriented
3. Schema-less
    No fixed schema — documents in the same collection can have different structures.
4. High Performance
   Designed for fast reads and writes.
   Built-in support for indexing, caching, and bulk operations.
5. Horizontal Scalability
   Uses sharding to split data across multiple servers (nodes).
6.  6. Replication
```

```python
When to Use MySQL (Relational Database)
***************************************
1. Strong Relationships
2. Fixed Schema
3. ACID Compliance Needed
4. Structured Data
5. Complex Queries
6. Business Reporting

Example Scenarios
*****************
Banking systems
Inventory and billing systems
HR and payroll systems
ERP/CRM platforms
```

```python
When to Use NoSQL (like MongoDB)
********************************
1.  Unstructured or Semi-Structured Data
2.   High Scalability
3.  Fast Writes/Reads
4. Flexible Schema
5. Big Data / High Volume
6.  Document Storage

Example Scenarios
Social media platforms
Real-time chat apps
Content management systems (CMS)
Product catalogs with varied fields
IoT device data storage
```
