```python
annotate & aggregate in django? (when working with related models.)
*******************************

annotate :-
 1.  result type QuerySet
    authors = Author.objects.annotate(total_price=Sum('book__price'))
    for author in authors:
        print(author.name, author.total_price)

aggregate :-

 1. Single dictionary
total_price = Book.objects.aggregate(Sum('price'))
print(total_price)  # {'price__sum': 120.00}

aggregate() returns a single dictionary

Useful for totals, averages, etc.
```

```python
What is celery?

Celery is an asynchronous task queue or distributed task scheduler for Python.


It’s used to run tasks in the background—outside of the main request/response cycle.

For example:

Sending emails 📧

Processing images 🖼️

Calling third-party APIs 🌐

Scheduling reports 


Advantages:
Asynchronous = fast user experience

Scalable = run many workers

Retry support, error handling

Works with Redis, RabbitMQ, etc.
```

```python
#jwt
A JWT consists of three parts: a header, a payload, and a signature. 
The header typically contains information about the token's signing algorithm, 
while the payload holds the claims or statements about the user or entity. 
The signature ensures the integrity of the token and verifies its authenticity.
```

```python

SELECT DISTINCT salary
FROM employees
ORDER BY salary DESC
LIMIT 1 OFFSET 3;

SELECT DISTINCT salary	Select unique salaries only (no duplicates)
FROM employees	From the employees table
ORDER BY salary DESC	Order salaries from highest to lowest
LIMIT 1 OFFSET 3	Skip the first 3 unique salaries, then take 1 salary
```
