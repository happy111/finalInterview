```python
git fetch → Downloads changes from remote but does not merge them.

git pull → Downloads changes and automatically merges into your local branch.
```

```python
git branch <branch_name>     # Create branch  Only create bracnch
git checkout <branch_name>   # Switch to branch
# Or in one step:
git checkout -b <branch_name> # create and switch
```

```python
How do you undo a commit?
Answer:

Undo last commit but keep changes staged:

git reset --soft HEAD~1
```

```python
Q11. What is cherry-pick in Git?
Answer:
git cherry-pick <commit_hash>
Applies the changes from a specific commit to your current branch without merging the whole branch.
```
