```python
Django request & response cycle
******************************
  1. Client Sends HTTP Request
  2. WSGI Server Receives the Request
  3. Django Settings and Middleware are Applied
     Django loads settings (settings.py) and applies middleware.
  4. URL Dispatcher (Routing)
     Django uses urls.py to match the requested path to a view function.
     If no match is found, Django returns a 404 Not Found.
  5. View is Called

    This view can:
    Query the database (via models)
    Render templates
    Perform logic
    Return a HttpResponse object (HTML, JSON, etc.)
  6. Middleware Processes Response
  7. Response is Returned to Client
```

```python
Http status code:-
******************

Category                    Code Range                        Meaning
1xx	                         100–199	                    Informational responses
2xx	                         200–299	                    Success
3xx	                         300–399	                    Redirection
4xx	                         400–499	                    Client errors
5xx	                         500–599	                    Server errors


2xx – Success
    200 OK – Request was successful.
    201 Created – Resource successfully created.
    204 No Content – Request successful, but no content to return.

400 Bad Request – Malformed request.
401 Unauthorized – Authentication required.
403 Forbidden – Server understood request, but refuses to authorize it.
404 Not Found – Resource not found.
405 Method Not Allowed – Request method not supported for this resource.

500 Internal Server Error – Server encountered an unexpected condition.
502 Bad Gateway – Invalid response from upstream server.
503 Service Unavailable – Server is temporarily overloaded or down.
```

```python
Django security:-
*****************
  1 . Cross Site Scripting (XSS) Protection:- 
    
      Cross-Site Scripting (XSS) is an attack where malicious scripts 
        are injected into web pages viewed by others. Django protects
        against XSS by automatically 
        escaping user input in templates.
        
 2 . Cross Site Request Forgery (CSRF) Protection

    In Django, CSRF protection works by adding a hidden token to every 
    POST form ({% csrf_token %}), and verifying that token on the server
    to ensure the request is legitimate.
    
 3. SQL Injection Protection

   Django protects against SQL injection by using its ORM 
    (Object-Relational Mapper)
 4. SSL/HTTPS
 5. Host header validation
 6. session
```

```python
Middleware :-

 Build in Middleware :-

  1. Security Middleware
  2. Session Middleware
  3. CommonMiddleware
  4. csrf middleware
  5. Authentication middleware


Methods are :- 


1. __init__(self, get_response)
    Called once when the server starts.

     Used for one-time configuration.

     get_response is a callable that processes the next middleware/view.

2. __call__(self, request)
   Called for each request.

    This is where you usually call get_response(request)
    to continue the middleware chain.

3. process_view(self, request, view_func, view_args, view_kwargs)
   Called before the view is executed.

    You can return an HttpResponse to short-circuit the view.

4. process_exception(self, request, exception)
   Called if a view raises an exception.

   You can return a custom HttpResponse or None.

5. process_template_response(self, request, response)
  
    Called after the view has returned a TemplateResponse.

    Allows you to modify the template response before rendering.
```

```python
class Singleton:
    _instance = None

    def __new__(cls):
        if cls._instance is None:
            print("Creating new instance")
            cls._instance = super(Singleton, cls).__new__(cls)
        else:
            print("Using existing instance")
        return cls._instance


# Test it
obj1 = Singleton()
obj2 = Singleton()

print(obj1 is obj2)  # True – both are the same instance
```

```python
 Get the 2nd highest salary
    *******************
    
class Employee(models.Model):
    name = models.CharField(max_length=100)
    salary = models.IntegerField()
    
 second_highest_salary = (
    Employee.objects
    .order_by('-salary')          # Sort descending
    .values_list('salary', flat=True)
    .distinct()                   # Make sure salaries are unique
)[1]                              # Index 1 = 2nd highest   
```
