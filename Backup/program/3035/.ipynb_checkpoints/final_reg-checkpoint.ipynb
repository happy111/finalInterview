```python
import re  # check start
s = re.match(r'u','umesh')
print(s.group())
```

```python
# Anywhere
result = re.search(r"o", "Hello World")
print(result.group())  # Output: World
```

```python
# FindAll
vowels = re.findall(r"[aeiou]", "Hello World")
print(vowels)  # Output: ['e', 'o', 'o']
```

```python
text = "Hello 123 World"
cleaned = re.sub(r"\d", "", text)
print(cleaned)  # Output: Hello  World
```

```python
import re

s = "I(*Like)**Coding*&&^in()*(Python"
words = re.findall(r'[a-zA-Z]+', s)
output = ' '.join(words)

print(output) 
```

```python
import re

text = "My phone number is 9876543210."

# Find all alphabetic characters
characters = re.findall(r"[a-zA-Z]", text)
print(characters)
```

```python
import re

text = "My phone number is 9876543210."

# Find all alphabetic characters
characters = re.findall(r"[a-zA-Z]+", text)
print(characters)
only_chars = " ".join(characters)
print(only_chars)
```
