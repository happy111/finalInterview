```python
class School:
    def __init__(self,name):
        self.schoolName = name
    @classmethod
    def set_message(cls):
        cls.message = "Hello from class method!"
    @classmethod
    def display(cls):
        cls.country = "India"
        
class Student(School):
    def __init__(self,studentName,name):
        super().__init__(name)
        self.studentName = studentName
    def disp(self):
        print(self.schoolName)
        print(self.studentName)
        
obj = Student("Umesh","SK")
obj.disp()
School.set_message()
print(School.message)
School.display()
print(School.country)
```

```python
class School:
    def __init__(self, name):
        self.schoolName = name

    @classmethod
    def set_message(cls):
        cls.message = "Hello from class method!"

    @classmethod
    def display(cls):
        cls.country = "India"

class Student(School):
    def __init__(self, studentName, name):
        super().__init__(name)
        self.studentName = studentName

    def disp(self):
        print(self.schoolName)
        print(self.studentName)

# Create a student object
obj = Student("Umesh", "SK")
obj.disp()

# Call classmethod to set class variable
School.set_message()

# ✅ Correct way to print the class variable
print(School.message)


School.display()
print(School.country)
```

```python
class Emp:
    def __new__(cls):
        print("__new__ called")
        instance = super().__new__(cls)  # actually create an object
        return instance

    def __init__(self):
        print("__init__ called")
        self.course = "MCA"

    def disp(self):
        print("Inside disp():", self.course)

# Now creating the object
obj = Emp()
print(obj)        # prints the object reference
obj.disp()        # works fine
```

```python
Method	Purpose	Must Return
__new__	Creates the object	A new object
__init__	Initializes the object	Must return None
```

```python
class Methods:-

__dict__  : return document of class
__doc__   : Show all details
__name__  : print class name
__module__ : FIle name
__bases__  : return parent class
```
