```python
class Node:
    def __init__(self,data):
        self.next = None
        self.data = data
class LinkedList:
    def __init__(self):
        self.head = None
    def insertBeg(self,data):
        self.new_data = Node(data)
        self.new_data.next = self.head
        self.head = self.new_data

    def insertEnd(self,data):
        self.new_data = Node(data)
        if self.head == None:
            self.new_data.next = self.head
            self.head = self.new_data
        else:
            self.temp = self.head
            while self.temp.next != None:
                self.temp = self.temp.next
            self.temp.next = self.new_data
            
            
    def display(self):
        self.temp = self.head
        while self.temp:
            print(self.temp.data,end = " ")
            self.temp = self.temp.next

    def reverse(self):
        prev = None
        curr = self.head
        fut = None
        while curr != None:
            fut = curr.next
            curr.next = prev
            prev = curr
            curr = fut
        self.head = prev

obj = LinkedList()
obj.insertBeg(1)
obj.insertBeg(2)
obj.insertBeg(3)
obj.insertEnd(10)
obj.insertEnd(11)
obj.display()
print()
print("<<<<<<<<<<<<<<<<<<<<Reverse of the Element<<<<<<<<<<<<<<<<<<<<<<<<")
obj.reverse()
obj.display()
```
