```python
In Django ORM, aggregate() and annotate() are both used for
performing calculations like count, sum, average, etc.,
but they serve different purposes:
```

```python
aggregate()
Performs a calculation over the entire QuerySet
and returns a single dictionary result.

Book.objects.aggregate(Avg('price'))
# Output: {'price__avg': 123.45}
```

```python
annotate()
Adds a calculated value to each row in the QuerySet.

from django.db.models import Count

Author.objects.annotate(book_count=Count('book'))
```

```python
aggregate()
***********
1. Single dictionary
2. Whole QuerySet
3. Total, Average of all records
4. Dict

annotate()
*********
1. QuerySet with extra fields
2. Row-level (each object)
3. Count, Sum per object (e.g., per Author)
4. QuerySet
```
