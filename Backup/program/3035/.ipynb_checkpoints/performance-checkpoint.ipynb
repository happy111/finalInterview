```python
How can you improve the performance of a Python API?
```

```python
Ans : -To improve the performance of a Python API, I focus on four key areas:

1. Efficient Database Access

     a. Use select_related and prefetch_related to reduce queries.
     b. Avoid N+1 problems and use aggregation instead of looping in Python.
     c. Add appropriate indexes and optimize SQL queries.

2.  Caching
3. Asynchronous Operations
4. Efficient Code
   1. Avoid unnecessary loops and large payloads.
   2. Use generators, list comprehensions, 
        and optimized data structures (e.g., set over list).
5. Load Management & Scalability
   1. Use Gunicorn + Nginx for production deployment.
   2. Enable GZIP compression and pagination to reduce payload size.
6. Monitoring & Profiling
```

```python
 Real-World Techniques
 **********************
     1. DB Optimization
     2. Cache
     3. Async/Parallel
     4. Pagination
     5. Compression   Enable GZIP
     6. Reduce Payload
     7. Rate Limiting
```

```python
1️⃣6️⃣ How do you secure microservices?

OAuth2 / JWT

mTLS

            mTLS (Mutual Transport Layer Security) is a security mechanism where
            both the client and the server authenticate each other using TLS certificates.
            
            👉 In normal TLS:
            
            Client verifies server
            
            Server does not verify client
            
            👉 In mTLS:
            
            Client verifies server
            
            Server verifies client
            ✔️ Two-way authentication


API Gateway auth
Role-based access control
```
