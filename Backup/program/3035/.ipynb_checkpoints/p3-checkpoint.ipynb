```python
What is ETL?

Extraction | Transform | Load

Extract → Transform → Load

It's a data pipeline process used to move and process data from one
system to another
(e.g., from databases to data warehouses).
```

```python
import pandas as pd
# Online CSV URL (raw GitHub link)
url = "https://raw.githubusercontent.com/uiuc-cse/data-fa14/gh-pages/data/iris.csv"
# Process the file in chunks
chunksize = 20  # small chunk size for demo; use larger for real data
# Initialize a counter
total_rows = 0
for chunk in pd.read_csv(url, chunksize=chunksize):
    print(f"\nProcessing chunk with {len(chunk)} rows:")
    print(chunk.head(2))  # preview first 2 rows of each chunk

    total_rows += len(chunk)

print(f"\nTotal rows processed: {total_rows}")
```
