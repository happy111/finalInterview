```python
def solve(s):
    res = []
    if len(s) == 0:
        return [""]
    for i in range(len(s)):
        c = s[i]
        for p in solve(s[:i]+s[i+1:]):
            res.append(c+p)
    return res
s = "abc"
res = solve(s)
print(res)
```

```python
def solve(res):
    temp={}
    for i in res:
        temp[i] = temp.get(i,0)+1
    s = list(temp.keys())
    # print(temp)
    #print(s)
    result=""
    for i in s:
        result += str(temp[i]) + i
    return result
res = "DDDDRRSS"
output = "4D2R2S"
result = solve(res)
print(result)
```

```python
def is_sublist(list1,list2):
    count = 0
    for i in range(len(list1) - len(list2)+1):
        if list1[i:i+len(list2)] == list2:
            count += 1
    print(count)

list1 = [1, 2, 3, 4, 5, 6]
list2 = [1, 2, 3]

print(is_sublist(list1, list2))
```

```python
def findIndex(arr,flag):
    left = 0
    right = len(arr) - 1
    k = 7
    while left <= right:
        mid = left+(right-left) // 2
        if arr[mid] == k:
            if arr[mid] == k and arr[mid+1] == k:
                return [mid,mid+1]
            elif arr[mid-1] == k:
                return [mid-1,mid]
            else:
                return [mid,mid]
        elif arr[mid] < k:
            left = mid + 1
        else:
            right = mid -1


def solve(arr):
    left = 0
    right = len(arr) - 1
    firstindex = findIndex(arr,True)
    return firstindex



arr = [5,7,7,8,8,10]
res = solve(arr)
print(res)
```

```python
def read_file(file_path):
    with open(file_path, 'r', encoding='utf-8') as b:
        s = b.readlines()
        for li in s:
            print(li)
 

file_path = "word.txt"
result = read_file(file_path)

# Using next() to get words one by one
print(next(result))  # First word
print(next(result))  # Second word
print(next(result))  # First word
print(next(result)) 
```

```python

In Python, shallow copy and deep copy behave the same when used on a simple (non-nested) list.
However, the difference becomes apparent with nested lists or complex objects.
With a shallow copy, if you modify an element inside a nested list, the change will reflect
in both the original list and the copied list — because they still share references to the inner objects.
In contrast, a deep copy creates entirely independent copies of all nested elements, so changes
in the copied list do not affect the original.

```

```python
def binary_to_decimal(n):
    output = 0
    s = len(n)-1
    for i in n:
        output += (int(i))*(2**int(s))
        s -= 1
    return output

n = "1011"
result = binary_to_decimal(n)
print(result)
```

```python
# Decimal to Binary
def binary_to_decimal(n):
    output = ""
    while n > 0:
        rem = int(n) % 2
        output = str(rem) + output
        n = n // 2
    return output

n = 11
result = binary_to_decimal(n)
print(result)
```

```python
#  Bobblesort

def bubbleSort(arr):
    n = len(arr)
    for i in range(n):
        for j in range(n-1-i):
            if arr[j] < arr[j+1]:
                arr[j],arr[j+1] = arr[j+1],arr[j]
    return arr
arr = [7,3,9,2,0,4,8,1,6,5]
res = bubbleSort(arr)
print(res)
```

```python
# Reverse a string

def reverse(s):
    ans = ""
    for i in s:
        ans =  i + s
    return ans

s = input("Enter a string")
res = reverse(s)
print(res)
```

```python
def decimal_sum(num1,num2):
    carry = 0
    output = ''
    while int(num1) > 0 or int(num2) >0 or carry:
        if int(num1) > 0:
            rem1 = int(num1) % 10
            carry += rem1
            num1 = int(num1) // 10
        if int(num2) > 0:
            rem2 = int(num2) % 10
            carry += rem2
            num2 = int(num2) // 10
        
        s = carry % 10
        output = str(s) +output
        carry = carry // 10
    return output


num1 = "345"
num2 = "846"
result = decimal_sum(num1,num2)
print(result)
```
