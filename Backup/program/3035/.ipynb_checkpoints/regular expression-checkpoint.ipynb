```python
import re
```

```python
.	Any character except newline	a.b matches acb
^	Start of string	^hello matches "hello world"
$	End of string	world$ matches "hello world"
*	0 or more repetitions	lo* matches "lo", "loo", "looo"
+	1 or more repetitions	lo+ matches "lo", "loo", but not "l"
?	0 or 1 repetition	lo? matches "l" or "lo"
[]	Matches one char in set	[aeiou] matches any vowel
[^]	Matches any char not in set	[^0-9] excludes digits
{n}	Exactly n repetitions	a{3} matches "aaa"
`	`	OR
()	Grouping	(ab)+ matches "ab", "abab", etc.
```

```python
vowels = re.findall(r"[aeiou]", "Hello World")
print(vowels)  # Output: ['e', 'o', 'o']
```
