```python
def print_star(n):
    for i in range(1,n+1):
        space = " " *(n+1-i)
        print(space+("* " * i))

n = int(input("Enter a number"))
result = print_star(n)
```

```python
def sort_list(arr):
    j = 0;
    while j < len(arr)-1:
        if arr[j] > arr[j+1]:
            arr[j],arr[j+1] = arr[j+1],arr[j]
            j = -1
        j +=1
    return arr

arr = [11, 2, 99, 9, 8,7, 6, 0, 5, 4, 3]
result = sort_list(arr)
print(result)
```

```python
li = [1,2,3,3,4,5]

def reverse_li(li):
    if len(li) == 0:
        return []
    else:
        return [li[-1]]+ reverse_li(li[:-1])

reverse_li(li)
```

```python
def palindrom(str1):
    newStr = str1[::-1]
    return str1 == newStr

result = palindrom("racecar")

if result:
    print("String Is Palindrome")
else:
    print("String is not palindrome")
    
```

```python
def are_anagrams(str1, str2):
    # Remove spaces and convert to lowercase
    str1 = str1.replace(" ", "").lower()
    str2 = str2.replace(" ", "").lower()
    
    # Check if the sorted characters of both strings are the same
    return sorted(str1) == sorted(str2)

# Test cases
string1 = "listen"
string2 = "silent"
print(are_anagrams(string1, string2))  # Output: True

```

```python
def is_valid_parenthese(str1):
    stack, pchar = [], {"(": ")", "{": "}", "[": "]"}
    for parenthese in str1:
        if parenthese in pchar:
            stack.append(parenthese)
        else:
            if not stack:
                return False
            if pchar[stack.pop()] != parenthese:
                return False
    return len(stack) == 0

print(is_valid_parenthese("{{}}"))
```

```python
def areRotations(s1, s2):
    n = len(s1)

    # Generate and check all possible rotations of s1
    for _ in range(n):
        
        # If current rotation is equal to s2, return true
        if s1 == s2:
            return True

        # Right rotate s1
        s1 = s1[-1] + s1[:-1]
        print(s1)

    return False

if __name__ == "__main__":
    s1 = "aab"
    s2 = "aba"

    print("true" if areRotations(s1, s2) else "false")
```

```python
def commonPrefix(li):
    s = sorted(li)
    first = s[0]
    last = s[-1]
    min_len = min(len(first),len(last))
    print(min_len)
    i = 0
    while(i < min_len and first[i] == last[i]):
        i +=1
    return first[0:i]

li = ["geeksforgeeks", "geeks", "geek", "geezer"]
result = commonPrefix(li)
print(result)
```
