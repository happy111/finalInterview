```python
def find_permutations(s):
    if len(s) == 1:
        return [s]
    result = []
    for i in range(len(s)):
        for p in find_permutations(s[:i]+s[i+1:]):
            result.append(s[i]+p)
    return result

s = "abc"
result = find_permutations(s)
for k in result:
    print(k)
```

```python
def find_permutations(s):
    if len(s) == 0:
        return ['']
    permu = []
    for i in range(len(s)):
        current_str = s[i]
        rema_chr = s[:i]+s[i+1:]
        rema_per = find_permutations(rema_chr)
        for per in rema_per:
            permu.append(current_str+per)
    return permu

s = "abc"
result = find_permutations(s)
for k in result:
    print(k)
```

```python
def solve(s):
    n = len(s)
    ans = [1] * n

    prefix  =1
    for i in range(n):
        ans[i] = prefix
        prefix *= s[i]

    sufix = 1
    for i in range(n-1,-1,-1):
        ans[i] *= sufix
        sufix *= s[i]
    print(ans)
    
s = [1,2,3,4]
result = solve(s)
```

```python
def solve(s):
    ans = [1] * len(s)

    prefix = 1
    for i in range(len(s)):
        ans[i] = prefix
        prefix *= s[i]

    sufix = 1
    for i in range(len(s)-1,-1,-1):
        ans[i] *= sufix
        sufix *= s[i]
    return ans

s = [1,2,3,4]
result = solve(s)
print(result)
```

```python
def fibo(n, temp):
    if n in temp:
        return temp[n]
    
    if n <= 1:
        temp[n] = n
        return temp[n]
    
    temp[n] = fibo(n - 1, temp) + fibo(n - 2, temp)
    return temp[n]

temp = {}
res = fibo(4, temp)

print(res)     # Output: 3
print(temp)    # Optional: shows memoized results
```

```python
def solve(li):
    temp = {}
    for i in li:
        new_data = ''.join(sorted(i))
        if new_data in temp:
            temp[new_data].append(i)
        else:
            temp[new_data] = [i]

    res = []
    for k in temp.values():
        if len(k) > 1:
            res.append(k)
        else:
            print("ddd")
            res.append(k[0])
    return res
    


li = ['ant','tna','abc','bca','der']
result = solve(li)
print(result)
```

```python
def solve(li,k):
    temp = {}
    res = []
    for i in li:
        tar = k - i
        if tar in temp:
            res.append([i,tar])
        else:
            temp[i] = True
    return res

li = [9,3,4,5,2,1,2]
k = 9
result = solve(li,k)
print(result)
```

```python
class Singleton:
    _instance = None
    def __new__(cls):
        if cls._instance == None:
            cls._instance = super().__new__(cls)
        else:
            print("Use existing instance")
        return cls._instance
```

```python
obj = Singleton()
obj1 = Singleton()
```
