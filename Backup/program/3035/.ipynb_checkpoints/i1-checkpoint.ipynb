```python
def fibonacci(n):
    if n <= 1:
        return n
    return fibonacci(n - 1) + fibonacci(n - 2)

# Testing the function
for i in range(10):
    print(f"Fibonacci({i}) = {fibonacci(i)}")
```

```python
### Why generator is fast compare to function?

1. Lazy Evalution
2. Generators use lazy evaluation, meaning they generate values on the fly, 
one at a time.
3. A generator doesn’t store the entire result in memory.
```

```python
Feature	                Lazy Evaluation	                 Eager Evaluation
When computed	         When needed	                    Immediately
Memory usage	         Lower (in many cases)	              Higher
Performance	Better (if unused values)	May compute unneeded values
```

```python
#  concurrency and parallelism

Concurrency:
    
Definition: Concurrency refers to the ability of a system to handle 
multiple tasks at the same time.
However,this does not necessarily mean that these tasks are being 
executed simultaneously.

Parallelism:
    Definition: Parallelism involves the simultaneous execution of 
    multiple tasks. 
    This means that tasks are literally running at the same time on
    multiple processors or cores.
```

```python
# monkeypatching

# Monkey patching in Python refers to the dynamic modification 
# or extension of a class or module at runtime.


class Dog:
    def bark(self):
        return "Woof!"

def show():
    print("Hello")
# Using the original method
dog = Dog()
print(dog.bark())  # Output: Woof!
dog.bark = show()
```

```python
class  Singleton:
    _instance = None
    def __new__(cls):
        if cls._instance is None:
            cls._instance = super().__new__(cls)
        else:
            print("Using existing instance")
        return cls._instance
    
# Test it
obj1 = Singleton()
obj2 = Singleton()

print(obj1 is obj2)  # True – both are the same instance
```

```python
Generator :-

A generator is a special kind of function in Python.

It uses the yield keyword to return values one at a time.

Each time you call it, it gives the next value and pauses.

It does not store all values in memory.

Generators are useful when working with large amounts of data.

They save memory and are faster in many cases.

A generator returns an iterator object.

You can use next() to get the next value from a generator.

Generators stop when there are no more values to yield.

They are easier to write than regular iterators.
```

```python
annotate & aggregate in django? (when working with related models.)
*******************************

ans. In Django, annotate and aggregate are query functions that allow you to perform calculations 
and aggregations on your database queries.
annotate adds new fields to each object in the queryset based on some calculations or aggregations. 
These fields are not stored in the database but are calculated on-the-fly when the query is executed.
Here's an example of using annotate:




annotate :-
 1.  result type QuerySet
    authors = Author.objects.annotate(total_price=Sum('book__price'))
    for author in authors:
        print(author.name, author.total_price)

aggregate :-

 1. Single dictionary
total_price = Book.objects.aggregate(Sum('price'))
print(total_price)  # {'price__sum': 120.00}

aggregate() returns a single dictionary

Useful for totals, averages, etc.
```

```python
Django ORM Equivalent of GROUP BY dept
***************************************

class Employee(models.Model):
    name = models.CharField(max_length=100)
    dept = models.CharField(max_length=50)
    salary = models.IntegerField()
    
from django.db.models import Sum

result = Employee.objects.values('dept').annotate(total_salary=Sum('salary'))

output : [
    {'dept': 'HR', 'total_salary': 50000},
    {'dept': 'IT', 'total_salary': 120000},
    {'dept': 'Sales', 'total_salary': 70000}
]
```
