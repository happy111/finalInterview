```python
def count_factor(n):
    count = 0
    for i in range(1,n+1):
        if n % i == 0:
            count += 1
    return count
def solve(n):
    nooffactor = count_factor(n)
    print(nooffactor)
    if nooffactor == 2:
        print("Number is prime")
    else:
        print("Number is not prime")

n = int(input("Enter a number"))
res = solve(n)
```

```python
What is namedtuple?
namedtuple is a factory function from the collections module that creates a tuple subclass with named fields.

It gives you:

The immutability and speed of a tuple

The readability of named attributes (like a class)
```

```python
from collections import namedtuple

Point = namedtuple('Point', ['x', 'y'])

p1 = Point(10, 20)
print(p1.x)  # 10
print(p1.y)  # 20
```

```python
class Student:
    school_name = "DPS"
    def __init__(self,fname,lname):
        self.fname = fname
        self.lname = lname
    @classmethod
    def display(cls):
        cls.school_name = "asdfs"
        print("School name is", cls.school_name)
      
    @staticmethod
    def display1():
        Student.school_name = "asdfs"
        print("School name is", Student.school_name)
obj = Student("umesh","samal")
obj.display1()
```

```python
## Linear Search Algorithm
## function definition
## time complexity: O(n)
## space complexity: O(1)

def linearSearch(arr, x):
    for i in range(len(arr)):
        if arr[i] == x:
            return i
    return -1
    
## Driver code
arr = [20, 45, 27, 47, 55, 67, 75, 88, 90]
x = 65
## function calling
result = linearSearch(arr, x)
print("Searching element is present at index:",result)
```

```python
## Binary Search Algorithm using recursion
## function definition
def binarySearch(arr, x, i, j):
    while i <= j:
        mid = i + (j-i)//2
        if arr[mid] == x:
            return mid
        elif arr[mid] < x:
            ## right side of the mid
            ## seach space - mid+1 to j
            ## recursion - calling the same function again inside the 
            ## method definition
            return binarySearch(arr, x, mid+1, j)
        else:
            ## left side of the mid
            ## search space - i to mid-1
            return binarySearch(arr, x, i, mid-1)
    ## searching element is not present in an array
    return -1

## Driver code
arr = [20, 30, 40, 50, 60, 70, 80, 90]
x = 80
i = 0
j = len(arr) - 1
## function calling
result = binarySearch(arr, x, i, j)
print("Searching element is present at the location:",result)
```

```python
## Binary Search Algorithm without using recursion(iterative)
## function definition
def binarySearch(arr, x, i, j):
    while i <= j:
        ## lower bound while division
        mid = i + (j-i)//2
        if arr[mid] == x:
            return mid
        elif arr[mid] < x:
            ## right side of the mid
            ## seach space - mid+1 to j
            i = mid + 1
            ## arr[mid] > x:
        else:
            ## left side of the mid
            ## search space - i to mid-1
            j = mid - 1
    ## searching element is not present in an array
    return -1

## Driver code
arr = [20, 30, 40, 50, 60, 70, 80, 90]
x = 20
i = 0
j = len(arr) - 1
## function calling
result = binarySearch(arr, x, i, j)
print("Searching element is present at the location:",result)
```

```python
def search_element(li,search_data,start,end):
    mid = start+(end-start)//2
   
    if li[mid] == search_data:
        print("Found element",mid)
    elif li[mid] > search_data:
        return search_element(li,search_data,start,mid-1)
    else:
        return search_element(li,search_data,mid+1,end)

    


search_data = 66
li = [4,22,66,11,2,34,1]
li.sort()
print(li)
start = 0
end = len(li)
search_element(li,search_data,start,end)
```

```python
def check_cyclic(str1,str2):
    dic = {}
    dic1 = {}
    for i in str1:
        dic[i] = dic.get(i,0)+1
    for i in str2:
         dic1[i] = dic1.get(i,0)+1

    if dic == dic1:
        print("str1 and str2 are cyclic")
    else:
        print("str1 and str2 are not cyclic")

str1 = "abcedef" 
str2 = "defabce"
check_cyclic(str1,str2)
```

```python
def sort_list(li):
    n = len(li)
    for i in range(n):
        for j in range(n-i-1):
            if li[j] > li[j+1]:
                li[j],li[j+1] = li[j+1],li[j]
    return li

list1 = [90, 112, 71, 70,10,20]
sort_list(list1)
```

```python
class MyDecorator:
    def __init__(self, function):
        self.function = function
    def __call__(self):
        # We can add some code 
        # before function call
        self.function()
        # We can also add some code
        # after function call.
  
# adding class decorator to the function
@MyDecorator
def function():
    print("GeeksforGeeks")
 
function()
```

```python
def is_sublist(lst, sublist):
    n = len(sublist)
    for i in range(len(lst) - n + 1):
        print(lst[i:i+n])
        if lst[i:i+n] == sublist:
            return True
    return False

list1 = [1, 2, 3, 4, 5, 6]
list2 = [1, 2, 3]

print(is_sublist(list1, list2))  # Output: True

list3 = [7, 8, 9]

print(is_sublist(list1, list3)) 
```
