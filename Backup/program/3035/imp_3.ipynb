```python
from abc import ABC, abstractmethod

# Abstract class
class Teacher(ABC):

    @abstractmethod
    def display(self):
        pass  # Abstract method (must be overridden)

    def sum(self, a, b):
        c = a + b
        print("Sum is:", c)  # Concrete method

# Subclass that extends the abstract class
class MathTeacher(Teacher):

    def display(self):
        print("I am a Math Teacher.")

# Create object of subclass
t = MathTeacher()
t.display()       # Output: I am a Math Teacher.
t.sum(10, 20)     # Output: Sum is: 30
```

```python
# Monkey patching example
# ***********************

# Monkey patching is the dynamic modification of a class or module at runtime 

class Student:
    def __init__(self):
        print("Umesh samal")

def display():
    print("Hello dosto")

Student.display = display

obj = Student
obj.display()
```
