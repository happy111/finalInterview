```python
import string
def print_rangoli(n):
    alp = string.ascii_lowercase
    print(alp)
    max_width = 4 * n -3
    
    lines = []
    for i in range(n):
        s = '-'.join(alp[n-1:i:-1] + alp[i:n])
        lines.append(s.center(max_width,'-'))
    
    full_rangaoli = lines[-1::-1] + lines[1:]
    
    for line in full_rangaoli:
        print(line)

n = int(input("Enter a number"))
result = print_rangoli(n)
print(result)
```

```python
def is_vowel(letter):
    return letter in ['a', 'e', 'i', 'o', 'u']

def score_words(words):
    score = 0
    words = words.split(" ")
    for word in words:
        num_vowels = 0
        for letter in word:
            if is_vowel(letter):
                num_vowels += 1
        print("ssssssssssssssssss",num_vowels,score)
        if num_vowels % 2 == 0:
            score += 2
        else:
            score += 1
    return score
s = "programming is awesome"
result = score_words(s)
print(result)
```
