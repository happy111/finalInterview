```python
✅ 1. What is vectorization in NumPy? Why is it faster?

NumPy is faster because:
    Uses optimized C/Fortran code internally
    Uses SIMD (Single Instruction Multiple Data)
    Works on contiguous memory blocks
Ex:-
arr = np.array([1,2,3])
result = arr * 2   # vectorized
```

```python
✅ 2. What is broadcasting? Explain rules.
Answer

Broadcasting allows operations on arrays of different shapes.

a = np.array([1,2,3])          # shape (3,)
b = np.array([[1],[2],[3]])    # shape (3,1)
```

```python
✅ 3. What is the difference between view() and copy()?

view()
    Creates a new array object
    Shares the same data
    Very fast
    Changes reflect in original

copy()
    Creates a new object with new data
    Changes do not reflect
```

```python
#Ex
import numpy as np
a = np.array([1,2,3])
b = a.view()
b[0] = 100
```

```python
c = a.copy()
c[0] = 200 
```

```python
✅ 4. What is the difference between contiguous and non-contiguous arrays?

A contiguous array has memory stored without gaps.

    Contiguous arrays are faster for operations.
```

```python
✅ 5. Explain "stride" in NumPy.

Stride = number of bytes to step to move to next element.
```

```python
a = np.arange(10)[::2]
a.strides
```

```python
✅ 7. How does NumPy store data internally?

NumPy stores data in:
    Homogeneous type
    Compact contiguous buffers
    Using C-style arrays

Uses ndarray object with metadata (shape, strides, dtype)
```

```python
import numpy as np

arr = np.arange(1, 26).reshape(5, 5)
arr
```

```python
# ✅ 7. Compute Euclidean distance between two vectors

a = np.array([1,2,3])
b = np.array([4,5,6])

dist = np.linalg.norm(a - b)
dist
```

```python
#✅ 8. Replace all values between 10 and 20 with 999
arr = np.array([5, 12, 15, 22, 35])
arr[(arr >= 10) & (arr <= 20)] = 999
arr
```
