```python
stocks = ['AMZN', 'CACC', 'EQIX', 'GOOG', 'ORLY', 'ULTA']

prices = [
    [12.81, 11.09, 12.11, 10.93, 9.83, 8.14],
    [10.34, 10.56, 10.14, 12.17, 13.1, 11.22],
    [11.53, 10.67, 10.42, 11.88, 11.77, 10.21]
]

stock_sum = {}

for i, stock in enumerate(stocks):
    stock_sum[stock] = prices[0][i] + prices[1][i] + prices[2][i]

print(stock_sum)
```

```python
import pandas as pd

stocks = ['AMZN', 'CACC', 'EQIX', 'GOOG', 'ORLY', 'ULTA']

prices = [
    [12.81, 11.09, 12.11, 10.93, 9.83, 8.14],
    [10.34, 10.56, 10.14, 12.17, 13.1, 11.22],
    [11.53, 10.67, 10.42, 11.88, 11.77, 10.21]
]

df = pd.DataFrame(prices, columns=stocks)
```

```python
stock_sum = df.sum(axis=0)
print(stock_sum)
```

```python
stock_sum = df.mean(axis=0)
print(stock_sum)
```

```python
top3_stock_avg = df.mean().sort_values(ascending=False).head(3)
print(top3_stock_avg)
```

```python
select salary from emp order by salary desc limit 1 offset 6


select salary, dept from emp group by dept order by salary desc

SELECT emp_id, emp_name, dept, salary
FROM (
    SELECT *,
           ROW_NUMBER() OVER (
                PARTITION BY dept
                ORDER BY salary DESC
           ) AS rn
    FROM emp
) t
WHERE rn <= 3
ORDER BY dept, salary DESC;


```

```python
def solve(st):
    output=[]
    for s in st.split(" "):
        result = ""
        for k in range(len(s)):
            if s[k].isupper() and (k == 0 or k == len(s)-1):
                p = s[k].lower()
                result += p
            if s[k].islower() and (k == 0 or k == len(s)-1):
                p =  s[k].upper()
                result += p
            if k != 0 and k != len(s)-1:
                result += s[k]
        output.append(result)
    return ' '.join(output)


st = "Today is My IntervieW"
output = "todaY IS mY interview"
res = solve(st)
print(res)
```

```python
import pandas as pd

data = {
    'dept': ['IT', 'IT', 'HR', 'HR', 'Finance'],
    'name': ['A', 'B', 'C', 'D', 'E'],
    'salary': [50000, 60000, 55000, 52000, 70000]
}

df = pd.DataFrame(data)
print(df)
```

```python
df.groupby('dept')['salary'].mean()
```

```python
df.groupby('dept')['name'].count()
```

```python
df.groupby('dept')['salary'].sum()
```

```python
import pandas as pd

data = {
    'dept': ['IT', 'IT', 'HR', 'HR', 'Finance'],
    'name': ['A', 'B', 'C', 'D', 'E'],
    'gender': ['F', 'F', 'F', 'F', 'm'],
    'salary': [50000, 60000, 55000, 52000, 70000]
}

df = pd.DataFrame(data)
print(df)

df.groupby(['dept', 'gender'])['salary'].mean()
```

```python
df.groupby('dept')['salary'].mean()
```

```python
df.groupby('dept')['name'].count()
```

```python
# 4. Using agg() to apply multiple functions

df.groupby('dept').agg({
    'salary': ['min', 'max', 'mean', 'count']
})
```

```python
# ⭐ 5. Reset index after groupby
df.groupby('dept')['salary'].mean().reset_index()
```

```python
# ⭐ 6. Sort after groupby
df.groupby('dept')['salary'].mean().sort_values(ascending=False)
```

```python
# ⭐ 7. Filter groups (e.g., departments with more than 1 employee)
df.groupby('dept').filter(lambda x: len(x) == 2)
```

```python
import pandas as pd

data = {
    'dept': ['IT', 'IT', 'HR', 'HR', 'Finance', 'Finance', 'Marketing'],
    'name': ['A', 'B', 'C', 'D', 'E', 'F', 'G'],
    'salary': [50000, 60000, 55000, 52000, 70000, 65000, 48000],
    'gender': ['M', 'F', 'M', 'F', 'F', 'M', 'F']
}

df = pd.DataFrame(data)
print(df)
```

```python
# 🎯 1. Find average salary per department

df.groupby('dept')['salary'].mean()
```

```python
#🎯 2. Count employees in each department
df.groupby('dept')['salary'].count()
```

```python
# 🎯 3. Find minimum and maximum salary per department
df.groupby('dept')['salary'].agg(['min', 'max']).reset_index()
```

```python
# 🎯 4. Apply multiple functions on multiple columns

df.groupby('dept').agg({
    'salary': ['mean', 'sum', 'count'],
    'name': 'count'
}).reset_index()
```

```python
# 🎯 5. Group by two columns (dept + gender)

df.groupby(['dept', 'gender'])['salary'].mean()
```

```python
# 🎯 6. Sort departments by total salary

df.groupby('dept')['salary'].sum().sort_values(ascending=False)
```

```python
# 🎯 7. Departments having more than 1 employee
df.groupby('dept').filter(lambda x: len(x) > 1)
```

```python
# 🎯 8. Find top-paid employee in each department

df.loc[df.groupby('dept')['salary'].idxmax()]
```

```python
df.loc[[4,2]]
```

```python
# 🎯 10. Add total salary of each department to each row (using transform)

df['dept_total_salary'] = df.groupby('dept')['salary'].transform('sum')
```

```python
df
```

```python
🎯 12. Percent salary contribution inside each department
```

```python
df['percent'] = df['salary'] / df.groupby('dept')['salary'].transform('sum') * 100
```

```python
df
```

```python
🎯 13. Rank employees inside each department (1 = highest salary)
```

```python
df['rank'] = df.groupby('dept')['salary'].rank(ascending=False)
```

```python
df
```

```python
df.groupby('dept').first()
```

```python
🎯 15. Get last record per department
```

```python
df.groupby('dept').last()
```
