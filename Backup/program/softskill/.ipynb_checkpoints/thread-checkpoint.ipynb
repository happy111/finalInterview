```python
# When to use Multi Threading
# I/O bound tasks :

   Tasks that spend more time waiting for i/o operations
# Concurrent execution
```

```python
import threading
import time

def print_number():
    for i in range(4):
        print(f"Number : {i}")

def print_letter():
    for letter in "asfds":
        print(f"Letter : {letter}")
```

```python
t = time.time()
print_number()
print_letter()
finished_time = time.time() - t
print(finished_time)
```

```python
import threading
import time

def print_number():
    for i in range(4):
        time.sleep(2)
        print(f"Number : {i}")

def print_letter():
    for letter in "asfds":
        time.sleep(2)
        print(f"Letter : {letter}")

t1 = threading.Thread(target=print_number)

t2 = threading.Thread(target=print_letter)
# Start thread
t1.start()
t2.start()
# Wait for the thread to complete
t1.join()
t2.join()
```

```python
# Multiprocessing

# processes that run in parallel
# CPU bound tasks that are heavy of CPU usage
# Paralel execution Multiple cores of the CPU
```

```python
import multiprocessing
import time

def print_number():
    print("dddddddddddddd")
    for i in range(4):
        time.sleep(2)
        print(f"Number : {i}")

def print_letter():
    for letter in "asfds":
        time.sleep(2)
        print(f"Letter : {letter}")

if __name__=="__main__":
    print("dddddddddddddd")
    p1 = multiprocessing.Process(target=print_number)
    
    p2 = multiprocessing.Process(target=print_letter)
    # Start thread
    p1.start()
    p2.start()
    # Wait for the thread to complete
    p1.join()
    p2.join()
```

```python
# Thread Pool Executor and Process Pool

from concurrent.futures import ThreadPoolExecutor

import time

def print_number(number):
    time.sleep(1)
    return f"Number : {number}"
```

```python
numbers = [1,2,3,4,5]
with ThreadPoolExecutor(max_workers=3) as executor:
    results = executor.map(print_number,numbers)
for re in results:
    print(re)
```
