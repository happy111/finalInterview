```python
✅ 1. What is FastAPI?

FastAPI is a modern, fast, async web framework for building APIs in Python.
It uses:

ASGI instead of WSGI

Pydantic for data validation

Starlette under the hood

Python async/await for high performance
```

```python
✅ 2. Why is FastAPI faster?

Because it uses:

Async I/O (async/await)

Starlette (high-performance ASGI framework)

Pydantic (fast JSON parsing in C)

Automatic data validation
```

```python
✅ 3. What is ASGI?

ASGI = Asynchronous Server Gateway Interface
It supports async communication, WebSockets, background tasks.

WSGI = old, synchronous only.

FastAPI requires ASGI servers like uvicorn or hypercorn.
```

```python
✅ 5. How to run FastAPI?
uvicorn main:app --reload
```

```python
✅ 6. What is dependency injection in FastAPI?

FastAPI allows functions (dependencies) to be automatically injected into API endpoints.

from fastapi import Depends

def get_db():
    return "DB Connection"

@app.get("/users")
def get_users(db=Depends(get_db)):
    return {"db": db}
```

```python
✅ 7. What are Pydantic models?

Data classes used for:

    request validation
    response validation
    type checking

from pydantic import BaseModel

class User(BaseModel):
    name: str
    email: str
```
