```python
import pandas as pd

data = {
    'Student': ['Alice', 'Bob', 'Charlie', 'David'],
    'Math':   [85, 90, 75, 80],
    'Science': [88, 85, 90, 82],
    'English': [82, 88, 85, 90]
}

df = pd.DataFrame(data)
print(df)
```

```python
df
```

```python

# ---- Child-wise average ----
df['child_average'] = df[['Math', 'Science', 'English']].mean(axis=1)
```

```python
df
```

```python

# ---- Subject-wise average ----
subject_average = df[['Math', 'Science', 'English']].sum()

print(subject_average)
top_subject = subject_average.sort_values(ascending=False).head(2)
```

```python
print(top_subject)
```
