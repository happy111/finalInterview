```python
def solve(string): 
    freq = [0] * 26
    is_odd_present = False
    ans = 0
    for ch in string:
        freq[ord(ch) - ord('a')] += 1
    for count in freq:  # go through all 26 letters
        if count % 2 == 0:
            ans += count
        else:
            is_odd_present = True
            ans += count - 1
    if is_odd_present:
        ans += 1
    return ans

string = "madamda"
result = solve(string)
print(result)  # Output: 3
```

```python
def solve(li):
    for i in range(len(li)):
        for j in range(i,len(li)):
            if li[i] > li[j]:
                li[i],li[j] = li[j],li[i]
    return li

li = [23,45,12,11,3]
result = solve(li)
print(result)
```

```python
def solve(li,k):
    sum = 0
    temp = {}
    ans = []
    for i in range(len(li)):
        start = i+1
        end = len(li)
        new_sum = li[i] + li[start] - k
        new_val = start+1
        while new_val < end:
            if new_sum == new_val:
                ans.append([li[i],li[start],li[new_sum]])
            else:
                temp[new_val] = new_val
    return ans
                

li = [3,4,5,2,1,2]
# find the distinct triplets whose sum = n
# where n= 9
k = 9
result = solve(li,k)
print(result)
```
