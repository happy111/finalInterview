```python
namedtuple  =  is a factory function for creating tuple subclasses 
with named fields.
```

```python
from collections import namedtuple
```

```python
Point = namedtuple('name' , ['x','y'])

# Create instances of Point
p1 = Point(2, 3)

print(p1.x,p1.y)
```

```python
##  Advantages of namedfield

1. Memory efficient
2. Immutable
3. readable
```

```python
## deque(Follow  pop)
from collections import deque
d = deque([1, 2, 3, 4, 5])

# Append elements to both ends
d.append(6)

d.appendleft(0)

# Pop elements from both ends
d.pop()
d.popleft()

print(d)
```

```python
###  counter ##
from collections import Counter
```

```python
cnt = Counter(['apple', 'banana', 'apple', 'orange', 'banana', 'apple'])
print(cnt)
```

```python
# Most common elements
print(cnt.most_common(1))  # Output: [('apple', 3), ('banana', 2)]
```

```python
### defaultdict  ### 

from collections import defaultdict


dd = defaultdict(list)

# Add some values
dd['fruits'].append('apple')
dd['fruits'].append('banana')
dd['vegetables'].append('carrot')

print(dd) 
```

```python
from collections import namedtuple
# Step 1: Define a class with 3 fields
Person = namedtuple("Person", "name age city")
# Step 2: Create an object with 3 values
p = Person(name="Alice", age=30, city="New York")

# Step 3: Access fields
print(p.name)  # Alice
print(p.age)   # 30
print(p.city)  # New York
```
