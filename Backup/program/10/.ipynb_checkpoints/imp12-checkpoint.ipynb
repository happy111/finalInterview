```python
Key Differences
Scope:

Decorators: Used at the function/method level within the same application.
Middleware: Used at the request/response level, typically in web applications to handle HTTP requests and responses.
Usage:

Decorators: Applied directly to functions or methods.
Middleware: Configured in the web framework settings and applied globally or to specific routes.
Functionality:

Decorators: Extend or modify the behavior of specific functions or methods.
Middleware: Provide a mechanism to process and modify requests and responses as they move through the application.
Typical Use Cases:

Decorators: Logging, access control, memoization.
Middleware: Authentication, logging, request modification, response compression.
```

```python
In Python, __str__ and __repr__ are special methods used to define how objects are represented as strings.
They are both used for string representation


__str__: For a user-friendly and readable string representation.
__repr__: For an unambiguous string representation useful for developers, often to recreate the object.
```

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __repr__(self):
        return f"Person('{self.name}', {self.age})"

person = Person("Alice", 30)
print(repr(person))  # Output: Person('Alice', 30)
```

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __str__(self):
        return f"Person named {self.name}, aged {self.age}"

person = Person("Alice", 30)
print(person)  # Output: Person named Alice, aged 30
```

```python
def display(file_name):
    with open(file_name,'r') as fr:
        s = fr.read()
        yield(s)

file_name = "demo.txt"
file_data = display(file_name)
print(file_data)
```
