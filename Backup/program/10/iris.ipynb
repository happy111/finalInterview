```python
def solve(s,n):
    for i in range(len(s)):
        for j in range(i,len(s)):
            if s[i] < s[j]:
                s[i],s[j] = s[j],s[i]
    return s[n-1]

s = [-1,3,4.6,4,7]
n = 2
result = solve(s,n)
print(result)
```

```python
def solve(li):
    sum = 0
    for i in li:
        if isinstance(i,list) == True:
            return sum + solve(i)
        else:
            sum += i
    return sum
```
