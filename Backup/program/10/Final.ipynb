```python
#  Check anagram
def solve(li):
    temp = {}
    for i in li:
        new_word = ''.join(sorted(i))
        if new_word not in temp:
            temp[new_word] = []
        temp[new_word].append(i)
    return list(temp.values())

li = ['ant','tna','abc','bca','der']
result = solve(li)
print(result)
```

```python
## sort dict via key
def solve(temp):
    new_li = list(temp.keys())
    new_li.sort()
    new_temp = {}
    for i in new_li:
        new_temp[i] = temp[i]
    return new_temp

temp = {1:"dd",2:"rwe",6:"vvcd",4:"sfdasf"}
result = solve(temp)
print(result)
```

```python
## sort dict via value
def solve(temp):
    new_li = list(temp.values())
    new_li.sort()

    new_temp = {}
    for i in new_li:
        for k,v in temp.items():
            if i == v:
                new_temp[k] = v
    return new_temp
temp = {1:"dd",2:"rwe",6:"vvcd",4:"aaa"}
result = solve(temp)
print(result)
```

```python
from collections import defaultdict

def group_anagrams(words):
    anagrams = defaultdict(list)
    for word in words:
        # Sort the word and use it as a key
        sorted_word = tuple(sorted(word))
        anagrams[sorted_word].append(word)
    
    # Convert the dictionary values to a list
    return list(anagrams.values())

words = ["eat", "tea", "tan", "ate", "nat", "bat"]
output = group_anagrams(words)
print(output)
```

```python
def solve(str):
    if not str:  # Edge case: empty string
        return 0
    ch = str[0]
    count = 1
    max_count = 1
    
    for i in range(1, len(str)):
        if str[i] == ch:
            count += 1
        else:
            if count > max_count:
                max_count = count
            ch = str[i]
            count = 1
    
    # Final check at the end of the string
    if count > max_count:
        max_count = count
    
    return max_count

str = "AAABBCCCDDDDAAAAAAAAAAAA"
result = solve(str)
print(result)
```

```python
def solve(A):
    temp = []
    for i in A:
        temp.append(len(i))
    temp.sort()
    
    output = []
    for i in temp:
        for k in A:
            if i == len(k):
                output.append(k)
    return output

A = ["This", "is", "a", "wonderful", "day"]
result = solve(A)
print(result)
```

```python
def solve(input):
    input = input.split(" ")
    smallest = ""
    largest = ""
    smallest_len = float('inf')
    largest_len = 0
    for i in input:
        if i == i[::-1]:
            if smallest_len > len(i):
                smallest_len = len(i)
                smallest = i        
            if largest_len < len(i):
                largest_len = len(i)
                largest = i
    print("Smallest palindrom is ", smallest)
    print("Largest palindrom is", largest)

input = "you abbba abcded own wow kayak"
result = solve(input)
```

```python
def solve(coins,target):
    pf = [coins[0]]
    for i in range(1,len(coins)):
        pf.append(pf[i-1] + coins[i])

    min_len = float('inf')
    ans = []
    for s in range(len(coins)):
        for e in range(s,len(coins)):
            sum = 0
            if s == 0:
                sum += pf[e]
            else:
                sum += pf[e] - pf[s-1]
            if sum == target:
                val = coins[s:e+1]
                if min_len > len(val):
                    min_len = len(val)
                    ans = val
    return ans


coins = [1,2,3,5,6,7]
target = 11
result = solve(coins,target)
print(result)
```

```python
def solve(li,k):
    ans = []
    for i in range(len(li)-2):
        temp = {}
        cs = k - li[i]
        for j in range(i+1,len(li)):
            if cs-li[j] in temp:
                ans.append([li[i],li[j],cs-li[j]])
            temp[li[j]] = j
    return ans

li = [3, 4, 5, 2, 1, 2]
k = 9
result = solve(li, k)
# Print the results in the desired format
print(result)
```
