```python
def solve(n, s):
    if n < 1*s or n > 9 * s:
        return "It is not possible"
        
    res = []
    for i in range(s):
        for d in range(9, -1, -1):
            remaining = n - d
            remaining_digits = s - i - 1
            if 0 <= remaining <= 9 * remaining_digits:
                res.append(d)
                n -= d
                break
    return res
print(solve(20, 4))
```

```python
def smallest_number(n,s):
    if n < 1 or n > (9*s):
        return "It is not possilbe"
    n -= 1
    res = [0] * s
    for i in range(s-1,0,-1):
        d = min(9,n)
        res[i] = d
        n -= d
    res[0] = n + 1
    print(res)
print(smallest_number(20, 3))
```

```python
from collections import defaultdict

def solve(event):
    user_event = defaultdict(list)
    for k in event:
        user_event[k['user_id']].append(k)

    duration_map = defaultdict(int)

    for user_id,eve in user_event.items():
        eve.sort(key = lambda x:x['time'])

        for i in range(len(eve)-1):
            cur = eve[i]
            nxt = eve[i+1]
            if cur['channel'] == nxt['channel']:
                diff = nxt['time'] - cur['time']
                key = (user_id, cur['channel'])
                duration_map[key] += diff
    result = [
        {"user_id" : user_id,"channel" : channel,"time" : time}
        for (user_id,channel),time in duration_map.items()
    ]

    return result
events = [
    {'user_id': 1, 'channel': 101, 'time': 10},
    {'user_id': 1, 'channel': 101, 'time': 50},
    {'user_id': 1, 'channel': 101, 'time': 40},
    {'user_id': 2, 'channel': 102, 'time': 10},
    {'user_id': 2, 'channel': 102, 'time': 20},
    {'user_id': 2, 'channel': 102, 'time': 40}
]

output = [
    {'user_id': 1, 'channel': 101, 'duration': 30}
]

res = solve(events)
print(res)
```

```python
from collections import defaultdict

def solve(events):
    user_event = defaultdict(list)

    # group events by user
    for e in events:
        user_event[e['user_id']].append(e)

    duration_map = defaultdict(int)

    for user_id, eve in user_event.items():
        # sort by time
        eve.sort(key=lambda x: x['time'])

        for i in range(len(eve) - 1):
            cur = eve[i]
            nxt = eve[i + 1]

            # ✅ only if same channel
            if cur['channel'] == nxt['channel']:
                diff = nxt['time'] - cur['time']
                key = (user_id, cur['channel'])
                duration_map[key] += diff

    result = [
        {"user_id": user_id, "channel": channel, "duration": duration}
        for (user_id, channel), duration in duration_map.items()
        if duration > 0
    ]

    return result
events = [
    {'user_id': 1, 'channel': 101, 'time': 10},
    {'user_id': 1, 'channel': 101, 'time': 50},
    {'user_id': 1, 'channel': 101, 'time': 40},
    {'user_id': 2, 'channel': 102, 'time': 10},
    {'user_id': 2, 'channel': 102, 'time': 20},
    {'user_id': 2, 'channel': 102, 'time': 40}
]

output = [
    {'user_id': 1, 'channel': 101, 'duration': 30}
]
res = solve(events)
print(res)
```

```python
def solve(st):
    temp = { "[": "]", "{": "}", "(":")" }
    s = []
    for i in st:
        if i in temp:
            s.append(i)
        else:
            if len(s) == 0 or temp[s.pop()] != i:
                return False
    return len(s) == 0

st = "[[()]()]{}"
res = solve(st)
print(res)
```
