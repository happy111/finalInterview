```python
7️⃣ What is TTL in CloudFront?

TTL (Time To Live) determines how long objects are cached at edge locations.
```

```python
def solve(nums):
    mnum = float('-inf')
    snum = float('-inf')
    for i in nums:
        if i > mnum:
            snum = mnum
            mnum = i
        elif i > snum and i != mnum:
            snum = i
    return snum
nums = [10, 20, 4, 45, 99]
res = solve(nums)
print(res)

```

```python
9️⃣ How does CloudFront work?

User requests content
CloudFront checks nearest edge location
If cached → return content
If not cached → fetch from origin
Cache it and deliver to user
```

```python
1️⃣1️⃣ How does CloudFront improve security?

HTTPS (SSL/TLS)
AWS WAF integration
AWS Shield (DDoS protection) (Distributed Denial of Service))
Signed URLs & Signed Cookies
Geo-restriction
```
