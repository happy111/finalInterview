```python
class MyIterator:
    def __init__(self, limit):
        self.limit = limit
        self.current = 0
    def __iter__(self):
        return self

    def __next__(self):
        if self.current < self.limit:
            self.current += 1
            return self.current
        else:
            raise StopIteration

s  = MyIterator(1)
print(next(s))  
print(next(s)) 
```

```python
| Feature       | Iterator                           | Generator                    |
| ------------- | ---------------------------------- | ---------------------------- |
| Creation      | Class with `__iter__` & `__next__` | Function with `yield`        |
| Laziness      | Optional                           | Always lazy                  |
| Memory Usage  | Higher                             | Very low                     |
| Complexity    | More code                          | Very simple                  |
| StopIteration | Manual                             | Automatic                    |
| Easy to write | ❌ Hard                             | ✅ Very Easy                  |
| Best use      | Custom iteration logic             | Streams, pipelines, big data |
```

```python
✅ 5. Use Cases
Iterator

When you want full control over iteration logic.
Complex state management.
Custom objects that need custom iteration.

Generator

When you want:

Simpler code
Lazy loading
Large data sequences
Pipelines
```
