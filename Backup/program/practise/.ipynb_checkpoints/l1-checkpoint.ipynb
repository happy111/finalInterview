```python
# Working of JWT
JWT : Json Web token

  . 1️⃣ User sends login credentials
         User enters username + password.
    2️⃣ Server verifies the user
        The server checks the credentials against the database.
        If valid, the server creates a JWT token.
        If invalid, the server returns an error.
    3️⃣ Server generates a JWT token
         A JWT contains 3 parts:
         Header → algorithm + token type
         Payload → user data (id, email, roles)
         Signature → created using secret key
    4️⃣ Server sends the JWT back to the client
        The token is sent in the login response.
        The client stores it (usually in localStorage, sessionStorage, or HTTP-only cookies).
    5️⃣ Client sends JWT with every request
    6️⃣ Server verifies the JWT
       The server extracts the token from the header.
    7️⃣ If the token is valid
        The request is allowed.
    8️⃣ If the token is invalid
```

```python
# Write program for Polymorphism explainh method overriding and overloading

# Polymorphism = same function name, different behavior depending on the object.

class Dog:
    def sound(self):
        return "Bark"

class Cat:
    def sound(self):
        return "Meow"

# Polymorphism in action
for animal in (Dog(), Cat()):
    print(animal.sound())
```

```python
# 🔥 2. Method Overriding (Run-time Polymorphism)

# Child class overrides parent class method.

class Vehicle:
    def start(self):
        print("Vehicle is starting")

class Car(Vehicle):
    # overriding parent method
    def start(self):
        print("Car is starting")

obj = Car()
obj.start()
```

```python
#Write a program for falatening a nested list

def solve(l):
    res = []
    for k in l:
        if isinstance(k,list):
            res.extend(solve(k))
        else:
            res.append(k)
    return res
            

s = [1,2,3,4,5,[4,5]]
res = solve(s)
print(res)
```

```python

def solve(l):
    res = 0
    for k in l:
        if isinstance(k,list):
            res += solve(k)
        else:
            res += k
    return res
            

s = [1,2,3,4,5,[4,5]]
res = solve(s)
print(res)
```

```python
import boto3
import csv
import os
import pymysql
from datetime import datetime

s3 = boto3.client("s3")

# RDS/MySQL connection details
DB_HOST = os.environ["DB_HOST"]
DB_USER = os.environ["DB_USER"]
DB_PASS = os.environ["DB_PASS"]
DB_NAME = os.environ["DB_NAME"]

BUCKET_NAME = "employee-monthly-reports"


def lambda_handler(event, context):
    # 1️⃣ Create DB connection
    conn = pymysql.connect(
        host=DB_HOST,
        user=DB_USER,
        password=DB_PASS,
        db=DB_NAME
    )

    cursor = conn.cursor()

    # 2️⃣ Fetch employee data
    query = """
        SELECT name, join_date, exit_date
        FROM employees;
    """
    cursor.execute(query)
    rows = cursor.fetchall()

    # 3️⃣ Generate CSV in memory
    month = datetime.now().strftime("%Y_%m")
    filename = f"employee_report_{month}.csv"

    # Create CSV file locally inside Lambda temp storage
    temp_path = f"/tmp/{filename}"

    with open(temp_path, "w", newline="") as file:
        writer = csv.writer(file)
        writer.writerow(["name", "join_date", "exit_date"])  # header
        for r in rows:
            writer.writerow(r)

    # 4️⃣ Upload to S3
    s3.upload_file(
        temp_path,
        BUCKET_NAME,
        f"reports/{month}/{filename}"
    )

    return {
        "statusCode": 200,
        "body": f"Report {filename} uploaded to S3 successfully!"
    }
```

```python
import boto3
import csv
import os
import pymysql
from datetime import datetime, timedelta

s3 = boto3.client("s3")

# ENV Variables
DB_HOST = os.environ["DB_HOST"]
DB_USER = os.environ["DB_USER"]
DB_PASS = os.environ["DB_PASS"]
DB_NAME = os.environ["DB_NAME"]
BUCKET_NAME = "employee-reports"

def get_month_range():
    """Return first and last day of previous month."""
    today = datetime.today()
    first_day_current = today.replace(day=1)
    last_day_prev = first_day_current - timedelta(days=1)
    first_day_prev = last_day_prev.replace(day=1)

    return first_day_prev.date(), last_day_prev.date()


def lambda_handler(event, context):

    start_date, end_date = get_month_range()
    month_label = start_date.strftime("%Y_%m")

    # 1️⃣ Connect to DB
    conn = pymysql.connect(
        host=DB_HOST,
        user=DB_USER,
        password=DB_PASS,
        db=DB_NAME
    )
    cursor = conn.cursor()

    # 2️⃣ Fetch Employees Joined in Month
    query_joined = """
        SELECT name, join_date
        FROM employees
        WHERE join_date BETWEEN %s AND %s;
    """
    cursor.execute(query_joined, (start_date, end_date))
    joined_rows = cursor.fetchall()

    # 3️⃣ Fetch Employees Exited in Month
    query_exited = """
        SELECT name, exit_date
        FROM employees
        WHERE exit_date BETWEEN %s AND %s;
    """
    cursor.execute(query_exited, (start_date, end_date))
    exited_rows = cursor.fetchall()

    # 4️⃣ Create CSV in /tmp/
    filename = f"employee_report_{month_label}.csv"
    csv_path = f"/tmp/{filename}"

    with open(csv_path, "w", newline="") as f:
        writer = csv.writer(f)

        # Joined
        writer.writerow([f"Employees Joined in {month_label.replace('_','-')}"])
        writer.writerow(["name", "join_date"])
        for row in joined_rows:
            writer.writerow(row)
        writer.writerow([])

        # Exited
        writer.writerow([f"Employees Exited in {month_label.replace('_','-')}"])
        writer.writerow(["name", "exit_date"])
        for row in exited_rows:
            writer.writerow(row)

    # 5️⃣ Upload the CSV to S3
    s3_key = f"{start_date.year}/{start_date.month:02}/{filename}"

    s3.upload_file(csv_path, BUCKET_NAME, s3_key)

    return {
        "statusCode": 200,
        "message": f"Report uploaded to s3://{BUCKET_NAME}/{s3_key}"
    }
```

```python
import boto3
import pymysql
import csv
import os
from datetime import datetime, timedelta

s3 = boto3.client("s3")

DB_HOST = os.environ["DB_HOST"]
DB_USER = os.environ["DB_USER"]
DB_PASS = os.environ["DB_PASS"]
DB_NAME = os.environ["DB_NAME"]
BUCKET_NAME = "employee-reports"

def lambda_handler(event, context):

    # Calculate previous month
    today = datetime.today()
    first_day_current_month = today.replace(day=1)
    last_month = first_day_current_month - timedelta(days=1)
    
    year = last_month.year
    month = last_month.month
    ym = last_month.strftime("%Y_%m")
    
    # Connect to RDS
    conn = pymysql.connect(
        host=DB_HOST, user=DB_USER,
        password=DB_PASS, db=DB_NAME
    )
    cursor = conn.cursor()

    # Employees Joined
    join_query = f"""
        SELECT name, join_date, exit_date, 'JOINED'
        FROM employees
        WHERE MONTH(join_date) = {month} AND YEAR(join_date) = {year};
    """
    
    cursor.execute(join_query)
    joined = cursor.fetchall()

    # Employees Left
    left_query = f"""
        SELECT name, join_date, exit_date, 'LEFT'
        FROM employees
        WHERE MONTH(exit_date) = {month} AND YEAR(exit_date) = {year};
    """
    
    cursor.execute(left_query)
    left = cursor.fetchall()

    # Combine both lists
    report_data = list(joined) + list(left)

    # Filename
    filename = f"employee_monthly_report_{ym}.csv"
    temp_path = f"/tmp/{filename}"

    # Write CSV
    with open(temp_path, "w", newline="") as csv_file:
        writer = csv.writer(csv_file)
        writer.writerow(["name", "join_date", "exit_date", "status"])
        for row in report_data:
            writer.writerow(row)

    # Upload to S3
    s3.upload_file(
        temp_path,
        BUCKET_NAME,
        f"reports/{year}/{month:02d}/{filename}"
    )

    return {
        "statusCode": 200,
        "body": f"Report {filename} uploaded successfully."
    }
```

```python
.design customer table,product table to handel transaction data ands any other
table needed for comercial purpose keep in mind normalization,
like customer is purchasing multiple product or 1 product purchased by multiple customer

Customer :
    CREATE TABLE customers (
        customer_id INT PRIMARY KEY AUTO_INCREMENT,
        full_name VARCHAR(100) NOT NULL,
        email VARCHAR(100) UNIQUE,
        phone VARCHAR(20),
        created_at DATETIME DEFAULT CURRENT_TIMESTAMP
    );

Category :

CREATE TABLE categories (
    category_id INT PRIMARY KEY AUTO_INCREMENT,
    category_name VARCHAR(100) NOT NULL
);

Product:- 

CREATE TABLE products (
    product_id INT PRIMARY KEY AUTO_INCREMENT,
    product_name VARCHAR(150) NOT NULL,
    product_price DECIMAL(10,2) NOT NULL,
    category_id INT,
    created_at DATETIME DEFAULT CURRENT_TIMESTAMP,
    FOREIGN KEY (category_id) REFERENCES categories(category_id)
);

Order:- 

CREATE TABLE orders (
    order_id INT PRIMARY KEY AUTO_INCREMENT,
    customer_id INT NOT NULL,
    order_date DATETIME DEFAULT CURRENT_TIMESTAMP,
    status VARCHAR(50),
    total_amount DECIMAL(10,2),
    FOREIGN KEY (customer_id) REFERENCES customers(customer_id)
);

Order_items :-

CREATE TABLE order_items (
    order_item_id INT PRIMARY KEY AUTO_INCREMENT,
    order_id INT NOT NULL,
    product_id INT NOT NULL,
    quantity INT NOT NULL,
    price_at_purchase DECIMAL(10,2) NOT NULL,
    FOREIGN KEY (order_id) REFERENCES orders(order_id),
    FOREIGN KEY (product_id) REFERENCES products(product_id)
);
```
