```python
def factor(i):
    count = 0
    for k in range(1,i):
        try:
            if i % k == 0:
                count += 1
        except Exception as e:
            print(e)
    return count

res = factor(6)
print(res)
```

```python
def solve(li):
    temp = {}
    for i in li:
        temp[i] = temp.get(i,0)+ 1
  #  print(temp)
    output = {}
    for k in temp.items():
        if k[1] > 1:
            output[k[0]] = k[1]
    return output

li =  [1,2,3,3,3,4,4,4,4,5,6,7,8,9]
#3:3 4:4

res = solve(li)
print(res)
```

```python
def factor(i):
    count = 0
    for k in range(1,i):
        try:
            if i % k == 0:
                count += 1
        except Exception as e:
            print(e)
    return count

def solve(n):
    output = []
    for i in range(2,n+1):
        count = factor(i)
        if count == 1:
            output.append(i)
    return output

n = int(input("Enter a number"))
res = solve(n)
print(res)
```

```python
i/p = " I love Python Programing"
o/p= "I evoL nohtyP gnimargorP"
def reverse_str(str1):
    str1 = str1.split(" ");
    str2 = []
    for i in str1:
        str2.append(i[::-1])
    return ' '.join(str2)


str1 = "I love Python Programing"
result = reverse_str(str1)
print(result)
```

```python
def solve(st):
    char = ['a','b','c','d','e','f','g','h']
    output='';
    for i in st:
        if i.isalpha() != True:
            output = output + char[int(i)]
        else:
            output = output + i
    return output

st= "a4m3r2"
result = solve(st)
print(result)
```
