```python
Commonly Used Functions in itertools
************************************
product: Cartesian product of input iterables.
permutations: All possible permutations of an iterable.
combinations: All possible combinations of a specified length from the iterable.
combinations_with_replacement: All possible combinations of a specified length from the iterable, 
                               allowing individual elements to be repeated.
accumulate: Accumulate sums (or other binary functions) over an iterable.
groupby: Group elements of the iterable by a specified key function.
islice: Slice an iterator to extract selected items by index.
cycle: Cycle through an iterable indefinitely.
count: Count indefinitely from a specified number.
repeat: Repeat a single element indefinitely or a specified number of times.
```

```python
#product
#*********

from itertools import product

a = [1,2,3]
b = [4,5,6]

s = product(a,b)
print(s)
print(tuple(s))
```

```python
# permutations


from itertools import permutations

a = [1,1,2]

p = list(permutations(a))

print(p)
```

```python
from itertools import combinations

a = [1, 2, 3]
comb = list(combinations(a, 2))
print(comb)
```

```python
from itertools import permutations

# Read input
s, k = input().split()
k = int(k)

# Generate permutations of the given length
perm = permutations(s, k)

print(type(perm))

for p in sorted(perm):
    print(''.join(p))
    
# input abc 2
```

```python
from itertools import combinations
s,n = input().split()
k = int(n)

perm = combinations(s, k)


for p in sorted(perm):
    print(''.join(p))
```
