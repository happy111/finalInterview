```python
## Reverse of numpy array

import numpy

def arrays(arr):
    # complete this function
    # use numpy.array
    arr = numpy.array(arr,float)
    
    reversed_arr = numpy.flip(arr)

    return reversed_arr
arr = input().strip().split(' ')
result = arrays(arr)
print(result)
```

```python
import numpy as np

# Read the input values
N, M = map(int, input().split())

# Create the matrix with main diagonal as 1's and all other elements as 0's
identity_array = np.zeros((N, M), dtype=int)

# Set the main diagonal to 1's
for i in range(min(N, M)):
    if N % 2 != 0:
        identity_array[i, i] = 1
    else:
        if i+1 < M:
            identity_array[i, i+1] = 1

        
        

# Print the resulting array
print(identity_array)
```
