```python
# 16  11  6  1 -4  1 6 11 16

def printPattern(n,m,flag):
    print(m,end=" ")
    if n == m and flag == False:
        return
    if flag:
        if m-5 > 0 and flag == True:
            printPattern(n,m-5,True)
        else:
            printPattern(n,m-5,False)
    else:
        printPattern(n,m+5,False)

# Driver Code 
if __name__ == "__main__": 
  n = 16
  printPattern(n, n, True) 
```

```python
def commonPrefix(li):
    s = sorted(li)
    first = s[0]
    last = s[-1]
    min_len = min(len(first),len(last))
    i = 0
    while i < min_len and first[i] == last[i]:
        i += 1
    return first[0:i]
li = ["geeksforgeeks", "geeks", "geek", "geezer"]
result = commonPrefix(li)
print(result)
```

```python
def solve(li):
    temp = {}
    for i in li:
        new_str = ''.join(sorted(i))
        if new_str not in temp:
            temp[new_str] = []
        temp[new_str].append(i)
    ans = []
    for i in temp.values():
        if len(i) > 1:
            ans.append(i)
        else:
            ans.append(i[0])
    return ans


li = ['ant','tna','abc','bca','der']
result = solve(li)
print(result)
```

```python
def solve(li):
    space_index = [i for i in range(len(li)) if li[i]==" "]
    ch_index = [li[i] for i in range(len(li)) if li[i] !=" "]
    new_str = ch_index[::-1]
    print(new_str)
    for i in space_index:
        new_str.insert(i," ")
    return ''.join(new_str)


input_str = "abc de ku"
result = solve(input_str)
print(result)

```

```python
def solve(string):
    map = [0] * 26
    n = len(string)
    for i in range(n):
        ch = ord(string[i]) - ord('a')
        map[ch] += 1
    print(map)
    
string = "abaw"
result = solve(string)
print(result)
```

```python
def longest_valid_permutation(input_string):
    max_len = 0
    start = 0
    temp = {}
    for end in range(len(input_string)):
        char = input_string[end]
        # If character is already in the window, shrink from the left
        if char in temp and temp[char] >= start:
            start = temp[char] + 1
        temp[char] = end
        max_len = max(max_len, end - start + 1)
    return max_len

# Example usage:
input_string = "abcabc"
result = longest_valid_permutation(input_string)
print("Longest valid permutation length:", result)
```

```python
def solve(res):
    right = len(res) - 1
    sc = [i for i in res if i.isdigit()][::-1]
    k = 0
    ans = ""
    for i in res:
        if i.isalnum() == False:
            ans += i
        else:
            ans += sc[k]
            k += 1
    return ans
res = "@#2345&98%"
res = solve(res)
print(res)
```

```python
from collections import Counter

def topMost(names, k):
    # Count frequency of each element
    freq = Counter(names)
    print(freq)
    # Get the k most common elements
    return [item[0] for item in freq.most_common(k)]

names = [1, 1, 1, 2, 2, 3,3,3,3]
k = 2
result = topMost(names, k)
print(result)
```

```python
Use namedtuple to create simple classes with named fields:
from collections import namedtuple
Person = namedtuple("Person", ["name", "age"])
person = Person("Alice", 30)
print(person.name) # Output: "Alice"
print(person.age) # Output: 30
```
