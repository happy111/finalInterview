```python
def extract_data(li):
    output = {}
    for i in li:
        key, value = map(str.strip, i.split(':', 1))
        if key == "tags":
            value = [tag.strip().strip("'") for tag in value.split(',')]
        output[key] = value
    return output



li = ["name : umesh","tags: 'a','b'"]
result = extract_data(li)
print(result)
```

```python
import sys

# Binary converter function
def decimal_binary(n):
    if n == 0:
        return "00000000000000000000000000000000"  # Special case for 0
    result = ""
    while n > 0:
        result = str(n % 2) + result
        n = n // 2
    
    # Add leading zeros if the binary length is less than 32
    result = result.zfill(32)  # Ensure the result is exactly 32 bits long
    
    return result

# Read input from stdin
try:
    # Read all input at once and split it by newlines
    input_data = sys.stdin.read().strip().splitlines()
    
    # Iterate over each line of input (each test case)
    for line in input_data:
        n = int(line.strip())  # Convert each line to an integer
        result = decimal_binary(n)  # Get the binary representation
        print(result)  # Print the result for each number
    
except ValueError as ve:
    print(f"Invalid input: {ve}")
except EOFError:
    print("No input provided")
decimal_binary(5)
```

```python
def is_leap(year):
    leap = False
    if year % 4 == 0: 
        if year % 100 == 0:
            if year % 400 == 0:
                leap = True
            else:
                leap = False
        else:
            leap = True
    else:
        leap = False
            

    
    return leap

year = int(input())
print(is_leap(year))
```

```python
# Read input
n, m = map(int, input().split())  # n = size of array, m = size of dislike set
arr = list(map(int, input().split()))  # array of integers
like_set = set(map(int, input().split()))  # "like" set
dislike_set = set(map(int, input().split()))  # "dislike" set

# Initialize happiness
happiness = 0

# Calculate happiness based on array elements
for num in arr:
    if num in like_set:
        happiness += 1
    elif num in dislike_set:
        happiness -= 1

# Output the final happiness
print(happiness)

```

```python
def print_formatted(number):
    # Find the width of the binary representation of the largest number
    width = len(bin(number)) - 2
    for i in range(1, number + 1):
        decimal_val = str(i)
        octal_val = oct(i)[2:]
        hex_val = hex(i)[2:].upper()
        binary_val = bin(i)[2:]
        
        print(f"{decimal_val:>{width}} {octal_val:>{width}} {hex_val:>{width}} {binary_val:>{width}}")

# Sample Input
n = 17
print_formatted(n)
```

```python
def print_rangoli(size):
    import string
    alphabet = string.ascii_lowercase

    # Calculate the width of the full rangoli
    max_width = 4 * size - 3
    
    # Generate the upper part of the rangoli
    lines = []
    for i in range(size):
        s = '-'.join(alphabet[size-1:i:-1] + alphabet[i:size])
        lines.append(s.center(max_width, '-'))
    
    # Generate the full rangoli by adding the lower part (reverse of upper part minus the middle line)
    full_rangoli = lines[-1::-1] + lines[1:]
    
    # Print the rangoli
    for line in full_rangoli:
        print(line)

# Example usage
print_rangoli(5)
```

```python
def mutate_string(string, position, character):
    s = list(string)
    s[position] = character
    string = ''.join(s)
    return string

if __name__ == '__main__':
    s = input()
    i, c = input().split()
    s_new = mutate_string(s, int(i), c)
    print(s_new)
```

```python
if __name__ == '__main__':
    # Read the number of commands
    N = int(input().strip())
    
    # Initialize the list
    lst = []
    
    # Iterate through each command
    for _ in range(N):
        command = input().strip().split()
        action = command[0]
        
        if action == "insert":
            index = int(command[1])
            value = int(command[2])
            lst.insert(index, value)
        elif action == "print":
            print(lst)
        elif action == "remove":
            value = int(command[1])
            lst.remove(value)
        elif action == "append":
            value = int(command[1])
            lst.append(value)
        elif action == "sort":
            lst.sort()
        elif action == "pop":
            lst.pop()
        elif action == "reverse":
            lst.reverse()
```

```python
def swap_case(s):
    result = ""  # Initialize an empty string to store the result
    for i in s:
        if i.islower():  # If the character is lowercase
            result += chr(ord(i) - 32)  # Convert to uppercase by subtracting 32 from ASCII value
        elif i.isupper():  # If the character is uppercase
            result += chr(ord(i) + 32)  # Convert to lowercase by adding 32 to ASCII value
        else:
            result += i  # If it's neither, just add the character as is
    return result  # Return the modified string

if __name__ == '__main__':
    s = input()  # Read the input string
    result = swap_case(s)  # Swap the case of all characters
    print(result)  # Print the result
```

```python
def count_substring(string, sub_string):
    c = 0
    for i in range(len(string) - len(sub_string) + 1):
        if string[i:i+len(sub_string)] == sub_string:
            c += 1
    return c
            

if __name__ == '__main__':
    string = input().strip()
    sub_string = input().strip()
    
    count = count_substring(string, sub_string)
    print(count)
```

```python
def check_string_properties(s):
    print(any(c.isalnum() for c in s))  # Checks if any character is alphanumeric
    print(any(c.isalpha() for c in s))  # Checks if any character is alphabetical
    print(any(c.isdigit() for c in s))  # Checks if any character is a digit
    print(any(c.islower() for c in s))  # Checks if any character is lowercase
    print(any(c.isupper() for c in s))  # Checks if any character is uppercase

if __name__ == '__main__':
    s = input().strip()  # Read input string
    check_string_properties(s)
```

```python

def wrap(string, max_width):
    return '\n'.join([string[i:i+max_width] for i in range(0, len(string), max_width)])

if __name__ == '__main__':
    string, max_width = input(), int(input())
    result = wrap(string, max_width)
    print(result)
```

```python
def door_mat(n, m):
    # Upper half of the mat (before 'WELCOME')
    for i in range(1, n, 2):
        pattern = ('.|.' * i).center(m, '-')

    # Middle line with 'WELCOME'
    welcome_line = 'WELCOME'.center(m, '-')
    print(welcome_line)

    # Lower half of the mat (after 'WELCOME')
    for i in range(n - 2, -1, -2):
        pattern = ('.|.' * i).center(m, '-')
        print(pattern)

if __name__ == '__main__':
    # Read input values
    n, m = map(int, input().split())
    
    # Call the function to print the door mat
    door_mat(n, m)
    

    
```
