```python
class Node:
    def __init__(self,data):
        self.next = None
        self.data = data
class LinkedList:
    def __init__(self):
        self.head = None
    def insertAtbeg(self,data):
        self.newdata = Node(data)
        self.newdata.next = self.head
        self.head = self.newdata
    def display(self):
        temp = self.head
        while temp != None:
            print(temp.data, end= " ")
            temp = temp.next
    def reverse(self):
        prev = None
        curr = self.head
        umesh = self.head
        fut = None
        while curr != None:
            fut = curr.next
            curr.next = prev
            prev = curr
            curr = fut
        self.head = prev



obj = LinkedList()
obj.insertAtbeg(1)
obj.insertAtbeg(2)
obj.insertAtbeg(3)
obj.display()
obj.reverse()
print("***********")
obj.display()
```
