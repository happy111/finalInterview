```python
 # Create a dictionary using list comprehension with square of numbers.

squares = {x: x**2 for x in range(1, 6)}
print(squares)

s = list(map(lambda x: x**2, [1, 2, 3]))
print(s)  # Output: [1, 4, 9]
```

```python
# Extract all values of a specific key from a list of dicts.
data = [
    {'name': 'Alice', 'age': 30},
    {'name': 'Bob', 'age': 25},
    {'name': 'Charlie', 'age': 35}
]

s = [d['name'] for d in data]
s
```

```python
#Convert two lists into a dictionary (one list of keys, one of values).

keys = ['name', 'age', 'city']
values = ['Alice', 30, 'New York']

result = dict(zip(keys, values))
print(result)
```

```python
dict1 = {'a': 1, 'b': 2, 'c': 3}
dict2 = {'a': 1, 'b': 4, 'd': 3}

intersection = {k: v for k, v in dict1.items() if k in dict2 and dict2[k] == v}
print(intersection)
```

```python
What is "blocking"?
Blocking means that a function or task stops everything until it 
finishes — it does not allow other work to proceed during that time.

import time

def download():
    print("Start download")
    time.sleep(5)  # BLOCKING call
    print("Download complete")

download()
print("Next task")
```

```python
async:-

1. Uses a single thread with non-blocking I/O.
2. I/O-bound tasks (e.g., network calls, file reads/writes, database queries).
3. Functions are marked with async, and use await to pause/resume execution.
4. Non-blocking


Multithreading
**************
    1.  Uses multiple threads within a process.
    2.  Threads run truly in parallel on multi-core CPUs 
    3.  blocking
```

```python
import asyncio

# Correct async functions
async def fun():
    await asyncio.sleep(4)
    print("Function 1")

async def fun1():
    await asyncio.sleep(4)
    print("Function 2")

async def fun2():
    await asyncio.sleep(4)
    print("Function 3")

# Main async function
async def main():
    await asyncio.gather(fun1(), fun2())

# If you're in a Jupyter cell, use this:
await main()
```
