```python
def traverse_nested_dict(nested_dict,indent=0):
    for key,value in nested_dict.items():
        print("  " * indent + f"Key: {key}")
        if isinstance(value,dict):
            traverse_nested_dict(value,indent+1)
        else:
            print("  " * (indent + 1) + f"Value: {value}")
nested_dict = {
    'a': 1,
    'b': {
        'b1': 2,
        'b2': {
            'b21': 3,
            'b22': 4
        },
        'b3': 5
    },
    'c': {
        'c1': {
            'c11': {
                'c111': 6
            }
        }
    }
}

# Run the recursive traversal
traverse_nested_dict(nested_dict)
```

```python
def solve(data):
    seen = set()
    res = []
    for d in data:
        tup = tuple(sorted(d.items()))
        if tup not in seen:
            seen.add(tup)
            res.append(d)
    return res

data = [
    {'name': 'Alice', 'age': 30},
    {'name': 'Bob', 'age': 25},
    {'name': 'Alice', 'age': 30},  # duplicate
    {'name': 'Charlie', 'age': 35}
]
res = solve(data)
print(res)
```

```python
# How do you remove duplicates from a list of dictionaries?

def remove(dicts):
    seen = set()
    result = []
    for d in dicts:
        # Convert dict to a sorted tuple of key-value pairs
        tup = tuple(sorted(d.items()))
        if tup not in seen:
            seen.add(tup)
            result.append(d)
    return result

data = [
    {'name': 'Alice', 'age': 30},
    {'name': 'Bob', 'age': 25},
    {'name': 'Alice', 'age': 30},  # duplicate
    {'name': 'Charlie', 'age': 35}
]
res = remove(data)
print(res)
```

```python

from collections import defaultdict

def group_by_field(data, field):
    grouped = defaultdict(list)
    for item in data:
        key = item.get(field)
        grouped[key].append(item)
    return dict(grouped)

employees = [
    {'name': 'Alice', 'department': 'HR'},
    {'name': 'Bob', 'department': 'IT'},
    {'name': 'Charlie', 'department': 'HR'},
    {'name': 'Dave', 'department': 'Finance'}
]

grouped = group_by_field(employees, 'department')
print(grouped)
```

```python
 # How to update all values in a nested dictionary where keys match a given pattern?

def update_nested_values(d, target_key, new_value):
    for key, value in d.items():
        if key == target_key:
            d[key] = new_value
        elif isinstance(value, dict):
            update_nested_values(value, target_key, new_value)

data = {
    'id': 1,
    'info': {
        'status': 'pending',
        'details': {
            'status': 'in_progress',
            'comments': 'Work started'
        }
    }
}

update_nested_values(data, 'status', 'complete')
print(data)
```

```python
 # Write a function that flattens a nested dictionary.

def flatten_dict(d, parent_key='', sep='.'):
    items = {}
    for key, value in d.items():
        new_key = f"{parent_key}{sep}{key}" if parent_key else key
        if isinstance(value, dict):
            items.update(flatten_dict(value, new_key, sep=sep))
        else:
            items[new_key] = value
    return items

nested = {
    'a': 1,
    'b': {
        'c': 2,
        'd': {
            'e': 3
        }
    },
    'f': 4
}

flattened = flatten_dict(nested)
print(flattened)
```

```python
# Given a nested dict, count total number of leaf key-value pairs.
def count_leaf_pairs(nested_dict):
    count = 0

    for key,value in nested_dict.items():
        if isinstance(value,dict):
            count+= count_leaf_pairs(value)
        else:
            count +=1
    return count
nested_dict = {
    'a': 1,
    'b': {
        'c': 2,
        'd': {
            'e': 3,
            'f': 4
        }
    },
    'g': 5
}

print(count_leaf_pairs(nested_dict)) 
```

```python
In Python, dictionary keys must be immutable (i.e., hashable). 
If you try to use a mutable type (like a list, dict, or set)
as a key, you'll get a TypeError.
```

```python
Why?
Dictionary keys are stored in a hash table.

Only immutable and hashable types like int, float, str, 
tuple (with only hashable elements), and frozenset can be used as keys.
```

```python
How to invert a dictionary (keys become values and vice versa)?

Basic Inversion (when values are unique and hashable):

original = {'a': 1, 'b': 2, 'c': 3}
inverted = {v: k for k, v in original.items()}

print(inverted)
# Output: {1: 'a', 2: 'b', 3: 'c'}
```

```python
1. Single key with max value

my_dict = {'a': 10, 'b': 25, 'c': 18}
max_key = max(my_dict, key=my_dict.get)
print(max_key)  # Output: 'b'
```

```python
2. All keys with the max value (if duplicates)

my_dict = {'a': 25, 'b': 10, 'c': 25}
max_val = max(my_dict.values())
max_keys = [k for k, v in my_dict.items() if v == max_val]

print(max_keys)  # Output: ['a', 'c']
```

```python
# Write a recursive function to print all values in a nested dictionary.

def print_values(d):
    for key, value in d.items():
        if isinstance(value, dict):
            print_values(value)  # recursive call
        else:
            print(value)

# Example usage:
nested_dict = {
    'a': 1,
    'b': {
        'c': 2,
        'd': {
            'e': 3,
            'f': 4
        }
    },
    'g': 5
}

print_values(nested_dict)
```
