```python
Q1 Answer: Django request–response lifecycle

. Client sends an HTTP request
. Request reaches Django through WSGI or ASGI server
. Request passes through request middleware
    Authentication
    Logging
    Security checks
. URL dispatcher resolves the requested URL
. URL is mapped to the appropriate view
. View executes business logic
. View interacts with models using Django ORM
. View may:
    Render HTML templates, or
    Serialize data (JSON response)
. Django generates the HTTP response
. Response passes through response middleware

    Add headers
    Compression
. Django sends the final HTTP response back to the client
```

```python
class Order(models.Model):
    user = models.ForeignKey(User, on_delete=models.CASCADE)

orders = Order.objects.select_related('user')


class Product(models.Model):
    orders = models.ManyToManyField(Order)
```

```python
Explain the difference between:

Serializer

ModelSerializer

👉 When would you avoid ModelSerializer in real projects?

In Django REST Framework, a Serializer is used to convert complex Python objects such as querysets or model instances into
JSON and also to validate and deserialize incoming data.

A ModelSerializer is a shortcut that automatically generates serializer fields and validation rules 
based on a Django model, reducing boilerplate code.

I would avoid ModelSerializer when I need full control over validation, fields, or business logic,
such as when dealing with multiple models, 
complex nested input, external APIs, or performance-critical serializers.

class UserSerializer(serializers.ModelSerializer):
    class Meta:
        model = User
        fields = ['id', 'username', 'email']

class LoginSerializer(serializers.Serializer):
    username = serializers.CharField()
    password = serializers.CharField(write_only=True)
```

```python
Performance & Optimization (Very Important)
Q6.
Your Django REST API is slow in production.
How would you debug and optimize performance?

. Layered debugging approach
    Diagnose performance issues step by step instead of guessing
    
. Application level
  Check for inefficient or unnecessary code execution

. Enable Django query logging to identify:
  N+1 query problems
  Fix N+1 queries using:
    select_related
    prefetch_related
. Apply pagination for large datasets to reduce response size

. Database level
    Analyze slow queries using EXPLAIN
    Add proper indexes on frequently filtered or joined columns
    Optimize query filters and ORM usage
. Caching level
. Infrastructure level
     Monitor:
        CPU usage
        Memory usage
        Worker utilization
    
    Tune Gunicorn/Uvicorn workers
    
    Use async views where applicable
    
    Offload long-running tasks to background workers like Celery
        
. Outcome
    Improved API response time
    Better scalability and stability in production
```

```python
How do you prevent race conditions in Django when multiple requests update the same record?

👉 Mention:
1️⃣ Database Transactions (Atomic Operations)

  Use transaction.atomic() to ensure all operations execute as a single unit of work.

    from django.db import transaction

    with transaction.atomic():
        order = Order.objects.get(id=1)
        order.quantity -= 1
        order.save()  
        
2️⃣ Row-Level Locking (select_for_update)

from django.db import transaction

with transaction.atomic():
    order = Order.objects.select_for_update().get(id=1)
    order.quantity -= 1
    order.save()
```

```python
Difference bet decorator,middlewar,signal

Decorator:-

A decorator is a function that wraps another function or method to add extra behavior before or after execution.

➡️ In Django, decorators are mostly used at view level.

✅ Where it works

Function-based views
Class-based views (using method_decorator)

✅ Common Django examples
    
    @login_required
    @permission_classes
    @csrf_exempt
    @api_view

✅ Use cases
    ✔ Authentication
    ✔ Authorization
    ✔ Logging
    ✔ Caching
    ✔ Rate limiting (per view)
```

```python
Middleware :-

is a component that processes requests and responses globally, before reaching the view and after leaving the view.

    Request → Middleware → View → Middleware → Response
```

```python
class CustomMiddleware:
    def __init__(self, get_response):
        self.get_response = get_response

    def __call__(self, request):
        print("Before view")
        response = self.get_response(request)
        print("After view")
        return response
```

```python
✅ Common Django middleware

    AuthenticationMiddleware
    SessionMiddleware
    CsrfViewMiddleware
    SecurityMiddleware

✅ Use cases
    ✔ Authentication (JWT, session)
    ✔ Logging
    ✔ Rate limiting
    ✔ Request/response modification
    ✔ Security headers
    ✔ Performance monitoring
```

```python
3️⃣ Signal

    Signals allow decoupled communication between components.
    One action triggers another automatically, without direct call.

    ✅ Common Django signals

    pre_save
    post_save
    pre_delete
    post_delete

✅ Use cases

✔ Send email after user registration
✔ Create profile automatically
✔ Audit logging
✔ Cache invalidation
✔ Async tasks trigger (Celery)
```

```python
| Feature   | Decorator         | Middleware           | Signal             |
| --------- | ----------------- | -------------------- | ------------------ |
| Scope     | View-level        | Global               | Event-based        |
| Trigger   | Function call     | Every request        | Model/event action |
| Use case  | Auth, permission  | Logging, security    | Post-save actions  |
| Execution | Before/after view | Before/after request | Async-style        |
| Coupling  | Tight             | Global               | Loosely coupled    |
```

```python
✔ Use Decorator when:

    Logic applies to specific APIs
    Authentication/authorization
    Per-view rate limiting

✔ Use Middleware when:

    Logic applies to all requests
    JWT validation
    Logging
    Request modification

✔ Use Signal when:

    Action should happen automatically
    After save/delete
    No direct dependency
```
