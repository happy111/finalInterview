```python
1️⃣ What is Django Template Language (DTL)?

DTL is Django’s built-in templating system used to generate
dynamic HTML by combining static templates with context data passed from views.

View receives request
View prepares context data (dictionary)
Template engine loads .html file
Variables & tags are processed
Final HTML is returned to client
```

```python
from rest_framework.generics import (
    CreateAPIView,
    ListAPIView,
    DestroyAPIView,
    UpdateAPIView,
)

class AreaUpdation(UpdateAPIView):
    # permission_classes = (IsAuthenticated, IsSuperAdmin | IsAdmin)
    serializer_class = AreaSerializer
    def put(self, request):
        try:
            data = request.data
            area = Area.objects.filter(id=data["id"])
            if area.count() == 0:
                raise Exception("No Area with given id.")
            serializer = AreaSerializer(area[0], data=data, partial=True)
            if serializer.is_valid(raise_exception=True):
                serializer.save()
                return Response(
                    {"success": True, "message": "Area Updated."}, status=HTTP_200_OK,
                )
        except Exception as e:
            return Response(
                {"success": False, "message": str(e)}, status=HTTP_406_NOT_ACCEPTABLE
            )


class AreaCreation(CreateAPIView):
    serializer_class = GenreSerializer
    def post(self, request):
        try:
            data = request.data
            serializer = GenreSerializer(data=data)
            if serializer.is_valid(raise_exception=True):
                serializer.save()
                return Response(
                    {"success": True, "message": "Genre Created."},
                    status=HTTP_201_CREATED
                )
        except Exception as e:
            return Response(
                {"success": False, "message": str(e)},
                status=HTTP_406_NOT_ACCEPTABLE
            )


class AreaList(ListAPIView):
    serializer_class = AreaSerializer

    def get(self, request):
        try:
            queryset = Area.objects.all()
            serializer = AreaSerializer(queryset, many=True)
            return Response(
                {"success": True, "data": serializer.data},
                status=HTTP_200_OK,
            )
        except Exception as e:
            return Response(
                {"success": False, "message": str(e)},
                status=HTTP_404_NOT_FOUND
            )
```
