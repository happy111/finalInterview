```python
Singleton Pattern
  
    Where Used
      . Database connection pools
      . Configuration loader
    Why Used
        . Ensure only one instance exists
        . Save memory and resources

Factory Pattern
        Notification services (Email)
	    Different report generation
    Why
      Object creation logic is centralized
      Easy to add new implementations
      Follows open/Closed principle

Strategy Pattern :
    Pricing calculation
    
    Discount logic
    
        Authentication strategies (JWT, OAuth)
        Allows dynamic behavior selection

8️⃣ Builder Pattern
    📌 Where Used
    
    Creating complex response objects
    Report generation
    API payload construction

   ✅ Why Used

        Improves readability
        Handles optional fields cleanly

7️⃣ Adapter Pattern
📌 Where Used

    Integrating third-party APIs
    Payment providers

    External authentication systems

✅ Why Used
    Converts external API response to internal format
    Avoids changing business logic
```

```python
OAUTH : 

 . It allows a third-party application to access a user’s data without sharing the user’s password.
 . ✅ OAuth does NOT authenticate users directly
   ✅ It authorizes access to resources

   Real-Life Example
        “Login with Google”

Step-by-step (Interview explanation)

    User clicks Login with Google
    App redirects user to Google Authorization Server
    User logs in & approves permissions
    Google sends authorization code to app
    App exchanges code for Access Token
    App calls API using token
    API validates token and returns data

3️⃣ Where OAuth is Used?
    ✅ Third-party login

OAuth handles authorization, JWT handles authentication
```

```python
What is RestAPI ?

 . REST API stands for Representational State Transfer Application Programming Interface
 . It is a set of architectural principles used to build scalable, stateless,
     and lightweight web services that communicate over HTTP.

| HTTP Method | Operation        | Example          |
| ----------- | ---------------- | ---------------- |
| GET         | Read             | GET /users       |
| POST        | Create           | POST /users      |
| PUT         | Update (full)    | PUT /users/10    |
| PATCH       | Update (partial) | PATCH /users/10  |
| DELETE      | Delete           | DELETE /users/10 |
```

```python
3️⃣ Statelessness
Each request must contain all required information.
✅ Server does not store client session

4️⃣ Representation

Resources are represented using:
JSON (most common)
XML (rare)
```

```python
| Feature     | REST   | SOAP       |
| ----------- | ------ | ---------- |
| Protocol    | HTTP   | HTTP, SMTP |
| Data Format | JSON   | XML        |
| Performance | Fast   | Slow       |
| Scalability | High   | Low        |
| Complexity  | Simple | Complex    |
```

```python
✅ Step-by-Step Query Optimization Process

1. Identify the slow query
2. Use EXPLAIN or EXPLAIN Analyze
3. Add Proper indexes
4. Avoid select *
5. Optimize WHERE conditions 
6. Reduce number of rows early
7. Optimize Joins
8. Avoid Subqueries when possible
9. Use Limit when possible
10.Check table size & normalization
     . Remove unused columns
     . 
11. Use Caching  
```

```python
EXCEPTION HANDLING:- 

Exception Handling is used to handle runtime errors
so that the program does not crash and can continue execution gracefully.
```

```python
try:
    # risky code
except ExceptionType:
    # handle error
else:
    # runs if no exception
finally:
    # always runs
```

```python
🔹 Types of Exception Handling

1️⃣ try–except

      try:
          a = int("abc")
      except ValueError:
         print("Invalid conversion")
          
2️⃣ Multiple except Blocks

try:
    x = int(input("Enter number: "))
    y = 10 / x
except ValueError:
    print("Please enter a number")
except ZeroDivisionError:
    print("Cannot divide by zero")
    
3️⃣ Single except for Multiple Exceptions
try:
    a = int("10")
    b = 10 / 0
except (ValueError, ZeroDivisionError) as e:
    print("Error:", e)
```
