```python
7. Give a simple RAG pipeline (high-level).

Preprocess documents → chunk + clean.
Embed chunks → store in vector DB with metadata.
User query → embed query.
Retrieve top-k chunks via ANN.
Optionally rerank using cross-encoder.
Generate answer using LLM with retrieved context (prompting).
Return answer + provenance (document ids/snippets).
```
