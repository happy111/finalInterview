```python
def MyRange(start,stop=None,step=1):
    if stop == None:
        start,stop=0,start
    while(start < stop and step > 0) or (start > stop and step < 0):
        yield start
        start += step
```

```python
def extract_data(li):
    output = {}
    for i in li:
        key, value = map(str.strip, i.split(':', 1))
        if key == "tags":
            value = [tag.strip().strip("'") for tag in value.split(',')]
        output[key] = value
    return output

li = ["name : umesh","tags: 'a','b'"]
result = extract_data(li)
print(result)
```

```python
def printPattern(n,m,flag):
    print(m,end=" ")
    if n == m and flag == False:
        return
    if flag:
        if m-5 > 0 and flag == True:
            printPattern(n,m-5,True)
        else:
            printPattern(n,m-5,False)
    else:
        printPattern(n,m+5,False)
# Driver Code 
# 16 11 6 1 -4 1 6 11 16 
if __name__ == "__main__": 
  n = 16
  printPattern(n, n, True)
```

```python
from collections import Counter

def topMost(names, k):
    # Count frequency of each element
    freq = Counter(names)
    print(freq)
    # Get the k most common elements
    return [item[0] for item in freq.most_common(k)]

names = [1, 1, 1, 2, 2, 3,3,3,3]
k = 2
result = topMost(names, k)
print(result)
```

```python
def solve(s):
    ans = [1] * len(s)
    prefix = 1
    for i in range(len(s)):
        ans[i] = prefix
        prefix *= s[i]
    sufix = 1
    for i in range(len(s)-1,-1,-1):
        ans[i] *= sufix
        sufix *= s[i]

    return ans

s = [1,2,3,4]
result = solve(s)
# [24, 12, 8, 6]
print(result)
```

```python
def solve(coins, target):
    start = 0
    curr_sum = 0
    for end in range(len(coins)):
        curr_sum += coins[end]
        while curr_sum > target and start <= end:
            curr_sum -= coins[start]
            start += 1
        if curr_sum == target:
            return coins[start:end+1]
    return None


coins = [1,2,3,5,6,7]
target = 11
result = solve(coins,target)
print(result)
```

```python
def solve(li,k):
    ans = []
    for i in range(len(li)-2):
        temp = {}
        cs = k - li[i]
        for j in range(i+1,len(li)):
            if cs-li[j] in temp:
                ans.append([li[i],li[j],cs-li[j]])
            temp[li[j]] = j
    return ans

li = [3, 4, 5, 2, 1, 2]
k = 9
result = solve(li, k)
# Print the results in the desired format
print(result)
```

```python
def check_pangram(st):
    ans = [0] * 26
    for i in st:
        s = ord(i) - ord('a')
        ans[s] = 1
    print(all(ans))

st = "thequickbrownfoxjumpsoverthelazydog"
res = check_pangram(st)
print("Is pangram:", res)
```
