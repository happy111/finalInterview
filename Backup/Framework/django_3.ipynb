```python
1. Django Request & response life cycle
   ************************************
  Request Phase
  1. Client Sends HTTP Request
  2. WSGI Server Receives Request
  3. Django Initializes Settings
     Django loads the settings from settings.py, including middleware, 
     URL configuration, installed apps, etc.
  4. Middleware Processing (Request Middleware)
  5. URL Dispatcher (URLConf Matching)
  6. View Function Execution
  7. Database Query/Business Logic Execution
        Inside the view, Django may interact with models (ORM),
        perform business logic, fetch data, etc.

  Response phase
  1. View Returns HttpResponse Object
    The view function constructs and returns an
        HttpResponse, JsonResponse, HttpResponseRedirect, etc.
  2. Middleware Processing (Response Middleware)
  3. Response Sent to WSGI Server
  4. WSGI Server Sends Response to Client
```

```python
rest_framework_simplejwt:-
  1. Provides access and refresh tokens.
  2. Tokens are stateless (not stored in DB).
  3. Access tokens expire quickly (e.g. 5 mins); refresh tokens used to get new ones.
      Supports:-
        1. Token blacklisting
        2. Sliding tokens
        3. Custom claims
        4. OAuth-friendly
Limitations:-
      1. Slightly more complex setup.
      2. JWTs must be securely stored on the client side (e.g., HTTP-only cookies).
```

```python
authtoken
**********
1. Static DB token
2. No expiry
3. In DB
4. Small, internal apps
5. Manual DB deletion  (Logout / Blacklist)
6. ❌ Not supported  (Refresh tokens)
7. No  Refresh Token

simplejwt
*********
1. JWT (stateless, self-contained)
2. Access/Refresh tokens with expiry
3. No DB (unless using blacklist)
4. Modern, secure APIs (e.g. SPA/mobile)
5. ✅ Yes Refresh Token
6. 	✅ Blacklist support (optional)
```

```python
JWT Implement in django application step
****************************************
1.  Install Required Packages
pip install djangorestframework
pip install djangorestframework-simplejwt

2.  Update settings.py
INSTALLED_APPS = [
    ...
    'rest_framework',
    'rest_framework_simplejwt',
]

# REST Framework Configuration
REST_FRAMEWORK = {
    'DEFAULT_AUTHENTICATION_CLASSES': (
        'rest_framework_simplejwt.authentication.JWTAuthentication',
    ),
}

3. Create Your API Views (Example: Login, Protected Route)

from rest_framework.views import APIView
from rest_framework.response import Response
from rest_framework.permissions import IsAuthenticated

class HelloView(APIView):
    permission_classes = [IsAuthenticated]

    def get(self, request):
        return Response({"message": f"Hello {request.user}"})

4. Configure URLs

# urls.py
from django.urls import path
from rest_framework_simplejwt.views import (
    TokenObtainPairView,
    TokenRefreshView,
)
from .views import HelloView

urlpatterns = [
    path('api/token/', TokenObtainPairView.as_view(), name='token_obtain_pair'),  # Login
    path('api/token/refresh/', TokenRefreshView.as_view(), name='token_refresh'),  # Refresh Token
    path('api/hello/', HelloView.as_view(), name='hello'),  # Protected Route
]
```
