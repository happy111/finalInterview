```python
request lifecycle:-

“The request lifecycle is the series of steps a web request goes through from when it is received by 
the server until the response is sent back to the client.”

2️⃣ Steps in Typical Web Request Lifecycle:

Client sends request – browser, app, or API call sends HTTP request.

Routing – server/framework decides which controller or handler will process the request.

Middleware / Pre-processing – authentication, logging, validation, rate limiting, etc.

Controller / Business Logic – main logic executes, interacts with database or services.

Database / Cache Interaction – fetch or update data as needed.

Response Generation – format data as JSON, HTML, or other format.

Middleware / Post-processing – additional steps like compression, headers, logging.

Send response back to client – server returns HTTP response.
```

```python
Read vs Write Optimization – Simple Points

1️⃣ Definition:

Read Optimization: Designing a system to make data retrieval (reads) faster.

Write Optimization: Designing a system to make data updates/inserts (writes) faster.

2️⃣ Purpose:

Read Optimization: Improve query speed, reduce latency for users.
Write Optimization: Reduce write latency, handle high volume of updates efficiently.

3️⃣ Techniques for Read Optimization:

Use caching (Redis, Memcached).
Denormalization to avoid joins.
Indexes (B-Tree, Hash, Composite).
Materialized views for precomputed results.

4️⃣ Techniques for Write Optimization:

Normalization to reduce redundant data.
Batch inserts/updates instead of single writes.
Write-back cache to reduce DB load.
Sharding/Partitioning to distribute writes across multiple nodes.
```
