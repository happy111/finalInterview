```python
LLD  : Bottom- up
       Top-up
```

```python
| Feature   | Aggregation (◇)             | Composition (◆ Black diamond) |
| --------- | --------------------------- | ----------------------------- |
| Ownership | Weak                        | Strong                        |
| Lifetime  | Part can live without whole | Part dies if whole dies       |
| Example   | Department–Teacher          | House–Rooms                   |
```

```python
Restaurant
-------------------
+ id
+ name
+ address
+ menu : List<MenuItem>
-------------------
+ addMenuItem()
+ removeMenuItem()

Restaurant ◆── MenuItem (1-M Composition)

MenuItem
--------------------
+ id
+ code
+ name
+ price


RestaurantManager
--------------------------
+ restaurants : List<Restaurant>
--------------------------
+ getInstance()
+ addRestaurant()
+ searchRestaurant()
+ listRestaurants()

RestaurantManager ◇── Restaurant (Weak Aggregation)

User
---------------------
+ id
+ name
+ address
+ cart : Cart
---------------------
+ addToCart()
+ clearCart()

User ◇── Cart (Aggregation)

Cart


--------------------------
+ id
+ restaurant : Restaurant
+ items : List<MenuItem>
+ totalPrice
--------------------------
+ addItem(MenuItem)
+ clear()
+ calculateTotal()


PaymentStrategy
---------------------
+ pay(amount)


UPIPayment implements PaymentStrategy
-------------------------------------
+ pay(amount)


NetBankingPayment implements PaymentStrategy
--------------------------------------------
+ pay(amount)


Order ◇── PaymentStrategy (STRATEGY)


Order
--------------------------
+ id
+ user : User
+ restaurant : Restaurant
+ items : List<MenuItem>
+ paymentStrategy : PaymentStrategy
--------------------------
+ getType()
+ processPayment()
+ notifyUser()

DeliveryOrder extends Order
---------------------------
+ deliveryAddress
+ getType() = "DELIVERY"




PickupOrder extends Order
-------------------------
+ pickupAddress
+ getType() = "PICKUP"


OrderFactory
---------------------
+ createOrder(type)


NowOrderFactory implements OrderFactory
---------------------------------------
+ createOrder(type) -> DeliveryOrder / PickupOrder

ScheduleOrderFactory implements OrderFactory
--------------------------------------------
+ scheduleTime
+ createOrder(type)


OrderManager
-------------------------
+ orders : List<Order>
-------------------------
+ addOrder(order)
+ listOrders()
+ findOrderById()

OrderManager ◇── Order  (Aggregation)
```

```python
NotificationService
----------------------------
+ sendOrderConfirmation(Order)
+ sendPaymentSuccess(Order)
+ sendOrderReady(Order)
+ sendOrderDelivered(Order)

NotificationService ◇── Order  (Aggregation)


Order
-----------------
+ notifyUser()      ---> calls NotificationService


paymentStrategy.pay()
NotificationService.sendPaymentSuccess(this)
```
