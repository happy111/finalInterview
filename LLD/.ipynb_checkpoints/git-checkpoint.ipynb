```python

You are working on a feature branch .Meanwhile, the main branch received a lot of updates. When
you try to merge, you face multiple merge conflicts.
Explain step-by-step how you would resolve conflicts and ensure a clean merge ?
```

```python
1. First, pull the latest updates from main
    git fetch origin
2. Bring those updates into your feature branch
   git checkout my-feature
   git merge origin/main
3. Git will show merge conflicts
4. Open each conflicted file and fix the conflicts manually
5. git add
6. git commit
7. git push
```
