```python
# Observer Design Pattern

The Observer Pattern is used when:

One object (Subject) wants to automatically notify
many dependent objects (Observers) whenever something changes.

A user subscribes to a channel → They become an observer.
The channel uploads a new video → A change happens in the subject.
YouTube notifies all subscribers → Observers are updated.
```

```python
                                                     Account1
    youtube channel                                  Account2
                                                     Account3

     Subject / Observable                            Observer  (1-M)

```

```python
IObservable                      IObserver
-------------------             -------------------
+ add(observer: IObserver)       + update(data)
+ remove(observer: IObserver)
+ notify()


 ConcreteObservable (Iobservable)                        ConcreteObserver(Iobserver)
   Vector<Iobserver>
+ add(observer: IObserver)                              + update(data)
+ remove(observer: IObserver)
+ notify()
```

```python
from abc import ABC, abstractmethod

# --------------------------------------
# IObserver Interface
# --------------------------------------
class IObserver(ABC):
    @abstractmethod
    def update(self, data):
        pass


# --------------------------------------
# IObservable Interface
# --------------------------------------
class IObservable(ABC):
    @abstractmethod
    def add(self, observer: IObserver):
        pass

    @abstractmethod
    def remove(self, observer: IObserver):
        pass

    @abstractmethod
    def notify(self, data):
        pass


# --------------------------------------
# ConcreteObservable
# --------------------------------------
class ConcreteObservable(IObservable):
    def __init__(self):
        self.observers = []  # vector<IObserver>

    def add(self, observer: IObserver):
        self.observers.append(observer)

    def remove(self, observer: IObserver):
        if observer in self.observers:
            self.observers.remove(observer)

    def notify(self, data):
        for observer in self.observers:
            observer.update(data)


# --------------------------------------
# ConcreteObserver
# --------------------------------------
class ConcreteObserver(IObserver):
    def __init__(self, name):
        self.name = name

    def update(self, data):
        print(f"{self.name} received update: {data}")


# --------------------------------------
# Example Usage
# --------------------------------------
if __name__ == "__main__":
    observable = ConcreteObservable()

    obs1 = ConcreteObserver("Observer A")
    obs2 = ConcreteObserver("Observer B")

    observable.add(obs1)
    observable.add(obs2)

    observable.notify("New Event Happened!")

    observable.remove(obs1)

    observable.notify("Another Event!")
```
