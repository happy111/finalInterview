```python
# Payment Method – Strategy Design Pattern

An application supports multiple payment methods:

        Credit Card
        UPI
        Net Banking
        Wallet

We want to:
✔ Add new payment methods easily
✔ Avoid if-else / switch blocks
✔ Choose payment method at runtime

Each payment method is a strategy
PaymentProcessor is the context
```

```python
from abc import ABC, abstractmethod

# ----------------------------------
# Strategy Interface
# ----------------------------------
class PaymentStrategy(ABC):
    @abstractmethod
    def pay(self, amount):
        pass


# ----------------------------------
# Concrete Payment Strategies
# ----------------------------------
class CreditCardPayment(PaymentStrategy):
    def pay(self, amount):
        print(f"Paid ₹{amount} using Credit Card")


class UPIPayment(PaymentStrategy):
    def pay(self, amount):
        print(f"Paid ₹{amount} using UPI")


class NetBankingPayment(PaymentStrategy):
    def pay(self, amount):
        print(f"Paid ₹{amount} using Net Banking")


class WalletPayment(PaymentStrategy):
    def pay(self, amount):
        print(f"Paid ₹{amount} using Wallet")


# ----------------------------------
# Context Class
# ----------------------------------
class PaymentProcessor:
    def __init__(self, payment_strategy: PaymentStrategy):
        self.payment_strategy = payment_strategy

    def process_payment(self, amount):
        self.payment_strategy.pay(amount)


# ----------------------------------
# Client Code
# ----------------------------------
if __name__ == "__main__":
    payment = PaymentProcessor(UPIPayment())
    payment.process_payment(500)

    payment.payment_strategy = CreditCardPayment()
    payment.process_payment(1200)
```
