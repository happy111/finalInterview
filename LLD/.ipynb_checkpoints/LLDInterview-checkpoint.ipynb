```python
Design a simple coffee vending machine. Machine supports 3 drinks: Espresso, Cappuccino, Latte.
Each drink has different ingredient proportions (water, milk, coffee powder).
How will you design this ?
```

```python
1 . Requiremnt 

    1. Requirements
            Machine supports 3 drinks:
                Espresso
                Cappuccino
                Latte
    2. Each drink has different ingredient proportions:
            water (ml)
            milk (ml)
            coffee powder (grams)
    3. Machine must:
            Create drink based on user choice
            Deduct ingredients from stock
            Check ingredient availability

2. Why Factory Design Pattern?
        You use Factory Method Pattern because:
        Object creation (Espresso/Cappuccino/Latte) varies by recipe
        Creation logic must be separated from usage
        Easy to add new drinks in future (OCP)
        Avoids long if-else blocks inside CoffeeMachine
```

```python
Class 1: Drink (Base Class / Product)
        name
        water
        milk
        coffee
Class 2: Concrete Drink Classes
        Espresso
        Cappuccino
        Latte
        Each class sets its own ingredient proportions.
Class 3: DrinkFactory
    Method create_drink(drink_type)
    Returns appropriate Drink object
Class 4: CoffeeMachine

    Stock management
    hecks ingredient availability
    Deducts ingredients
    Calls Factory to create drink
```

```python
          +----------------------+
          |       Drink          |  (Base class)
          +----------------------+
          | name                 |
          | water                |
          | milk                 |
          | coffee               |
          +----------------------+
                 ▲      ▲      ▲
                 |      |      |
   +-------------+      |      +----------------+
   |                    |                       |
+--------+       +-------------+        +---------------+
|Espresso|       | Cappuccino  |        |    Latte      |  (Concrete classes)
+--------+       +-------------+        +---------------+
 
                 +-------------------------------+
                 |        DrinkFactory           |
                 +-------------------------------+
                 | create_drink(drink_type)      |
                 +-------------------------------+
                           |
                           v
               +--------------------------+
               |      CoffeeMachine       |
               +--------------------------+
               | stock                    |
               | make_drink(drink_type)   |
               | check_stock(drink)       |
               +--------------------------+
```

```python
# ------------------------
# Base Product
# ------------------------
class Drink:
    def __init__(self, name, water, milk, coffee):
        self.name = name
        self.water = water
        self.milk = milk
        self.coffee = coffee
```

```python
class Espresso(Drink):
    def __init__(self):
        super().__init__("Espresso", water=50, milk=0, coffee=18)

class Cappuccino(Drink):
    def __init__(self):
        super().__init__("Cappuccino", water=30, milk=50, coffee=18)

class Latte(Drink):
    def __init__(self):
        super().__init__("Latte", water=30, milk=100, coffee=18)
```

```python
# ------------------------
# Factory Class
# ------------------------
class DrinkFactory:
    @staticmethod
    def create_drink(drink_type):
        drink_type = drink_type.lower()

        if drink_type == "espresso":
            return Espresso()
        elif drink_type == "cappuccino":
            return Cappuccino()
        elif drink_type == "latte":
            return Latte()
        raise ValueError("Invalid drink type requested")
```

```python
# ------------------------
# Coffee Machine
# ------------------------
class CoffeeMachine:
    def __init__(self):
        self.stock = {
            "water": 500,
            "milk": 500,
            "coffee": 200
        }
    def check_stock(self, drink):
        if self.stock["water"] < drink.water:
            return False, "Not enough water"
        if self.stock["milk"] < drink.milk:
            return False, "Not enough milk"
        if self.stock["coffee"] < drink.coffee:
            return False, "Not enough coffee"
        return True, None

    def make_drink(self, drink_type):
        drink = DrinkFactory.create_drink(drink_type)

        ok, error = self.check_stock(drink)
        if not ok:
            print(error)
            return None

        # Deduct ingredients
        self.stock["water"] -= drink.water
        self.stock["milk"] -= drink.milk
        self.stock["coffee"] -= drink.coffee

        print(f"{drink.name} is ready! ☕")
        return drink
```

```python
if __name__ == "__main__":
    machine = CoffeeMachine()

    machine.make_drink("espresso")
    machine.make_drink("latte")
    machine.make_drink("cappuccino")

    print("Remaining stock:", machine.stock)
```
