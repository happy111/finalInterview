```python
Decorator:- 

Decorator is a structural design pattern
that lets you add new behavior to an object
dynamically without modifying its existing code
```

```python
A decorator takes a function as input and returns another function.
Its responsibility is to add new functionality without disturbing or modifying the existing function’s behavior.
```

```python
def multiply(fun):
    def wrapper(a,b):
        s = fun(a,b)
        return a * b
    return wrapper

@multiply
def sum(a,b):
    return a + b
```

```python
sum(4,5)
```

```python
class Multiply:
    def __init__(self,function):
        self.function = function
    def __call__(self,*args,**kwargs):
        s = self.function(*args,**kwargs)
        return args[0] * args[1]

@Multiply
def sum(a,b):
    return a + b
```

```python
sum(5,6)
```

```python
def validate(fun):
    def wrapper(str1,str2):
        try:
            if not isinstance(str1,str) or not isinstance(str2,str):
                raise ValueError("Please enter valid data")
            else:
                s = fun(str1,str2)
                return s
        except Exception as e:
            print(e)
    return wrapper

@validate
def solve(str1,str2):
    return str1 + str2
```

```python
solve("umesh",3333)
```

```python
Decency injection
*****************

Dependency Injection in FastAPI allows us to inject required components (like DB sessions, auth, services)
into endpoints in a clean, reusable, and testable way using Depends.

Adv :
  1. Avoid tight coupling
  2. Improve reusability
  3. Easy testing 

from fastapi import FastAPI,Depends

app = FastAPI()

def get_db():
	return "DB Connection"

@app.get("/users")
def get_users(db=Depends(get_db)):
	return {"db" : db}
```

```python
2️⃣ Encapsulation

   Binding data and methods together and restricting direct access.
       
Public
Protected (_var)
Private (__var)
```

```python
class Account:
    def __init__(self):
        self.__balance = 0   # private variable

    def deposit(self, amount):
        self.__balance += amount

    def get_balance(self):
        return self.__balance
class Umesh(Account):
    pass

obj = Umesh()
obj.deposit(400)
print(obj.get_balance())
```

```python
from abc import ABC, abstractmethod

class Vehicle(ABC):
    @abstractmethod
    def start(self):
        pass

class Car(Vehicle):
    def start(self):
        print("Car started")
obj = Car()
obj.start()
```

```python
# Class Method
class Company:
    name = "ABC"

    @classmethod
    def change_name(cls, new_name):
        cls.name = new_name
        print(cls.name)
    
obj = Company()
obj.change_name("umesh")
```

```python
#  How Python Manages Memory

Python handles memory automatically

  1️⃣ Python Memory Architecture
        Python memory is divided into two main parts:

        🔹 1. Stack Memory
                Stores function calls
                Stores local variables & references
                Managed automatically

                def add():
                    x = 10
                    y = 20
                    ➡ x and y references are stored in stack

            🔹 2. Heap Memory
                 Stores objects
                 Managed by Python’s Memory Manager
 2️⃣ Reference Counting
    Each object keeps a reference count.
            a = []
            b = a
            c = a
     Reference count = 3
     When count becomes 0, memory is released.
         
 3️⃣ Garbage Collection (GC)
     Used to handle circular references.
```

```python
class A:
    def __init__(self):
        self.ref = None

a = A()
b = A()
a.ref = b
b.ref = a

Python uses reference counting and garbage collection to manage memory automatically inside a private heap.
```

```python
# s3 lifecycle 

S3 lifecycle is mainly used to automate cost optimization and data retention without manual intervention.

“For example, in a production system, application logs are stored in S3 Standard.
After 30 days, they move to Standard-IA, after 90 days to Glacier,
and after 1 year they are deleted automatically using lifecycle rules.”
```

```python
“In S3 Lifecycle, we define rules based on object age, prefix, or tags.
These rules can move objects from S3 Standard to Standard-IA, Glacier, 
or Deep Archive, or permanently delete them after a retention period.
It also supports versioned buckets where we can manage current and non-current object versions separately.”
```

```python
2. What is the typical workflow of an AWS Lambda function integrated with API Gateway?

1. 1️⃣ Client Sends HTTP Request
    2. API Gateway receives the request.
    3. API Gateway Validates and Routes the Request

API Gateway performs:

    Authentication/Authorization (IAM, Cognito, JWT, API Keys)
    Throttling & Rate Limiting
    Request validation (Query params, headers, body)
    Mapping templates (if using REST API)
    CORS handling

Routing to correct resource + method

Then API Gateway converts the incoming HTTP request into a JSON event object.

    4. 3️⃣ API Gateway Invokes AWS Lambda


API Gateway calls your Lambda function with the generated event:

5. Lambda Executes Your Code
6. 5️⃣ Lambda Sends a Response Back to API Gateway
7. 6️⃣ API Gateway Converts Lambda Output → HTTP Response

Client → API Gateway → Request Validation → Event → Lambda → JSON Response →
API Gateway transforms response → Client
```

```python
5. Have you used FastAPI? If yes, explain how you would deploy a FastAPI application behind API Gateway using AWS Lambda.
(Optional: explain how request handling works using Django/FastAPI or any Python framework.)


    Client → API Gateway → Lambda (FastAPI + Mangum) → DynamoDB/RDS/S3


1. Create a fastapi applications

        from fastapi import FastAPI
        from mangum import Mangum
        
        app = FastAPI()
        
        @app.get("/hello")
        def say_hello():
            return {"message": "Hello from FastAPI on Lambda!"}
        
        handler = Mangum(app)

        AWS Lambda does not understand FastAPI or ASGI apps directly.
        Mangum is an adapter that converts:
        API Gateway events → ASGI request
        and
        ASGI response → API Gateway response

2. Package Dependencies
    For AWS Lambda, you must package:
     Zip the package OR use Lambda layers.

3. Create a Lambda Function
4. Configure API Gateway

Choose API Gateway → Lambda Proxy Integration.

5. Test the API

API Gateway will hit Lambda → Mangum → FastAPI → Response.

    🌟 What Happens Internally (Request Flow Explanation)

    🔹 Step 1: Client Request
    
         Client → API Gateway
               GET /hello

   🔹 Step 2: API Gateway Converts HTTP → Lambda Event
    API Gateway does NOT forward raw HTTP.
         Instead, it sends a JSON event:
{
  "httpMethod": "GET",
  "path": "/hello",
  "headers": {},
  "body": null,
  "isBase64Encoded": false
}

🔹 Step 3: Lambda Receives Event + Context
AWS Lambda calls your function:
🔹 Step 4: Mangum Converts Lambda Event → ASGI Scope

🔹 Step 5: FastAPI Processes Request

    FastAPI:
        Validates request with Pydantic
        Executes endpoint function
        Builds response object

🔹 Step 6: Response Goes Back

FastAPI → Mangum → Lambda → API Gateway → Client.
```
