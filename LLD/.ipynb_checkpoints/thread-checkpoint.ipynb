```python
Python is not strictly single-threaded, but the Global Interpreter Lock (GIL) makes it
behave like it in many cases. You can create multiple threads in Python,
but only one thread executes Python bytecode at a time.
```

```python
Global Interpreter Lock (GIL): 

The GIL is a mutex that allows only one thread to execute Python code at a time.
```

```python
thread is a smaller unit of execution within a process that shares the same memory but runs independently
```

```python
🧵 Thread
        Definition: A thread is the smallest unit of execution within a process.
        Characteristics:
        Multiple threads can exist inside a single process.
        Threads share the same memory space of the process (code, data, files).
        Each thread has its own stack and registers.
        Lightweight compared to processes — faster to create and switch between.

Example: In a web browser process, one thread might handle rendering a page while another downloads files.
```

```python
Process

   A process is a program that is currently running on your computer.
   Has its own memory space (code, data, stack, registers).
```
