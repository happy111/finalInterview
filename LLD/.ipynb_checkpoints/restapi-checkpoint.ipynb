```python
import requests

def getTopRatedFoodOutlets(city):
    page = 1
    max_rating = 0.0
    top_outlets = []
    while True:
        url = f"https://jsonmock.hackerrank.com/api/food_outlets?city={city}&page={page}"
        response = requests.get(url).json()
        data = response["data"]
        for outlet in data:
            rating = outlet["user_rating"]["average_rating"]
            name = outlet["name"]

            if rating > max_rating:
                max_rating = rating
                top_outlets = [name]   # reset list
            elif rating == max_rating:
                top_outlets.append(name)

        if page >= response["total_pages"]:
            break

        page += 1

    return top_outlets
```

```python
res = getTopRatedFoodOutlets("Denver")
```
