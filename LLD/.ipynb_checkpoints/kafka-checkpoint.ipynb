```python
Kafka service is not working properly, there could be a wide range of issues
   1.  Kafka Broker Not Running
         .  1. Check if the Broker Process Is Running ps -ef | grep kafka
            2. Check Kafka Logs  <kafka-dir>/logs/server.log
            3. Check Port Availability
            4. Start Kafka Broker Manually
   2.   Incorrect Configuration

        Symptoms: Connection issues, wrong topics, client failures
        Common Config Errors:
        listeners, advertised.listeners
        broker.id, log.dirs, zookeeper.connect
        Fix: Double-check configs in server.properties and client settings
   3.   Network or DNS Issues
   4.   Disk Full or I/O Issues
   5.   Too Many Partitions or Topics
        Symptoms: High memory/CPU usage; slow performance
        Fix:
            Reduce number of partitions
   6. Security Issues 
   7. Error Logs
        Check Kafka logs:
            /logs/kafka-server.log
            /logs/zookeeper.log
        Client errors (from producers/consumers)
```

```python
Kafka :

Apache kafka is a distributed event streaming platform.
used for building real-time data pipelines
high-speed messaging system
High throughput

Architecture :


Producer      : Sends data (messages/events) to Kafka
Kafka Broker  : Receives messages, stores them, and serves them to consumers
Consumer      : Reads messages from Kafka
Topic         : Named data stream (like a table or folder)
Partition     : Topics are split into parts to allow parallelism
Consumer Group	A group of consumers that share the work
Zookeeper (older versions): Coordinates Kafka cluster (optional in 
                                                       newer versions)
```

```python
Can multiple consumers read the same message?
✅ Yes, if they are in different consumer groups.

❌ No, if they are in the same group — message is load-balanced.
```

```python
What is a Kafka Broker?
A Kafka Broker is a Kafka server that stores data and serves it to consumers.
```

```python
7. What happens if a broker fails?
Kafka replicates data across brokers. If one fails, 
another broker with a replica can serve the data — this ensures high availability.
```

```python
11. What are retention policies in Kafka?
Kafka can:

Keep messages for a time period (e.g., 7 days)

Or until a log size limit is reached
```

```python
12. What is the role of Zookeeper in Kafka (old versions)?
Zookeeper manages:
Broker metadata
Leader election
Cluster coordination
```

```python
 ✅ Parallelism

Definition:
Parallelism means executing multiple tasks at the exact same time.

🟢 Requires multiple CPU cores
🟢 Tasks run simultaneously
🟢 Faster execution for heavy computations
```

```python
Definition:
Concurrency means dealing with multiple tasks at the same time,
but not necessarily executing them simultaneously.

🟢 Tasks overlap in time
🟢 System switches between tasks quickly
🟡 Single CPU core can handle concurrency using context switching
🟢 Improves responsiveness
```
