```python
1. Adapter

The Adapter Pattern acts as a bridge between incompatible interfaces 
so they can collaborate without changing their existing code.
    
      🎯 Real-World Example: Charging a Phone with Different Plugs


Adapter Pattern is heavily used in payment gateways, cloud services, logging, database drivers,
and microservice integrations to make
incompatible systems work together without changing existing code.

```

```python
# Old Interface
class OldPrinter:
    def print_text(self, text):
        return f"Old Printer: {text}"

# New Interface
class NewPrinter:
    def print_new(self, text):
        return f"New Printer: {text}"

# Adapter
class PrinterAdapter:
    def __init__(self, old_printer):
        self.old_printer = old_printer
    def print_new(self, text):
        return self.old_printer.print_text(text)

# Usage
old = OldPrinter()
adapter = PrinterAdapter(old)
print(adapter.print_new("Hello"))  # Old Printer: Hello
```

```python
2. Bridge

   Abstraction = WHAT you do
   Implementation = HOW you do it

Bridge Design Pattern is a structural pattern that separates an abstraction from its implementation,
```

```python
# 1. Implementor (Sender Interface)
class NotificationSender:
    def send(self, message):
        pass
```

```python
# 2. Concrete Implementors
class SMSSender(NotificationSender):
    def send(self, message):
        print("SMS Sent:", message)


class EmailSender(NotificationSender):
    def send(self, message):
        print("Email Sent:", message)
```

```python
#3. Abstraction

class Notification:
    def __init__(self, sender: NotificationSender):
        self.sender = sender

    def notify(self, message):
        pass
```

```python
class AlertNotification(Notification):
    def notify(self, message):
        print("ALERT:", end=" ")
        self.sender.send(message)


class InfoNotification(Notification):
    def notify(self, message):
        print("INFO:", end=" ")
        self.sender.send(message)
```

```python
# Client code
sms = SMSSender()
email = EmailSender()

alert = AlertNotification(sms)
info = InfoNotification(email)

alert.notify("CPU Utilization High!")
info.notify("Your bill is generated.")
```

```python
7. Proxy

    Proxy acts as a middleman between client and actual object.

    Useful when you want:
        Access control
        Lazy loading
        Logging
        Security
        Caching
        Remote access

    ✅ Why use Proxy Pattern?
       Add extra functionality (logging, authentication) without modifying the real object.


       
```

```python
class RealImage:
    def display(self):
        print("Displaying high-resolution image.")

class ImageProxy:
    def __init__(self, user_role):
        self.user_role = user_role
        self.real_image = None

    def display(self):
        if self.user_role != "admin":
            print("Access Denied: Only admin can view image.")
            return
        
        if self.real_image is None:
            self.real_image = RealImage()   # lazy load
            print("Loading image for the first time...")

        self.real_image.display()
        
proxy1 = ImageProxy("guest")
proxy1.display()

proxy2 = ImageProxy("admin")
proxy2.display()
proxy2.display()
```

```python
5. Facade

   .  The Facade Pattern provides a simple, unified interface to a complex subsystem.
   .  Instead of interacting with multiple classes, the client interacts with one facade class,
         which internally calls all required operations.

    Real-Life Analogy (Very Easy)
        Example: Ordering Food

    When you order food from Swiggy/Zomato, you don’t talk to:
    The restaurant
    The delivery person
    The payment gateway
    The notification system

    Instead, you use one simple interface → the app.
    👉 Swiggy = Facade
    👉 Complex internal services = Subsystems
```
