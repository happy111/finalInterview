```python
✅ Database Schema Design

1️⃣ authors table
CREATE TABLE authors (
    author_id INT PRIMARY KEY AUTO_INCREMENT,
    name VARCHAR(255) NOT NULL,
    bio TEXT
);

2️⃣ books table

CREATE TABLE books (
    book_id INT PRIMARY KEY AUTO_INCREMENT,
    title VARCHAR(255) NOT NULL,
    published_year INT,
    genre VARCHAR(100)
);

3️⃣ book_authors (junction table)
CREATE TABLE book_authors (
    id INT PRIMARY KEY AUTO_INCREMENT,
    book_id INT NOT NULL,
    author_id INT NOT NULL,

    FOREIGN KEY (book_id) REFERENCES books(book_id),
    FOREIGN KEY (author_id) REFERENCES authors(author_id),

    UNIQUE(book_id, author_id)  -- prevents duplicate pairings
);
```

```python
| Feature            | **CHAR(n)**                                             | **VARCHAR(n)**                    |
| ------------------ | ------------------------------------------------------- | --------------------------------- |
| Storage Type       | **Fixed-length**                                        | **Variable-length**               |
| Storage Behavior   | Always stores **exact n characters** (pads with spaces) | Stores **only actual characters** |
| Speed              | Faster (because fixed size)                             | Slightly slower                   |
| Storage Efficiency | Wastes space if data is short                           | Saves space                       |
| Best Use Case      | Fixed-size data                                         | Data with variable length         |
```

```python
class Old_printer:
    def oldPrint(self):
        return "Old Once"

class New_printer:
    def newprint(self):
        return "New Once"

class Adapter:
    def __init__(self,old):
        self.old = old
    def newprint(self):
        return self.old.oldPrint()

obj= Old_printer()
obj = Adapter(obj)
obj.newprint()
```

```python
from abc import ABC,abstractmethod

class PaymentStrategy(ABC):
    @abstractmethod
    def pay(self,amount):
        pass

class CreditCard(PaymentStrategy):
    def pay(self,amount):
        print(f"Paid {amount} using Credit Card")

    

class PaymentProcessor:
    def __init__(self,strategy:PaymentStrategy):
        self.strategy = strategy

    def process(self, amount):
        self.strategy.pay(amount)

obj = PaymentProcessor(CreditCard())
obj.process(400)
```

```python
You have a FastAPI application that needs to call three external APIs. These API calls are independent from each other.

How will you improve the performance using Python concurrency?
Explain your approach and code outline.


AsyncClient() → makes non-blocking HTTP calls
await client.get(url) → waits for response without blocking other tasks
resp.json() → convert response to Python dict

httpx → async HTTP client
asyncio → enables concurrency / async tasks
FastAPI → web framework                       

```

```python
  
import httpx
import asyncio
from fastapi import FastAPI

app = FastAPI()

async def call_api(url):
    async with httpx.AsyncClient() as client:
        resp = await client.get(url)
        return resp.json()

@app.get("/aggregate")
async def aggregate():
    url1 = "https://service1/api"
    url2 = "https://service2/api"
    url3 = "https://service3/api"

    # Run all 3 calls concurrently
    result1, result2, result3 = await asyncio.gather(
        call_api(url1),
        call_api(url2),
        call_api(url3),
    )

    return {
        "s1": result1,
        "s2": result2,
        "s3": result3
    }
```
