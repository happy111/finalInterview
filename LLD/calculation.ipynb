```python
Back of the envelope calculations :-
**********************************

MY application support how many users

QPS : Query per second (Read Request and Write request)

Read : select

Write : insert,delete,update

Instgram
Back of the envelope calc :

power of 2

1 GB = 1024 MB   2 ^ 10
1 MB = 1024 KB : 2 ^ 10

10   Thousands : kilobyte
20   Millions  : Megabyte
30   Billions  : Gegabyte
40   Trillions : Terabyte
50   Quadrillions : Petabyte
60   Exabyte


Monthly active users : 2 billion
60% of these users use instadaily 1.2 billion
DAU : Daily active users 1.2B

Users see feed (1 feed request - 1 query)

user post photo/video/reel (1 post request 1 query)

User check feed 30 types a day(average) 
Estiamate of 5 years

User publish one picture/reel a day(Average)



FEED VIEW QPS
QPS  :  DAILY FEED REQUEST : 1.2B * 30 / 86400(1 DAY = 86400)  = ~420K

Upload QPS  :  Daily upload request : 1.2B * 1 / 86400 = ~14k/sec

PEEK QPS :- 2 * QPS = ~840K/SEC
peek qps : ~ 2 * upload qps = ~28k/sec


Storage Unit :

  Assumptions : 
     20% of users : Video 50MB
     80% of users : photo 1MB

photos per day = 1.2B * 80 % = ~1B
1B * 1MB = 2^30 * 1*2^20 = 2^ 50 = ~1pB

Videos per day = 1.2B * 20% = ~0.25B * 50
0.25 * 2^ 30 * 50 * 2^ 20 = 2^50*12 = ~12PB

Total 12PB + 1PB = 13PB

Calculation : 13PB * 365 * 5


Slightly clearer (still simple, interview-safe):

Cassandra / DynamoDB → feeds, posts, likes, comments (high QPS, availability)
PostgreSQL / MySQL → users, auth, relationships
Object storage (S3 + CDN) → photos & videos

👉 DynamoDB is used for feeds and posts because it can handle massive read/write traffic with high availability and horizontal scaling.
```

```python
🔹 Storage Calculation
Per URL record size

Assume:

Short URL key = 8 bytes

Long URL = 2 KB

Metadata (timestamps, flags, user_id) = 1 KB

  ≈ 3 KB per record

Daily writes
10,000 × 86400 ≈ 864M URLs/day

Storage per day
864M × 3 KB ≈ 2.6 TB/day

Storage per year
2.6 TB × 365 ≈ 949 TB ≈ ~1 PB/year

Storage for 5 years
1 PB × 5 ≈ 5 PB

| Layer        | Technology           |
| ------------ | -------------------- |
| Cache        | Redis                |
| Primary DB   | DynamoDB / Cassandra |
| ID generator | Snowflake            |
| TTL cleanup  | DB TTL               |
```
