```python
1. Strategy Pattern
   Used to choose different algorithms at runtime.
       
   Strategy pattern allows switching persistence behavior at runtime.

       class Persistence(ABC):
           def save(self, data):

        class FileStorage(Persistence)
        class DBStorage(Persistence)

📌 Use Case

Payment system selects a payment method (UPI, Card, PayPal).
```

```python
# Strategy Interface
class PaymentStrategy:
    def pay(self, amount):
        pass

# Concrete Strategies
class UpiPayment(PaymentStrategy):
    def pay(self, amount):
        print(f"Paid {amount} using UPI")

class CardPayment(PaymentStrategy):
    def pay(self, amount):
        print(f"Paid {amount} using Credit Card")

# Context
class PaymentProcessor:
    def __init__(self, strategy: PaymentStrategy):
        self.strategy = strategy
        
    def process(self, amount):
        self.strategy.pay(amount)

# Usage
processor = PaymentProcessor(UpiPayment())
processor.process(500)

processor = PaymentProcessor(CardPayment())
processor.process(500)
```

```python
#StrategyDesignPattern.py Strategy Design Pattern (Robot)
#***************************

from abc import ABC, abstractmethod

# -------------------------------
# Strategy Interface for Walk
# -------------------------------
class WalkableRobot(ABC):
    @abstractmethod
    def walk(self):
        pass


# Concrete Walk Strategies
class NormalWalk(WalkableRobot):
    def walk(self):
        print("Walking normally...")


class NoWalk(WalkableRobot):
    def walk(self):
        print("Cannot walk.")


# -------------------------------
# Strategy Interface for Talk
# -------------------------------
class TalkableRobot(ABC):
    @abstractmethod
    def talk(self):
        pass


# Concrete Talk Strategies
class NormalTalk(TalkableRobot):
    def talk(self):
        print("Talking normally...")


class NoTalk(TalkableRobot):
    def talk(self):
        print("Cannot talk.")


# -------------------------------
# Strategy Interface for Fly
# -------------------------------
class FlyableRobot(ABC):
    @abstractmethod
    def fly(self):
        pass


# Concrete Fly Strategies
class NormalFly(FlyableRobot):
    def fly(self):
        print("Flying normally...")


class NoFly(FlyableRobot):
    def fly(self):
        print("Cannot fly.")


# -------------------------------
# Robot Base Class (Context)
# -------------------------------
class Robot(ABC):
    def __init__(self, walk_behavior: WalkableRobot,
                 talk_behavior: TalkableRobot,
                 fly_behavior: FlyableRobot):
        self.walk_behavior = walk_behavior
        self.talk_behavior = talk_behavior
        self.fly_behavior = fly_behavior

    def walk(self):
        self.walk_behavior.walk()

    def talk(self):
        self.talk_behavior.talk()

    def fly(self):
        self.fly_behavior.fly()

    @abstractmethod
    def projection(self):
        pass


# -------------------------------
# Concrete Robot Types
# -------------------------------
class CompanionRobot(Robot):
    def projection(self):
        print("Displaying friendly companion features...")


class WorkerRobot(Robot):
    def projection(self):
        print("Displaying worker efficiency stats...")


# -------------------------------
# Client Code
# -------------------------------
if __name__ == "__main__":
    robot1 = CompanionRobot(
        NormalWalk(),
        NormalTalk(),
        NoFly()
    )

    robot1.walk()
    robot1.talk()
    robot1.fly()
    robot1.projection()

    print("--------------------")

    robot2 = WorkerRobot(
        NoWalk(),
        NoTalk(),
        NormalFly()
    )

    robot2.walk()
    robot2.talk()
    robot2.fly()
    robot2.projection()
```

```python
✅ 2. Observer Pattern
**********************

👉 Defines a one-to-many dependency so that when one object changes state,

all its dependents are notified automatically

📌 Use Case

Email/SMS notifications when user registers.
```

```python
class Observer:
    def update(self, message):
        pass

class EmailObserver(Observer):
    def update(self, message):
        print(f"Email sent: {message}")

class SMSObserver(Observer):
    def update(self, message):
        print(f"SMS sent: {message}")

# Subject
class NotificationSubject:
    def __init__(self):
        self.observers = []
        
    def subscribe(self, obs):
        self.observers.append(obs)
        
    def notify(self, msg):
        for obs in self.observers:
            obs.update(msg)

# Usage
subject = NotificationSubject()
subject.subscribe(EmailObserver())
subject.subscribe(SMSObserver())

subject.notify("User registered successfully!")
```

```python
#  State Pattern

Object changes its behavior based on internal state.

📌 Use Case

Order state: NEW → PACKED → SHIPPED → DELIVERED.
```

```python
class State:
    def handle(self):
        pass

class NewState(State):
    def handle(self):
        print("Order in NEW state")

class ShippedState(State):
    def handle(self):
        print("Order SHIPPED")

class Order:
    def __init__(self):
        self.state = NewState()

    def set_state(self, state):
        self.state = state
    
    def process(self):
        self.state.handle()


order = Order()
order.process()

order.set_state(ShippedState())
order.process()
```

```python
Observer	            Notify dependents of state change
Strategy	            Encapsulate interchangeable algorithms
Command	                Encapsulate requests as objects
Chain of Responsibility	Pass request along a chain of handlers
State	                Change behavior when state changes
Iterator	            Access collection elements sequentially
Mediator	            Encapsulate interaction between objects
```
