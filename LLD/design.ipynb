```python
 1. Singleton Pattern
Definition: Ensures that a class has only one instance and provides a global access point.
    
Use Case: 
        Database connection pooling – 
        A single instance manages all DB connections efficiently instead of creating multiple instances.

🔹 2. Factory Pattern

Definition: Provides an interface for creating objects 
            in a superclass but allows subclasses to alter the type of objects that will be created.

        Use Case: Logger Framework – 
          
                Different loggers (FileLogger, ConsoleLogger) are created using a LoggerFactory without modifying client code.

🔹 3. Adapter Pattern

Definition: Allows incompatible interfaces to work together by acting as a bridge between them.
    
    Use Case: 
         Payment Gateway Integration 
             – Adapting different payment providers (PayPal, Stripe, Razorpay) to work with a common interface.

🔹 4. Builder Pattern
    
Definition: Simplifies object creation by constructing complex objects step by step instead of using a large constructor.
Use Case: Creating an HTTP Request – The Request.Builder() in OkHttp helps in constructing complex HTTP requests with headers, body, etc.

🔹 5. Observer Pattern
    
Definition: Defines a dependency between objects so that when one object changes, all its dependents are notified.
Use Case: Event Listeners in UI – When a button is clicked, multiple listeners (like logging, analytics) respond automatically.

🔹 6. Strategy Pattern
    
Definition: Defines a family of algorithms, encapsulates each one, and makes them interchangeable at runtime.
Use Case: Payment Strategy – Choosing between different payment modes (Credit Card, UPI, Net Banking) dynamically at checkout.

🔹 7. Decorator Pattern
    
    : Adds additional behaviour to objects dynamically without modifying their structure.
Use Case: Adding Features in a Text Editor – Features like spell check, grammar check,
        and auto-correct can be added dynamically to a text editor component.
```
