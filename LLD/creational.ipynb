```python
Creational Design Pattern
 
    There are 5 main creational design patterns:

        Singleton
        Factory Method
        Abstract Factory
        Builder
        Prototype
```

```python
SingleTon :-  Ensure only one instance of a class exists and provide a global access point.

    When to use?
            Database connection
            Logger class
            Thread pool

__new__ controls object creation before __init__.
```

```python
class SingleTon:
    __instance__=None
    def __new__(cls):
        if cls.__instance__ == None:
            cls.__instance__ = super().__new__(cls)
        return cls.__instance__
```

```python
obj1 = SingleTon()
obj2 = SingleTon()
obj1 == obj2
```

```python
2️⃣ Factory Method Pattern

    Let subclass decide which object to create
```

```python
# Example

class Shape:
  def draw(self):
      pass
      
class Circle(Shape):
    def draw(self):
        return "Drawing Circle"

class Square(Shape):
    def draw(self):
        return "Drawing Square"

class ShapeFactory:
    def get_shape(self, shape_type):
        if shape_type == "circle":
            return Circle()
        elif shape_type == "square":
            return Square()
factory = ShapeFactory()
shape = factory.get_shape("circle")
print(shape.draw())
```

```python
3️⃣ Abstract Factory Pattern

Abstract Factory Pattern allows you to create families of related objects without 
specifying their concrete classes, ensuring that objects created together are compatible.
    
    When to use?
        UI frameworks (Dark theme, Light theme)
        Multi-database support (MySQL / Postgres)
```

```python
from abc import ABC, abstractmethod

# ===== Abstract Products =====
class Button(ABC):
    @abstractmethod
    def render(self):
        pass

class Checkbox(ABC):
    @abstractmethod
    def render(self):
        pass


# ===== Concrete Products =====
class WindowsButton(Button):
    def render(self):
        return "Rendering a Windows Button"

class MacButton(Button):
    def render(self):
        return "Rendering a Mac Button"

class WindowsCheckbox(Checkbox):
    def render(self):
        return "Rendering a Windows Checkbox"

class MacCheckbox(Checkbox):
    def render(self):
        return "Rendering a Mac Checkbox"


# ===== Abstract Factory =====
class UIFactory(ABC):
    @abstractmethod
    def create_button(self) -> Button:
        pass

    @abstractmethod
    def create_checkbox(self) -> Checkbox:
        pass


# ===== Concrete Factories =====
class WindowsFactory(UIFactory):
    def create_button(self) -> Button:
        return WindowsButton()

    def create_checkbox(self) -> Checkbox:
        return WindowsCheckbox()

class MacFactory(UIFactory):
    def create_button(self) -> Button:
        return MacButton()

    def create_checkbox(self) -> Checkbox:
        return MacCheckbox()


# ===== Client Code =====
def app(factory: UIFactory):
    button = factory.create_button()
    checkbox = factory.create_checkbox()

    print(button.render())
    print(checkbox.render())


# ===== Usage =====
if __name__ == "__main__":
    os_type = "Windows"  # this could come from config/env

    if os_type == "Windows":
        factory = WindowsFactory()
    else:
        factory = MacFactory()

    app(factory)
```

```python
4️⃣ Builder Pattern

Build complex objects step-by-step, allowing different representations.

    When to use?
        Building an object with too many optional parameters
        Creating different variants of an object
```

```python
class Computer:
    def __init__(self):
        self.cpu = None
        self.ram = None
        self.gpu = None

    def __str__(self):
        return f"CPU={self.cpu}, RAM={self.ram}, GPU={self.gpu}"

class ComputerBuilder:
    def __init__(self):
        self.computer = Computer()

    def add_cpu(self, cpu):
        self.computer.cpu = cpu
        return self

    def add_ram(self, ram):
        self.computer.ram = ram
        return self

    def add_gpu(self, gpu):
        self.computer.gpu = gpu
        return self

    def build(self):
        return self.computer

# Usage
pc = (ComputerBuilder()
      .add_cpu("i7")
      .add_ram("16GB")
      .add_gpu("RTX 3060")
      .build())

print(pc)
```

```python
5️⃣ Prototype Pattern

    Clone existing objects instead of creating new ones from scratch.

    When to use?
    
    Costly object creation
    Need object duplication with minimal changes
```

```python
import copy

class Car:
    def __init__(self, model, color):
        self.model = model
        self.color = color

    def clone(self):
        return copy.copy(self)

# Usage
car1 = Car("BMW", "Black")
car2 = car1.clone()
car2.color = "Red"

print(car1.color)  # Black
print(car2.color)  # Red
```

```python
| Pattern              | Goal                                   | Best For                        |
| -------------------- | -------------------------------------- | ------------------------------- |
| **Singleton**        | Only one instance                      | DB connections, logger          |
| **Factory Method**   | Delegate object creation to subclasses | Frameworks, extensibility       |
| **Abstract Factory** | Create families of related objects     | UI themes, cross-platform apps  |
| **Builder**          | Build complex objects step-by-step     | Large objects with many options |
| **Prototype**        | Clone objects                          | Costly initialization           |
```

```python
3. Abstract Factory
******************

    that provides an interface to create families of related or dependent objects 
        without specifying their concrete classes.
```

```python
4. Builder

Why use the Builder pattern?
****************************
     1. To handle complex object creation
         Some objects have many parameters (especially optional ones).
     2.  Improved readability
     3.  Immutability
     4.   Avoid large constructors

Use the Builder Pattern when:
    You have a class with many parameters (especially optional ones).
    You want readable and maintainable object creation.
    You want the construction process to be independent from the representation.
    You want to create immutable objects.


class Computer:
    def __init__(self):
        self.parts = []
    def add(self, part):
        self.parts.append(part)
    def show(self):
        return ", ".join(self.parts)

# Builder
class ComputerBuilder:
    def __init__(self):
        self.computer = Computer()
    def add_cpu(self): self.computer.add("CPU")
    def add_ram(self): self.computer.add("RAM")
    def add_gpu(self): self.computer.add("GPU")
    def build(self): return self.computer

# Usage
builder = ComputerBuilder()
builder.add_cpu()
builder.add_ram()
pc = builder.build()
print(pc.show())  # CPU, RAM
```

```python
5. Prototype

👉 Creates new objects by cloning existing ones instead of creating from scratch.
```

```python
import copy

class Prototype:
    def __init__(self, name):
        self.name = name
    def clone(self):
        return copy.deepcopy(self)

# Usage
obj1 = Prototype("Original")
obj2 = obj1.clone()
obj2.name = "Cloned"
print(obj1.name)  # Original
print(obj2.name)  # Cloned
```

```python
Pattern	Short Definition
Singleton	        One instance only
Factory Method	    Subclass decides object creation
Abstract Factory	Creates families of related objects
Builder	            Step-by-step object construction
Prototype	        Clone existing objects
```
