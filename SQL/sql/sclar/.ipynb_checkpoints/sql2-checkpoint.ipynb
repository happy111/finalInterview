```python
create database scaler;

create table students(
   id int,
   firstname varchar(50) not null,
   lastname varchar(50) default 'samal',
   email varchar(50) not null unique default 'umeshsamal3@gmail.com',
   dateofBirth date,
   enrollmentDate timestamp not null,
   psp decimal(3,2) default 0,
   batchId int default 1,
   isActive boolean not null,
   primary key(id,email)

)
```

```python
create table students(
   id int AUTO_INCREMENT PRIMARY KEY,
   batch_id int,
   firstname varchar(50) not null,
   lastname varchar(50) default 'samal',
   email varchar(50) not null unique default 'umeshsamal3@gmail.com',
   dateofBirth date,
   enrollmentDate timestamp not null,
   psp decimal(3,2) default 0,
   batchId int default 1,
   isActive boolean not null,
   foreign key(batch_id) references batches(batch_id)
   on delete cascade
   on update cascade

)
```

```python
SELECT 
    a.city
FROM 
    person p
LEFT JOIN 
    address a USING(personId)
ORDER BY 
    a.city;
```

```python
select distinct c.title from content as c   join TVProgram as t on c.content_id = t.content_id where
    c.kids_content = 'Y'
    AND MONTH(t.program_date) = 6
    AND YEAR(t.program_date) = 2020 and c.content_type='MOVIES'  order by c.title asc
```
