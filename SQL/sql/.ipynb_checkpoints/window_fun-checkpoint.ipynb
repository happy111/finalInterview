```python
import re
text = "Hello, world! This is a test-string; with multiple delimiters."

# Define the delimiters (e.g., comma, space, exclamation mark, hyphen, semicolon)
delimiters = r'[.,\s!;-]+'

# Split the string using the delimiters
result = re.split(delimiters, text)
print(result)
```

```python
SELECT id,name,if(ps=1,'A',
			   if(ps=2,'B',
               if(ps=3,'C',
               if(ps=4,'D',
               if(ps=5,'E', 'G' ))))) as performance FROM onecard.performance;
```

```python
SELECT id, 
       name, 
       CASE 
           WHEN ps = 1 THEN 'A'
           WHEN ps = 2 THEN 'B'
           WHEN ps = 3 THEN 'C'
           WHEN ps = 4 THEN 'D'
           WHEN ps = 5 THEN 'E'
           ELSE 'G'
       END as performance 
FROM onecard.performance;
```

```python
SELECT COALESCE(column1, column2, 'default_value') AS result
FROM table_name;
```

```python
SELECT id, 
       name, 
       CASE 
           WHEN COALESCE(ps, 0) = 1 THEN 'A'
           WHEN COALESCE(ps, 0) = 2 THEN 'B'
           WHEN COALESCE(ps, 0) = 3 THEN 'C'
           WHEN COALESCE(ps, 0) = 4 THEN 'D'
           WHEN COALESCE(ps, 0) = 5 THEN 'E'
           ELSE 'G'
       END as performance 
FROM onecard.performance;
```

```python
SELECT id,COALESCE(name,salary) as p FROM onecard.employee;
```

```python
###  VIEWS  ###

In MySQL, a view is a virtual table based on the result set of a SELECT query.
Views are useful for simplifying complex queries, abstracting data,
and enhancing security by restricting access to specific data.
Here's how you can create, use, and manage views in MySQL.
```

```python
CREATE VIEW view_name AS
SELECT column1, column2, ...
FROM table_name
WHERE condition;
```

```python
SELECT emp_id,
       sum(sales) over() As sales,
       Avg(sales) over() As Abg
	   from onecard.wf;
```

```python
SELECT emp_id,
	   product_id,dept,
       sum(sales) over(partition by dept) As sales,
       Avg(sales) over(partition by dept) As Abg
	   from onecard.wf;
```

```python
       
select emp_id,
       product_id,
       dept,
       sales,
       rank() over (order by sales desc) as sale_rank from onecard.wf;
```
