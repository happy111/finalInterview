```python
# Rank Score

| Id | Score |
+----+-------+
| 1 | 3.50 |
| 2 | 3.65 |
| 3 | 4.00 |
| 4 | 3.85 |
| 5 | 4.00 |
| 6 | 3.65 |


Output

+-------+------+
| Score | Rank |
+-------+------+
| 4.00 | 1 |
| 4.00 | 1 |
| 3.85 | 2 |
| 3.65 | 3 |

select score, (select count(distinct score) from practise.score where score >= s.score) 
as Ranks
from practise.score as s
order by Score desc;
```

```python
#  . Consecutive Numbers

Write a SQL query to find all numbers that appear at least three times
consecutively
```

```python
SELECT num
FROM practise.consutive
GROUP BY num
HAVING COUNT(num) == 3)
```
