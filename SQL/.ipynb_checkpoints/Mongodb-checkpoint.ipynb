```python
1. What is MongoDB?

MongoDB is a NoSQL, document-oriented database that stores data in BSON format (JSON-like).
```

```python
2. Difference between SQL and NoSQL?

| SQL          | MongoDB (NoSQL)    |
| ------------ | ------------------ |
| Tables       | Collections        |
| Rows         | Documents          |
| Fixed schema | Schema-less        |
| Joins        | Embedded documents |
```

```python
3. What is a Document in MongoDB?

A JSON-like structure:

{
  "name": "Umesh",
  "age": 25
}
```

```python
4. What is a Collection?

A group of documents (similar to a table).
```

```python
5. What is BSON?

Binary JSON — faster storage, allows extra data types (date, binary).
```

```python
16. What is Aggregation Framework?

Used for data analytics operations like:

    $match
    $group
    $sort
    $limit
```

```python
User data:-

[
  { "name": "Amit",   "age": 25 },
  { "name": "Rohan",  "age": 30 },
  { "name": "Neha",   "age": 25 },
  { "name": "Sonal",  "age": 22 },
  { "name": "Deepak", "age": 30 },
  { "name": "Rahul",  "age": 22 },
  { "name": "Kiran",  "age": 25 }
]
```

```python
db.users.aggregate([
  { $group: { _id: "$age", count: { $sum: 1 } } }
])

Output
[
  { "_id": 22, "count": 2 },
  { "_id": 25, "count": 3 },
  { "_id": 30, "count": 2 }
]
```

```python
db.users.aggregate([
  {
    $group: {
      _id: "$age",
      count: { $sum: 1 },
      users: { $push: "$name" }
    }
  }
])

[
  { "_id": 25, "count": 3, "users": ["Amit", "Neha", "Kiran"] },
  { "_id": 22, "count": 2, "users": ["Sonal", "Rahul"] },
  { "_id": 30, "count": 2, "users": ["Rohan", "Deepak"] }
]
```

```python
21. What is a Replica Set?

A group of MongoDB servers:

1 primary
Multiple secondaries
```

```python
| Property            | Meaning                                                     |
| ------------------- | ----------------------------------------------------------- |
| **A – Atomicity**   | All operations succeed or ALL fail.                         |
| **C – Consistency** | Database moves from one valid state to another valid state. |
| **I – Isolation**   | One transaction does not affect another transaction.        |
| **D – Durability**  | Once committed, data is permanently stored.                 |
```

```python
29. What is an Explain Plan? Why used?

Used to measure query performance.
db.users.find({name: "Umesh"}).explain("executionStats")

Explain Plan in MongoDB shows how the database executes a query.
    It is used to analyze performance, detect slow operations,
understand index usage, and optimize queries. Using explain("executionStats"),
developers can see the number 
of documents scanned, execution time, and whether indexes are used.
```

```python
30. How to troubleshoot slow $sort + $group?

Create index for sort field
Use $group first, then sort
Avoid large memory use → allowDiskUse: true

Use pipeline optimization
```

```python
31. What is the Aggregation Pipeline Optimization?

MongoDB automatically rearranges pipeline stages for performance.
```

```python
33. Why is $lookup slow?

Because it is like a left join and scans both collections.
```

```python
34. How to optimize $lookup?

Use indexes on join fields

Reduce document size

Use pipeline $lookup with filters
```

```python
40. How to reduce document size?

Use short field names

Use embedded docs instead of referencing

Avoid arrays with huge lengths
```
