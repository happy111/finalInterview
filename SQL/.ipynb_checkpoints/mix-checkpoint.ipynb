```python
Write a query to get the latest order for each customer
from orders(cust_id, order_id, order_date, amount)

Using Join:-
************
select O.* from orders O JOIN(
  select cust_id, Max(order_date) AS latest_date from orders group by cust_id
) t on O.cust_id = t.cust_id AND o.order_date = t.latest_date;

Using Subquery
**************
SELECT o.*
FROM orders o
WHERE o.order_date = (
    SELECT MAX(order_date)
    FROM orders
    WHERE cust_id = o.cust_id
);

Window Function
***************
    
SELECT cust_id, order_id, order_date, amount
FROM (
    SELECT 
        cust_id,
        order_id,
        order_date,
        amount,
        ROW_NUMBER() OVER (PARTITION BY cust_id ORDER BY order_date DESC) AS rn
    FROM orders
) AS t
WHERE rn = 1;

```
