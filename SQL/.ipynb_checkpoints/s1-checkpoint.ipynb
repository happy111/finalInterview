```python
Q1. Show total sales for each department without using GROUP BY.

SELECT DISTINCT
    department,
    SUM(sales) OVER (PARTITION BY department) AS dept_total
FROM orders;
```

```python
Q2. Show each employee’s rank based on sales within their department.

select name, department, sales,
Rank() over(PARTITION BY department order by sales DESC) AS dept_rank
from orders.
```

```python
Scenario 1: Find the highest salary employee in each department

SELECT 
  emp_name,
  department,
  salary,
  MAX(salary) OVER (PARTITION BY department) AS max_salary_in_dept
FROM emp;
```

```python
# 🧩 Scenario 2: Assign a unique row number to each employee ordered by salary

SELECT 
  emp_name,
  department,
  salary,
  Rank() OVER (order by salary desc) AS row_num
FROM employees;
```

```python
🧩 Scenario 3: Find the difference between current and previous employee salary

SELECT emp_name, salary,
       LAG(salary) OVER (ORDER BY salary) AS prev_salary,
       salary - LAG(salary) OVER (ORDER BY salary) AS diff
FROM employees;

| emp_name | salary | prev_salary | diff |
| -------- | ------ | ----------- | ---- |
| John     | 5000   | NULL        | NULL |
| David    | 5500   | 5000        | 500  |
| Mary     | 6000   | 5500        | 500  |
| Lisa     | 6500   | 6000        | 500  |
| Tom      | 7000   | 6500        | 500  |
```

```python
🧩 Scenario 4: Compare each employee’s salary with the next employee

SELECT emp_name, salary,
       LEAD(salary) OVER (ORDER BY salary) AS next_salary,
       LEAD(salary) OVER (ORDER BY salary) - salary AS diff
FROM employees;
```

```python
🧩 Scenario 5: Calculate total and average salary for each department

SELECT emp_name, dept_id, salary,
       SUM(salary) OVER (PARTITION BY dept_id) AS dept_total,
       AVG(salary) OVER (PARTITION BY dept_id) AS dept_avg
FROM employees;
```

```python
🧩 Scenario 6: Get top 2 highest salaries in each department

SELECT emp_name, dept_id, salary
FROM (
  SELECT emp_name, dept_id, salary,
         DENSE_RANK() OVER (PARTITION BY dept_id ORDER BY salary DESC) AS rnk
  FROM employees
) t
WHERE rnk <= 2;
```

```python
🧩 Scenario 10: Find salary rank within the entire company

SELECT emp_name, salary,
       RANK() OVER (ORDER BY salary DESC) AS overall_rank
FROM employees;
```
