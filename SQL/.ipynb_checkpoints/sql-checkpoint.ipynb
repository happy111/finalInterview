```python
ROW_NUMBER()
***********
  	Assigns a unique row number.

    Ex : Pagination, deduplication.

SELECT 
    name, salary,
    ROW_NUMBER() OVER (ORDER BY salary DESC) AS row_num
FROM employees;

select first_name,last_name, ROW_NUMBER() OVER (ORDER BY customer_id)
as xz FROM sakila.customer;
```

```python
SELECT 
    name, salary,
    RANK() OVER (ORDER BY salary DESC) AS rank_num,
    DENSE_RANK() OVER (ORDER BY salary DESC) AS dense_rank_num
FROM employees;
```

```python
Aspect	RANK()	
Skips the next rank(s) after ties	

DENSE_RANK()
No gaps; assigns next consecutive rank
```

```python
Nth max salary
*************

SELECT salary
FROM (
    SELECT 
        salary,
        DENSE_RANK() OVER (ORDER BY salary DESC) AS rank_val
    FROM employees
) AS ranked
WHERE rank_val = 3;
```

```python
Case 
***

SELECT 
    name,
    salary,
    CASE department
        WHEN 'HR' THEN 'Human Resources'
        WHEN 'IT' THEN 'Information Technology'
        ELSE 'Other Department'
    END AS department_full_name
FROM employees;
```

```python
SELECT 
    CASE col_name
        WHEN 0 THEN 1
        WHEN 1 THEN 0
        ELSE col_name
    END AS converted_value
FROM your_table;
```

```python
Write a query to get the latest order for each customer
from orders(cust_id, order_id, order_date, amount)

SELECT cust_id, order_id, order_date, amount
FROM (
    SELECT o.*,
           ROW_NUMBER() OVER (PARTITION BY cust_id ORDER BY order_date DESC) AS rn
    FROM orders o
) t
WHERE rn = 1;
```
