```python
What is the difference between ROW_NUMBER(), RANK(), and DENSE_RANK()

| Function         | Handles Ties?            | Behavior                                                     | Example Order Values |
| ---------------- | ------------------------ | ------------------------------------------------------------ | -------------------- |
| **ROW_NUMBER()** | ❌ *Does NOT handle ties* | Always gives a **unique** sequential number                  | 1, 2, 3, 4           |
| **RANK()**       | ✅ Yes                    | Gives same rank for ties but **skips** the next rank         | 1, 1, 3, 4           |
| **DENSE_RANK()** | ✅ Yes                    | Gives same rank for ties but **does NOT skip** the next rank | 1, 1, 2, 3           |
```

```python
| emp_id | sale_date  | amount |
| ------ | ---------- | ------ |
| 1      | 2024-01-01 | 500    |
| 1      | 2024-01-05 | 700    |
| 1      | 2024-01-10 | 300    |
| 2      | 2024-01-03 | 900    |
| 2      | 2024-01-05 | 400    |
| 3      | 2024-01-02 | 200    |
Q2. Write an SQL query to get the top 2 highest sales per employee.

SELECT emp_id, sale_date, amount
FROM (
    SELECT 
        emp_id,
        sale_date,
        amount,
        ROW_NUMBER() OVER (
            PARTITION BY emp_id 
            ORDER BY amount DESC
        ) AS rn
    FROM sales
) t
WHERE rn <= 2;
```

```python
| order_id | cust_id | amount |
| -------- | ------- | ------ |
| 101      | 1       | 500    |
| 102      | 1       | 700    |
| 103      | 2       | 900    |
We want total sales per customer.

SELECT cust_id, SUM(amount) AS total_sales
FROM orders
GROUP BY cust_id;
```

```python
CTE  : Common table expression

WITH total_sales AS (
    SELECT cust_id, SUM(amount) AS total_amount
    FROM orders
    GROUP BY cust_id
)
SELECT *
FROM total_sales;
```
