```python
# NTILE is a SQL window function used to divide ordered rows into equal buckets (groups).

| name | salary |
| ---- | ------ |
| A    | 10000  |
| B    | 20000  |
| C    | 30000  |
| D    | 40000  |
| E    | 50000  |
| F    | 60000  |

# Example 1️⃣: Divide employees into 3 salary groups
SELECT
    name,
    salary,
    NTILE(3) OVER (ORDER BY salary) AS salary_group
FROM employees;

| name | salary | salary_group |
| ---- | ------ | ------------ |
| A    | 10000  | 1            |
| B    | 20000  | 1            |
| C    | 30000  | 2            |
| D    | 40000  | 2            |
| E    | 50000  | 3            |
| F    | 60000  | 3            |
```
