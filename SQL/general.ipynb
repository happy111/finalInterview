```python
#  Memory Management:-

  Splits memory into:
      
    Stack memory → stores function calls, local variables
    Heap memory  → stores objects (created by Python interpreter)
 
  Every object in Python has a reference count

  Python’s Garbage Collector is responsible for allocation and dellocation of the memory.
```

```python
Memory Management Best Practices

Use Local Variables: Local variables have a shorter lifespan and are freed sooner than global variables.

Avoid Circular References: Circular references can lead to memory leaks if not properly managed.
    
Use Generators: Generators produce items one at a time and only keep one item in memory at a time, 
making them memory efficient.
    
Explicitly Delete Objects: Use the del statement to delete variables and objects explicitly.
    
Profile Memory Usage: Use memory profiling tools like tracemalloc and memory_profiler to identify memory leaks 
    and optimize memory usage.
```

```python
Multithreading:-
 1. Multiple threads in a single process
 2. Threads share the same memory
 3. I/O-bound tasks
 4. Very fast

Multiprocessing :- 
 1. Multiple separate processes
 2. Each process has its own memory space
 3. CPU-bound tasks
 4. Slow
```
