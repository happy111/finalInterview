```python
5. VIEWS

1 . A View is a saved SQL query.
2.  A View is a virtual table (not a real one) that is created from the result of a SQL query.
3.  It looks like a table but doesn’t store data physically — it stores only the query definition.

    ✅ Syntax

CREATE VIEW view_name AS
SELECT column1, column2, ...
FROM table_name
WHERE condition;
```

```python
CREATE VIEW it_employees AS
SELECT e.emp_id, e.emp_name, e.salary
FROM employees e
JOIN department d ON e.dept_id = d.dept_id
WHERE d.dept_name = 'IT';


SELECT * FROM it_employees;
```

```python
6. INDEXES
***********
CREATE INDEX idx_emp_name ON employees(emp_name);

1. An index is like a “book index” — it helps MySQL find data faster without scanning the entire table.
    
2. It’s a data structure (usually a B-tree) stored separately that keeps track of
  where each value is located in the table.
```

```python
⚙️ How It Works Internally

MySQL creates a B-tree (balanced tree) or hash structure.

This structure keeps column values sorted and stores a pointer to the row’s location in the table.

When searching, MySQL navigates the tree, like how you search alphabetically in a book index.
```

```python
🧱 Types of Indexes in MySQL
****************************

1. PRIMARY KEY
   ***********
     Uniquely identifies each row; automatically creates a unique index
     Example:-
       PRIMARY KEY (emp_id)
2. UNIQUE INDEX
   ************
     Prevents duplicate values in a column

     CREATE UNIQUE INDEX idx_email ON users(email);
3. INDEX (Non-Unique)
   *****************
     Speeds up lookups but allows duplicates

     CREATE INDEX idx_name ON employees(emp_name);

4. FULLTEXT INDEX
   **************
     Used for text searching

     CREATE FULLTEXT INDEX idx_desc ON articles(description);
5. COMPOSITE INDEX
   ***************
     Index on multiple columns

    CREATE INDEX idx_emp_dept ON employees(emp_name, dept_id);
```

```python
⚡ Performance Example

🔹 Without Index:

   SELECT * FROM employees WHERE emp_name = 'John';

   MySQL scans all rows — slow (O(n)).

🔹 With Index:

   CREATE INDEX idx_emp_name ON employees(emp_name);

    SELECT * FROM employees WHERE emp_name = 'John';
    ⏱️ MySQL finds row using B-tree — fast (O(log n)).
```

```python
Check Existing Indexes

SHOW INDEX FROM employees;

🧹 Remove an Index

DROP INDEX idx_emp_name ON employees;
```

```python
7. NORMALIZATION
****************

Goal: Remove redundancy, avoid anomalies, and make data efficient.

Normalization organizes data into multiple tables and sets rules (normal forms).
It prevents problems like:

Duplicate data
Update anomalies
Delete anomalies
Insert anomalies

| Normal Form | Rule                                | Simple Meaning                                             |
| ----------- | ----------------------------------- | ---------------------------------------------------------- |
| **1NF**     | No repeating columns, atomic values | Each cell should have one value only                       |
| **2NF**     | 1NF + no partial dependency         | Non-key columns must depend on whole primary key           |
| **3NF**     | 2NF + no transitive dependency      | Non-key columns should not depend on other non-key columns |
```
