```python
Types of index:-

    In MySQL, indexes are used to speed up data retrieval from tables

    1. PRIMARY KEY
        Uniquely identifies each record in a table.
        Automatically creates a unique index.
        Only one primary key per table.
        CREATE TABLE users (
            id INT PRIMARY KEY,
            name VARCHAR(50)
        );
    2. UNIQUE
        Ensures that all values in the column(s) are unique.
        Allows NULL values (but only one NULL if a single column).

                CREATE TABLE users (
            email VARCHAR(100) UNIQUE
        );
    3. INDEX (or KEY)
       Standard index to speed up queries.
           CREATE INDEX idx_name ON users(name);
    4. FULLTEXT
```

```python
You have two tables:
Orders(order_id, customer_id, amount) and
Customers(customer_id, name).
Write a query to get each customer's name and total amount spent.??

SELECT 
    c.name,
    SUM(o.amount) AS total_amount
FROM 
    Customers c
LEFT JOIN 
    Orders o 
ON 
    c.customer_id = o.customer_id
GROUP BY 
    c.name;
```

```python
SELECT
    c.name,
    o.amount,
    SUM(o.amount) OVER (PARTITION BY c.customer_id) AS total_amount
FROM
    Customers c
LEFT JOIN
    Orders o
ON
    c.customer_id = o.customer_id;
```
