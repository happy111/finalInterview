```python
What is SQL?

SQL (Structured Query Language) is used to interact with databases — create, read, update, delete data.
```

```python
Create Database & Table
***********************

CREATE DATABASE company_db;

USE company_db;

CREATE TABLE employees (
    emp_id INT PRIMARY KEY,
    emp_name VARCHAR(100),
    dept_id INT,
    salary DECIMAL(10,2),
    hire_date DATE
);
```

```python
INSERT Data
***********
INSERT INTO employees (emp_id, emp_name, dept_id, salary, hire_date)
VALUES (101, 'Alice', 1, 50000, '2020-01-15'),
       (102, 'Bob', 2, 60000, '2019-07-22'),
       (103, 'Charlie', 1, 55000, '2021-03-10');
```

```python
SELECT Data
***********
SELECT * FROM employees;
SELECT emp_name, salary FROM employees;
SELECT emp_name FROM employees WHERE salary > 55000;
```

```python
UPDATE & DELETE
***************
UPDATE employees SET salary = 58000 WHERE emp_id = 103;
DELETE FROM employees WHERE emp_id = 101;
```

```python
🟡 2. INTERMEDIATE CONCEPTS
***************************
SELECT emp_name AS Name, salary AS "Monthly Salary" FROM employees;
```

```python
FILTERING with WHERE, AND, OR, BETWEEN, LIKE, IN
************************************************
SELECT * FROM employees WHERE salary BETWEEN 50000 AND 60000;
SELECT * FROM employees WHERE emp_name LIKE 'A%';
SELECT * FROM employees WHERE dept_id IN (1, 2);
```

```python
ORDER BY & LIMIT
****************
SELECT * FROM employees ORDER BY salary DESC;
SELECT * FROM employees ORDER BY hire_date LIMIT 3;
```

```python
🧮 AGGREGATE FUNCTIONS
***********************
SELECT COUNT(*) AS total_employees FROM employees;
SELECT AVG(salary) AS avg_salary FROM employees;
SELECT dept_id, SUM(salary) FROM employees GROUP BY dept_id;
```

```python
HAVING Clause
*************
SELECT dept_id, AVG(salary) AS avg_sal
FROM employees
GROUP BY dept_id
HAVING AVG(salary) > 55000;
```

```python
# Difference bet group by and having
************************************
having
*****
    Used to Filter Groups
SELECT department, COUNT(*) AS total_employees
FROM employees
GROUP BY department
HAVING COUNT(*) > 5;

group by
*********
GROUP BY is used to group rows that have the same values in specified columns into summary rows.
    SELECT department, COUNT(*) AS total_employees
FROM employees
GROUP BY department;
```

```python
🔵 4. SUBQUERIES

Example 1: Employee earning more than average

SELECT emp_name, salary
FROM employees
WHERE salary > (SELECT AVG(salary) FROM employees
```

```python
Subquery vs Join — Which is Faster and Why
*****************************************
1. JOIN
A JOIN combines data from two or more tables based on a related column.

SELECT e.emp_name, d.dept_name
FROM employees e
JOIN department d ON e.dept_id = d.dept_id;

What Happens Internally:
***********************
The database engine retrieves data from both tables.
Then it combines them using the join condition.
Modern SQL engines optimize joins efficiently using indexing, hashing, or merge algorithms.
⚡ Performance:
Faster than subqueries in most cases, especially with proper indexes.
SQL optimizers can handle joins more efficiently since they are set-based operations.

🧩 2. SUBQUERY
SELECT emp_name
FROM employees
WHERE dept_id IN (SELECT dept_id FROM department WHERE location = 'Delhi');


✅ What Happens Internally:
    ***********************
The inner query executes first.
Its result is used by the outer query.
If not optimized properly, it can cause multiple executions or temporary tables.
⚡ Performance:
Usually slower than joins when used as correlated subqueries
(where the inner query runs once per row of the outer query).
    
But non-correlated subqueries can perform fine depending on the optimizer.
```
