```python
1. Types of join
1. INNER JOIN (Only matching rows from both tables)

SELECT employees.name, departments.department_name
FROM employees
INNER JOIN departments
ON employees.dept_id = departments.id;
```

```python
dept_id	dept_name
10	    HR
20	    IT
40	    Finance
```

```python
SELECT 
    e.emp_id, 
    e.emp_name, 
    e.dept_id, 
    d.dept_name
FROM 
    employee AS e
INNER JOIN 
    dept AS d
ON 
    e.dept_id = d.dept_id;
```

```python
2. LEFT JOIN  (All from left + matching from right)
***************************************************
If there’s no match, you’ll get NULL for columns from the right table.

SELECT employees.name, departments.department_name
FROM employees
LEFT JOIN departments
ON employees.dept_id = departments.id;
    
```

```python
Table 1: employee
emp_id	emp_name	dept_id
1	    Alice	    10
2	    Bob	        20
3	    Charlie	    30
4	    David	    NULL
```

```python
Table 2: dept
dept_id	dept_name
10	    HR
20	    IT
40	    Finance
```

```python
SELECT 
    e.emp_name, 
    d.dept_name
FROM 
    employee AS e
LEFT JOIN 
    dept AS d
ON 
    e.dept_id = d.dept_id;
```

```python
3. RIGHT JOIN  (All from right + matching from left)

SELECT employees.name, departments.department_name
FROM employees
RIGHT JOIN departments
ON employees.dept_id = departments.id;
```

```python
Table 1: employee
emp_id	emp_name	dept_id
1	    Alice	    10
2	    Bob	        20
3	    Charlie	    30
4	    David	    NULL
```

```python
Table 2: dept
dept_id	dept_name
10	    HR
20	    IT
40	    Finance
```

```python
SELECT 
    e.emp_name, 
    d.dept_name
FROM 
    employee AS e
RIGHT JOIN 
    dept AS d
ON 
    e.dept_id = d.dept_id;
```

```python
4. FULL JOIN    (All rows from both tables)
Returns all rows from both tables.
Non-matching rows from each side are filled with NULLs

SELECT employees.name, departments.department_name
FROM employees
FULL OUTER JOIN departments
ON employees.dept_id = departments.id;

```

```python
5. CROSS JOIN    (Every combination (Cartesian product)
every row from the first table is paired with every row from the second table.

SELECT employees.name, departments.department_name
FROM employees
CROSS JOIN departments;
```

```python
 6. SELF JOIN      (Table joined with itself)

SELECT e.name AS Employee, m.name AS Manager
FROM employees e
JOIN employees m
ON e.manager_id = m.id;
```

```python
| Join Type      | What it Returns                    |
| -------------- | ---------------------------------- |
| **INNER JOIN** | Only matching rows                 |
| **LEFT JOIN**  | All left + matching right          |
| **RIGHT JOIN** | All right + matching left          |
| **FULL JOIN**  | All from both, matched + unmatched |
| **CROSS JOIN** | Every combination                  |
| **SELF JOIN**  | Table joined with itself           |
```
