```python
A window function performs a calculation across a set of table rows that are related to
the current row — this set of rows is called a window.

    Advantages :-
     1. Ranking and ordering
     2. Partitioning data
     3. Flexibility and readability
```

```python
1. Aggregate Window Functions

Function	Description
SUM()	    Calculates running or window total
AVG()	    Calculates average over window
COUNT()	    Counts rows in the window
MIN()	    Finds minimum value
MAX()	    Finds maximum value

Ex:- SUM()

SELECT 
  emp_name,
  dept_id,
  salary,
  SUM(salary) OVER (PARTITION BY dept_id) AS dept_total_salary
FROM employees;

Explanation:
Calculates total salary per department.
Unlike GROUP BY, all employee rows remain visible.
    
| emp_id | emp_name | dept_id | salary |
| ------ | -------- | ------- | ------ |
| 1      | John     | 10      | 5000   |
| 2      | Mary     | 10      | 6000   |
| 3      | David    | 20      | 5500   |
| 4      | Lisa     | 20      | 6500   |
| 5      | Tom      | 30      | 7000   |

SELECT 
  emp_name,
  dept_id,
  salary,
  SUM(salary) OVER (PARTITION BY dept_id) AS dept_total_salary
FROM employees;

| emp_name | dept_id | salary | dept_total_salary |
| -------- | ------- | ------ | ----------------- |
| John     | 10      | 5000   | 11000             |
| Mary     | 10      | 6000   | 11000             |
| David    | 20      | 5500   | 12000             |
| Lisa     | 20      | 6500   | 12000             |
| Tom      | 30      | 7000   | 7000              |

```

```python
AVG()	    Calculates average over window

| emp_id | emp_name | dept_id | salary |
| ------ | -------- | ------- | ------ |
| 1      | John     | 10      | 5000   |
| 2      | Mary     | 10      | 6000   |
| 3      | David    | 20      | 5500   |
| 4      | Lisa     | 20      | 6500   |
| 5      | Tom      | 30      | 7000   |

SELECT
  emp_id,
  emp_name,
  dept_id,
  AVG(salary) OVER (PARTITION BY dept_id) AS dept_avg_salary
FROM emp;

output
******
| emp_id | emp_name | dept_id | dept_avg_salary |
| ------ | -------- | ------- | --------------- |
| 1      | John     | 10      | 5500            |
| 2      | Mary     | 10      | 5500            |
| 3      | David    | 20      | 6000            |
| 4      | Lisa     | 20      | 6000            |
| 5      | Tom      | 30      | 7000            |
```

```python
🥈 2. Ranking Functions
       These assign a rank or order to rows based on a condition.

| Function       | Description                        |
| -------------- | ---------------------------------- |
| `ROW_NUMBER()` | Gives a unique sequential number   |
| `RANK()`       | Gives rank with gaps if duplicates |
| `DENSE_RANK()` | Gives rank without gaps            |
| `NTILE(n)`     | Divides rows into `n` equal groups |
```

```python
Ex:-
1. ROW_NUMBER()    Gives a unique sequential number  

**************

PARTITION BY → (optional) divides rows into groups.

ORDER BY → defines the order in which the row numbers are assigned.

| emp_id | emp_name | dept_id | salary |
| ------ | -------- | ------- | ------ |
| 1      | John     | 10      | 5000   |
| 2      | Mary     | 10      | 6000   |
| 3      | David    | 20      | 5500   |
| 4      | Lisa     | 20      | 6500   |
| 5      | Tom      | 10      | 7000   |
| 6      | Alex     | 20      | 6200   |

SELECT
  emp_id,
  emp_name,
  dept_id,
  salary,
  ROW_NUMBER() OVER (PARTITION BY dept_id ORDER BY salary DESC) AS row_num
FROM employees;

Step-by-Step Explanation
The data is divided by dept_id (two partitions: 10 and 20).
Inside each department, rows are ordered by salary (highest first).
ROW_NUMBER() assigns 1, 2, 3… for each row in that order.

| emp_id | emp_name | dept_id | salary | row_num |
| ------ | -------- | ------- | ------ | ------- |
| 5      | Tom      | 10      | 7000   | 1       |
| 2      | Mary     | 10      | 6000   | 2       |
| 1      | John     | 10      | 5000   | 3       |
| 4      | Lisa     | 20      | 6500   | 1       |
| 6      | Alex     | 20      | 6200   | 2       |
| 3      | David    | 20      | 5500   | 3       |
```

```python
Function	    Handles Ties How?
RANK()	        Skips ranks for ties
DENSE_RANK()	Doesn’t skip ranks
```

```python
Example : RANK()
****************

| emp_name | salary |
| -------- | ------ |
| John     | 5000   |
| Mary     | 6000   |
| David    | 6000   |
| Lisa     | 7000   |
SELECT emp_name, salary,
       RANK() OVER (ORDER BY salary DESC) AS rank
FROM employees;
| emp_name | salary | rank |
| -------- | ------ | ---- |
| Lisa     | 7000   | 1    |
| Mary     | 6000   | 2    |
| David    | 6000   | 2    |
| John     | 5000   | 4    |
```

```python
Example : DENSE_RANK()
****************

| emp_name | salary |
| -------- | ------ |
| John     | 5000   |
| Mary     | 6000   |
| David    | 6000   |
| Lisa     | 7000   |
SELECT emp_name, salary,
       DENSE_RANK() OVER (ORDER BY salary DESC) AS dense_rank
FROM employees;
| emp_name | salary | rank |
| -------- | ------ | ---- |
| Lisa     | 7000   | 1    |
| Mary     | 6000   | 2    |
| David    | 6000   | 2    |
| John     | 5000   | 3    |
```

```python
SELECT salary
FROM (
    SELECT salary,
           DENSE_RANK() OVER (ORDER BY salary DESC) AS rnk
    FROM employees
) t
WHERE rnk = 2;
```

```python
2nd Highest Salary
******************
select salary from(
   select salary, DENSE_RANK() over(order by salary desc) as rnk
    from employees
) t where rnk = 2;
    
```

```python
ROW_NUMBER() removes duplicates

👉 Even though A and B have same salary, they get different row numbers.

DENSE_RANK() :-

👉 Duplicate salaries get same rank.
👉 Next rank is not skipped → (1,1,2,3)
```

```python
A cluster is a database object that stores rows from one or more tables together in the same
data blocks because they are frequently accessed together.

Clustered tables share the same data blocks.

Why Clusters Are Used

	Reduce disk I/O
	Improve JOIN performance
	Faster retrieval of related records


Types of cluster :-
   Indexed Cluster
    2️⃣ Hash Cluster
```

```python
| Feature          | Cluster       | Normal Table    |
| ---------------- | ------------- | --------------- |
| Storage          | Shared blocks | Separate blocks |
| Disk I/O         | Less          | More            |
| Join performance | Faster        | Slower          |
| Insert speed     | Slower        | Faster          |
| Maintenance      | Complex       | Simple          |
```
