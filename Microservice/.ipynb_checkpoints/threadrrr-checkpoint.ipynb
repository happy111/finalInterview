```python
import threading
def print_sentence(sentence):
    print(sentence)
sentences = [
    "Sentence 1: Hello!",
    "Sentence 2: Welcome to Python.",
    "Sentence 3: Multithreading is cool.",
    "Sentence 4: This is concurrent printing.",
    "Sentence 5: All printed at once!"
]
threads = []
for s in sentences:
    t = threading.Thread(target=print_sentence, args=(s,))
    threads.append(t)
    t.start()

for t in threads:
    t.join()
```

```python
asyncio is Python’s built-in library to write asynchronous and non-blocking
programs, especially for I/O-bound and high-level structured network code.
```

```python
import asyncio

async def solve():
    print("How are you")

async def solve1():
    print("I am fine")

async def solve2():
    print("I am good")

async def main():
    await asyncio.gather(solve(), solve1(), solve2())

#asyncio.run(main())
await main()
```

```python
asyncio:-
  1. A module (library) in Python
  2. Provides tools to run asynchronous code
  3. asyncio.run(main())	

await:-
  1.  A keyword in Python
  2.  Tells Python to pause and wait
  
```

```python
Difference bet thread and asyncio
*********************************

thread
******
1. Uses OS-level threads
2. OS switches threads
3. Harder (race conditions, locks, deadlocks)
4. High memory usage, context-switching overhead   
asyncio
*******
1. Uses single-threaded event loop
2. we control switches using await 
3. Easier (no locks needed usually)
4. Lightweight, efficient in many connections
```

```python
Real-Life Analogy:
Threading = Like hiring multiple people to do different jobs at the same time.

Asyncio = Like one person doing multiple tasks, switching between them smartly when waiting.
```

```python
When to Use What?
Use Case	                                                 Use
Handling 1000+ HTTP requests	                             asyncio
Reading/writing many files	                                 asyncio
Heavy image or video processing	                             threading or multiprocessing
Background job with blocking APIs	                          threading
GUI + background tasks	                                      threading (commonly)
```

```python
What Causes a Deadlock?
Imagine two threads and two locks:

Thread A acquires Lock 1, then tries to acquire Lock 2.

Thread B acquires Lock 2, then tries to acquire Lock 1.

Both threads are now stuck, 
waiting for the other to release the lock — this is a deadlock.
```

```python
Managing deadlocks in Python
****************************
    1. Acquire locks in a fixed order	
    2. Use timeouts	
    3. Minimize critical section time
    4. Avoid nested locks if possible
```

```python
| Feature       | Threads           | Multiprocessing    | Asyncio                     |
| ------------- | ----------------- | ------------------ | --------------------------- |
| Runs in       | Same process      | Separate processes | Same process                |
| Parallel?     | ❌ No (GIL blocks) | ✅ Yes              | ❌ No                        |
| For CPU-bound | Bad               | Best               | Bad                         |
| For I/O-bound | Good              | Overkill           | Best                        |
| Memory        | Shared            | Separate           | Shared                      |
| Switch type   | Preemptive (OS)   | Preemptive (OS)    | Cooperative (`await`)       |
| Overhead      | Low               | High               | Very low                    |
| Typical use   | I/O tasks         | Heavy CPU tasks    | High concurrency networking |
```

```python
1. No OS Thread Overhead
    Asyncio uses coroutines, 
    not OS threads → no heavy thread creation, no per-thread memory, no thread stacks.
2. No Context Switching Cost
   Thread switching is expensive (done by OS).
   Asyncio switches tasks inside Python, which is extremely lightweight.
3. No GIL Contention
    Asyncio runs in one thread, so no locking or GIL blocking.
```
