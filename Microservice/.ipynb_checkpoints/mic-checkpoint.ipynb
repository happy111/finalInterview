```python
Saga Pattern:

  1. One of the most well-known patterns for distributed transactions is called saga
  
  2.  A saga is a sequence of local transactions where each transaction
updated data within a single service.


Distributed : One transaction is depend to other transaction


Types is saga
  1. Choreography   / event based

     Each local transaction publishes domain events that trigger local transactions in other
     services.

     One service triggers an event, then other services listen and start their work

     order service -> order created event -> payment service ->build order event-> stock service

    -< order prepared event -> delivery service

  2. Orchestration  / Command Based
```
