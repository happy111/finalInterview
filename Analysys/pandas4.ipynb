```python
import pandas as pd

data = {
    'A': [1, 2, None, 4],
    'B': [None, 2, 3, 4]
}

df = pd.DataFrame(data)
print(df)
```

```python
df_filled = df.fillna(0)
print(df_filled)
```

```python
df_dropped = df.dropna()
print(df_dropped)
```

```python
df = df.rename(columns={'A': 'AA'})
```

```python
import pandas as pd

data = {
    'City': ['Mumbai', 'Pune', 'Mumbai', 'Pune', 'Delhi'],
    'Temperature': [34, 30, 33, 31, 28],
    'Humidity': [80, 75, 82, 70, 65]
}

df = pd.DataFrame(data)
print(df)
```

```python
df1 = df.groupby("City")["Temperature"].mean()
```

```python
df1
```

```python
df1 = df.groupby("City")["Temperature"].sum()
```

```python
df1
```

```python
# Pivot table
import pandas as pd

data = {
    'Date': ['2025-07-18', '2025-07-18', '2025-07-19', '2025-07-19', '2025-07-19'],
    'City': ['Mumbai', 'Pune', 'Mumbai', 'Pune', 'Delhi'],
    'Sales': [200, 150, 220, 180, 90]
}

df = pd.DataFrame(data)
print(df)
```

```python
pivot_result = pd.pivot_table(df, values='Sales', index='Date',aggfunc='sum')
print(pivot_result)
```
