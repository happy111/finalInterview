```python
import pandas as pd

data = {
    'Student': ['Alice', 'Bob', 'Charlie', 'David'],
    'Math':   [85, 90, 75, 80],
    'Science': [88, 85, 90, 82],
    'English': [82, 88, 85, 90]
}

df = pd.DataFrame(data)
print(df)
```

```python
col_avg = df[['Math', 'Science', 'English']].mean(axis=0)
print(col_avg)
```

```python
df["Row_Avg"] = df[['Math', 'Science', 'English']].mean(axis=1)
print(df)
```
