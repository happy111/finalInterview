```python
import pandas as pd

data = {
    'Student': ['Alice', 'Bob', 'Charlie', 'David'],
    'Math':   [85, 90, 75, 80],
    'Science': [88, 85, 90, 82],
    'English': [82, 88, 85, 90]
}

df = pd.DataFrame(data)
print(df)
```

```python
df
```

```python
df[["Math","Science"]]
```

```python

# ---- Child-wise average ----
df['child_average'] = df[['Math', 'Science', 'English']].mean(axis=1)
```

```python
df['child_average']
```

```python
# ---- Subject-wise average ----
subject_average = df[['Math', 'Science', 'English']].sum()

print(subject_average)
top_subject = subject_average.sort_values(ascending=False).head(2)
```

```python
print(top_subject)
```

```python
import pandas as pd

# Sample data
data1 = {
    "ID": [1, 2, 3, 4],
    "Name": ["Alice", "Bob", "Charlie", "David"]
}

data2 = {
    "ID": [5, 6],
    "Name": ["Eve", "Frank"]
}

# Create DataFrames
df1 = pd.DataFrame(data1)
df2 = pd.DataFrame(data2)

# Concatenate DataFrames along rows
concat_df = pd.concat([df1, df2], ignore_index=True)

print("Concatenated DataFrame (along rows):")
print(concat_df)
```

```python
import pandas as pd

# Sample data
data_left = {
    "ID": [1, 2, 3, 4],
    "Name": ["Alice", "Bob", "Charlie", "David"]
}

data_right = {
    "Score": [85, 90, 88, 92]
}

# Create DataFrames
df_left = pd.DataFrame(data_left)
df_right = pd.DataFrame(data_right, index=[1, 2, 3, 4])

# Set index for the left DataFrame
df_left.set_index('ID', inplace=True)

# Perform join
joined_df = df_left.join(df_right)

print("Joined DataFrame (left join):")
print(joined_df)
```

```python
import pandas as pd

# Sample data
data_left = {
    "ID": [1, 2, 3, 4],
    "Name": ["Alice", "Bob", "Charlie", "David"]
}

data_right = {
    "ID": [3, 4, 5, 6,7],
    "Score": [85, 90, 88, 92,44]
}

# Create DataFrames
df_left = pd.DataFrame(data_left)
df_right = pd.DataFrame(data_right)

# Perform merge (inner join by default)
merged_df = pd.merge(df_left, df_right, on="ID")

print("Merged DataFrame (inner join):")
print(merged_df)
```

```python
data = {
    "Name" : [1,2,3,4,'umesh','samal']
}
data1 = {
    "Name1" : [1,2,3,4,'umesh','samal',8,9]
}
df = pd.DataFrame(data)
df1 = pd.DataFrame(data1)
name_set = set(df['Name'])
name1_set = set(df1['Name1'])
# Find the difference
difference = name1_set - name_set

# Convert the result to a list and sort (if needed)
output = list(difference)

# Print the output
print(output)
```

```python
| Feature       | `merge()`                        | `join()`               | `concat()`                           |
| ------------- | -------------------------------- | ---------------------- | ------------------------------------ |
| Purpose       | SQL-style joins                  | Index-based joins      | Stacking DataFrames                  |
| Join Type     | inner/left/right/outer           | inner/left/right/outer | vertical or horizontal concatenation |
| Works On      | Columns (or index)               | Index (or left column) | Rows or columns                      |
| Syntax Style  | SQL-like                         | Method of DataFrame    | Simple stacking                      |
| Key Required? | Yes                              | Usually index          | No                                   |
| Best Use Case | Combine tables using key columns | Combine by index       | Add rows or columns                  |
```
