```python
1. What is NLP?

NLP (Natural Language Processing) is a field of AI that helps computers understand, interpret, and generate human language.
```

```python
3. What is Tokenization?

Breaking text into smaller units (words, subwords, sentences).

Word tokenization
Sentence tokenization
Subword tokenization (BPE, WordPiece)
```

```python
4. What is Stopwords Removal?

Stopwords (the, is, a, an) are removed to reduce noise
```

```python
5. What is Stemming?

Reduces words to root form using rules.
Example:

"playing" → "play"

"studies" → "studi"
```

```python
6. What is Lemmatization?

Converts words to their dictionary/base form using POS tags.
Example:

"better" → "good"
"running" → "run"


```

```python
| Stemming              | Lemmatization     |
| --------------------- | ----------------- |
| Rule-based            | Dictionary-based  |
| Fast                  | Slower            |
| Not always meaningful | Always meaningful |
```

```python
8. What is Bag of Words (BoW)?

Represents text as word frequency.
```

```python
9. What is TF-IDF?

Term Frequency – Inverse Document Frequency

It helps in:
    Feature extraction
    Text classification
    Search engines
    Removing common words (“and”, “the”)

Giving importance to rare but meaningful words
```

```python
10. What is n-gram?

Sequence of n words.

    Unigram: I
    Bigram: I love
    Trigram: I love NLP
```

```python
11. What is Word Embedding?

Dense numeric vector representation of words.

Examples:

    Word2Vec
    GloVe
    FastText
Word Embedding is a technique that represents words as numerical vectors, where:

👉 Words with similar meaning have similar vectors
👉 Words are placed close together in a continuous vector space
👉 It captures semantic meaning (king, queen, man, woman relationships)

Famous Word Embedding Techniques
    
    Word2Vec (Skip-gram / CBOW)
    GloVe
    FastText
    BERT / Transformer embeddings (contextual embeddings)
```

```python
13. What is One-hot Encoding?

Binary vector representing presence/absence of words.
```

```python
14. What is POS Tagging?

Assigns grammatical labels (noun, verb, adjective).
```

```python
16. What is Sentiment Analysis?

Classifying text as positive, negative, neutral.
```

```python
17. What is Text Classification?

Categorizing text into predefined labels (spam/not spam).
```

```python
18. What is a Confusion Matrix (NLP context)?

Evaluates classification performance with TP, FP, FN, TN.
```

```python
19. What is Cosine Similarity?

Measures similarity between two text vectors.
Used in:

Search engines
Recommendation systems
Semantic matching
```

```python
| Feature           | Sparse Vector                              | Dense Vector                            |
| ----------------- | ------------------------------------------ | --------------------------------------- |
| Meaning           | Mostly **zeros**, very few non-zero values | Mostly **non-zero** values              |
| Size              | Very large                                 | Much smaller                            |
| Memory usage      | High                                       | Low                                     |
| Example use       | Bag of Words, TF-IDF                       | Word Embeddings (Word2Vec, GloVe, BERT) |
| Captures meaning? | ❌ No                                       | ✅ Yes                                   |
| Representation    | Hard for ML models to learn patterns       | Easy for ML models                      |
| Speed             | Slow (high dimensional)                    | Fast (low dimensional)                  |
```

```python
| Bag-of-Words        | Embeddings       |
| ------------------- | ---------------- |
| Sparse vectors      | Dense vectors    |
| Large dimensions    | Small dimensions |
| No semantic meaning | Captures meaning |
```

```python
21. Explain Language Modeling.

Predicting the next word in a sentence.
Used in:
    Autocomplete
    Speech recognition
    Transformers
```
