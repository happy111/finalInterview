```python
10. What is a confusion matrix?

   A table showing TP, TN, FP, FN measuring model accuracy.

       1️⃣ Precision
              precision : TP / (TP + FP)

              We predict 10 customers Will buy products
              Out of them, 7 actually bought (correct).
              3 did not (false positives
              Precision = 7 / (7 + 3) = 0.70
              70% of your positive predictions were correct.

        2️⃣ Recall
        
        Recall = TP / (TP + FN)

        Out of all actual positives, how many did we correctly identify?
        
        ✔ Example:
        
        There are 12 actual buyers.
        You correctly predicted 7 (TP).
        You missed 5 buyers (FN).
        
        Recall = 7 / (7 + 5) = 0.58
        
        3️⃣ F1-score
        
        F1 = Harmonic Mean of Precision & Recall
        
        Formula:
             F1=2×   Precision+Recall  /  Precision×Recall
```

```python
🎯 When to use which?

Precision important
When false positives are costly
Example: Spam email detection (don't mark real emails as spam)

Recall important
When missing positives is dangerous
Example: Cancer detection (don’t miss patients)

F1-score important
When you need balance between precision & recall
Example: Imbalanced datasets
```
