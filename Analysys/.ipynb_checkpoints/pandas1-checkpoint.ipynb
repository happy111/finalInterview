```python
import pandas as pd

data = {
    'Student': ['Alice', 'Bob', 'Charlie', 'David'],
    'Math':   [85, 90, 75, 80],
    'Science': [88, 85, 90, 82],
    'English': [82, 88, 85, 90]
}

df = pd.DataFrame(data)
print(df)

# ---- Child-wise average ----
df['child_average'] = df[['Math', 'Science', 'English']].mean(axis=1)

# ---- Subject-wise average ----
subject_average = df[['Math', 'Science', 'English']].mean()

# ---- OPTIONAL: If you want "percentage" (out of 100) ----
df['percentage'] = df['child_average']   # same because max marks = 100 each

print("\nUpdated DataFrame:")
print(df)

print("\nSubject-wise Average:")
print(subject_average)
```

```python
def calculate_percentage(row):
    return row[['Math', 'Science', 'English']].mean()

df['percentage'] = df.apply(calculate_percentage, axis=1)
```
