```python
import pandas as pd

df = pd.DataFrame({
    'ID': [1, 2, 2, 3, 3],
    'CITY': ['BLR', 'HYD', 'HYD', 'DEL', 'DEL'],
    'SALARY': [50000, 60001, 60000, 55000, 55000]
})

print(df)
```

```python
df = df.drop_duplicates()
```

```python
df
```

```python
🔹 2. Remove duplicates based on one column
df_city = df.drop_duplicates(subset=['CITY'])
print(df_city)
```

```python
| Feature             | merge | join       | concat |
| ------------------- | ----- | ---------- | ------ |
| Similar to SQL JOIN | ✅     | ✅          | ❌      |
| Join on column      | ✅     | ⚠️ Limited | ❌      |
| Join on index       | ⚠️    | ✅          | ❌      |
| Stack DataFrames    | ❌     | ❌          | ✅      |
| Requires key        | ✅     | Usually    | ❌      |

merge() → join on columns

join() → join on index
```

```python
import pandas as pd

df1 = pd.DataFrame({
    'EMP_ID': [1, 2, 3],
    'NAME': ['Amit', 'Neha', 'Rahul']
})

df2 = pd.DataFrame({
    'EMP_ID': [2, 3, 4],
    'SALARY': [60000, 70000, 80000]
})
```

```python
df1
```

```python
df2
```

```python
# Merge
pd.merge(df1, df2, on='EMP_ID', how='right')
```

```python
# 2️⃣ join() → Join using index (or key column)
df1_indexed = df1.set_index('EMP_ID')
df2_indexed = df2.set_index('EMP_ID')
```

```python
df1_indexed.join(df2_indexed)
```

```python
🔹 Row-wise concat (default)
pd.concat([df1, df1])
```

```python
🔹 Column-wise concat
pd.concat([df1, df2], axis=1)
```
