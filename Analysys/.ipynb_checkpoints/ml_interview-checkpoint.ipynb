```python

You have a supervised learning problem but only 60% of the data is labeled. What ML approach do
you use?
```

```python
“When only 60% of the data is labeled, the correct approach is Semi-Supervised Learning.
I first train a model on the labeled data, then use it to predict labels for the unlabeled part.
High-confidence predictions become pseudo-labels, and I retrain using both sets.
This approach improves accuracy and uses the full dataset effectively.”


⭐ Example: Classifying Emails as Spam or Not Spam

You have 10,000 emails, but only 6,000 are labeled:
Spam
Not Spam
The remaining 4,000 emails have no labels.

✔️ Step-by-Step (Semi-Supervised Learning)

Train on Labeled Data
Use the 6,000 labeled emails to train a spam classifier (e.g., Logistic Regression or Random Forest).

Predict Labels for Unlabeled Emails
Apply the trained model to the 4,000 unlabeled emails.

Select High-Confidence Predictions
Suppose the model is 95% sure about 2,500 emails.
We treat these as pseudo-labeled.

Retrain Using Both Sets
Train the final model using:

6,000 labeled emails

2,500 pseudo-labeled emails

Final Model is More Accurate
Because it used almost the entire dataset, not just 60
```
