```python
You have 40% missing data in some columns. How do you handle this?

  1️⃣ Understand the Missingness Type
     Before deciding anything
  2️⃣ Analyze Feature Importance
  3️⃣ Choose the Right Strategy Based on Column Type

        A. Numerical Column (40% NA)
               Mean/Median (if data is normally distributed or skewed)
        B. Categorical Column (40% NA)
            Option 1 — Fill with Mode
            df['col'].fillna(df['col'].mode()[0], inplace=True)
4️⃣ Drop Column Only if Necessary
```

```python
“If a column has 40% missing data, I don’t blindly drop it.
I first analyze why the data is missing (MCAR, MAR, MNAR).
If the feature is important, I try advanced imputation like
KNN or MICE for numerical variables, or mode/‘Unknown’ imputation for categorical variables.
I also create a missingness indicator because missing values often carry useful signals.
I drop the column only if it has low importance or cannot be reliably imputed.
Finally, I compare model performance before and after imputation.
```
